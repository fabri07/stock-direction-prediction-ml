```python
# =============================================
#  NVDA BULL/BEAR — Dos modelos por época + Ensemble + Calibración + Stacking
#  File: src/hybrid_nvda_bullbear_2models.py
# =============================================
import os
import json
from datetime import datetime, UTC

import joblib
import numpy as np
import pandas as pd
import requests
import torch
import matplotlib.pyplot as plt

from tqdm import tqdm
from transformers import AutoTokenizer, AutoModelForSequenceClassification

from sklearn.base import BaseEstimator, TransformerMixin, clone
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import RobustScaler
from sklearn.feature_selection import VarianceThreshold
from sklearn.metrics import (
    f1_score, classification_report, confusion_matrix,
    accuracy_score, precision_recall_curve
)
from sklearn.calibration import CalibratedClassifierCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

# =============================================
#              CONFIG
# =============================================
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"🧠 Device Torch: {device}")

print("🔄 Cargando FinBERT (esto puede tardar unos segundos)...")
tokenizer = AutoTokenizer.from_pretrained("yiyanghkust/finbert-tone", local_files_only=False)
model_sent = AutoModelForSequenceClassification.from_pretrained(
    "yiyanghkust/finbert-tone",
    local_files_only=False
).to(device)
print("✅ FinBERT cargado")

FRED_API_KEY = os.getenv("FRED_API_KEY", "")
FMP_API_KEY = os.getenv("FMP_API_KEY", "")
FINNHUB_API_KEY = os.getenv("FINNHUB_API_KEY", "")
USE_GPU = torch.cuda.is_available()

# =============================================
#  Utils
# =============================================
def _safe_pct_change(s: pd.Series, periods: int) -> pd.Series:
    return s.pct_change(periods)

def _lag_cols(df: pd.DataFrame, cols: list[str], lag: int = 1) -> pd.DataFrame:
    lagged = df[cols].shift(lag)
    lagged.columns = [f"{c}_lag{lag}" for c in cols]
    return lagged

def _best_f1_threshold(y_true: np.ndarray, y_prob: np.ndarray) -> float:
    p, r, t = precision_recall_curve(y_true, y_prob)
    f1 = np.where((p + r) > 0, 2 * p * r / (p + r), 0.0)
    if len(t) == 0:
        return 0.5
    best = int(np.nanargmax(f1[:-1]))
    return float(np.clip(t[best], 0.01, 0.99))

def _weighted_average(p1: np.ndarray, p2: np.ndarray, w1: float, w2: float) -> np.ndarray:
    s = (w1 + w2) if (w1 + w2) > 0 else 1.0
    return (w1 * p1 + w2 * p2) / s
# === Helpers de deduplicación ===
def _unique_cols(df: pd.DataFrame) -> pd.DataFrame:
    """Devuelve el DF con columnas únicas, conservando la primera aparición."""
    if not isinstance(df, pd.DataFrame):
        return df
    return df.loc[:, ~df.columns.duplicated()].copy()

def _unique_list(seq):
    """Devuelve una lista sin duplicados, preservando el orden."""
    seen = set()
    out = []
    for x in seq:
        if x not in seen:
            seen.add(x)
            out.append(x)
    return out

# =============================================
#  Transformadores que preservan nombres/shape
# =============================================
class ColumnAligner(BaseEstimator, TransformerMixin):
    def fit(self, X, y=None):
        self.columns_ = list(X.columns) if isinstance(X, pd.DataFrame) else [f"f_{i}" for i in range(np.shape(X)[1])]
        return self
    def transform(self, X):
        df = X if isinstance(X, pd.DataFrame) else pd.DataFrame(X, columns=self.columns_)
        return df.reindex(columns=self.columns_)

class DataFrameSimpleImputer(SimpleImputer):
    def fit(self, X, y=None):
        self._cols = list(X.columns)
        return super().fit(X, y)
    def transform(self, X):
        arr = super().transform(X)
        return pd.DataFrame(arr, index=X.index, columns=self._cols)

class DataFrameRobustScaler(RobustScaler):
    def fit(self, X, y=None):
        self._cols = list(X.columns)
        return super().fit(X, y)
    def transform(self, X):
        arr = super().transform(X)
        return pd.DataFrame(arr, index=X.index, columns=self._cols)

class OutlierClipper(BaseEstimator, TransformerMixin):
    def __init__(self, lower_q: float = 0.01, upper_q: float = 0.99):
        self.lower_q = float(lower_q); self.upper_q = float(upper_q)
    def fit(self, X, y=None):
        X_df = X if isinstance(X, pd.DataFrame) else pd.DataFrame(X)
        self.columns_ = list(X_df.columns)
        self.lower_ = X_df.quantile(self.lower_q)
        self.upper_ = X_df.quantile(self.upper_q)
        return self
    def transform(self, X):
        X_df = X if isinstance(X, pd.DataFrame) else pd.DataFrame(X, columns=self.columns_)
        return X_df.clip(lower=self.lower_, upper=self.upper_, axis=1)

class CorrelationSelector(BaseEstimator, TransformerMixin):
    def __init__(self, threshold: float = 0.90):
        self.threshold = float(threshold)
    def fit(self, X, y=None):
        X_df = X if isinstance(X, pd.DataFrame) else pd.DataFrame(X)
        corr = X_df.corr().abs().fillna(0.0)
        upper = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))
        to_drop = [c for c in upper.columns if (upper[c] > self.threshold).any()]
        self.keep_ = [c for c in X_df.columns if c not in to_drop]
        return self
    def transform(self, X):
        X_df = X if isinstance(X, pd.DataFrame) else pd.DataFrame(X, columns=self.keep_)
        return X_df[self.keep_]
    def get_feature_names_out(self, input_features=None):
        return np.array(self.keep_)

class AllNaNToConstant(BaseEstimator, TransformerMixin):
    """Evita que el Imputer salte columnas 100% NaN (shape mismatch)."""
    def __init__(self, constant: float = 0.0):
        self.constant = constant
        self.all_nan_cols_: list[str] = []
    def fit(self, X, y=None):
        X_df = X if isinstance(X, pd.DataFrame) else pd.DataFrame(X)
        self.all_nan_cols_ = [c for c in X_df.columns if not X_df[c].notna().any()]
        return self
    def transform(self, X):
        X_df = X if isinstance(X, pd.DataFrame) else pd.DataFrame(X, columns=self.all_nan_cols_)
        if self.all_nan_cols_:
            X_df = X_df.copy()
            X_df[self.all_nan_cols_] = X_df[self.all_nan_cols_].fillna(self.constant)
        return X_df

# =============================================
#  Dataset (anti-leak; target t+1 | X_t lag1)
# =============================================
def build_dataset_all(ticker: str, start_date: str = "2000-01-01"):
    print(f"📡 Descargando datos diarios de {ticker} desde FMP...")
    url_price = (
        f"https://financialmodelingprep.com/api/v3/historical-price-full/"
        f"{ticker}?from={start_date}&to=2025-11-10&apikey={FMP_API_KEY}"
    )
    r = requests.get(url_price).json()
    if "historical" not in r:
        raise ValueError("❌ No se pudieron obtener precios históricos desde FMP.")
    df_prices = pd.DataFrame(r["historical"])
    df_prices["Date"] = pd.to_datetime(df_prices["date"], errors="coerce")
    df_prices = df_prices.sort_values("Date").reset_index(drop=True)
    df_prices = df_prices[df_prices["Date"] >= pd.to_datetime(start_date)]
    print(f"✅ {len(df_prices)} días {df_prices['Date'].min().date()} → {df_prices['Date'].max().date()}")

    print("🏦 Fundamentales FMP...")
    endpoints = {
        "marketcap": f"https://financialmodelingprep.com/stable/historical-market-capitalization?symbol={ticker}&apikey={FMP_API_KEY}",
        "income": f"https://financialmodelingprep.com/stable/income-statement?symbol={ticker}&period=quarter&apikey={FMP_API_KEY}",
        "balance": f"https://financialmodelingprep.com/stable/balance-sheet-statement?symbol={ticker}&period=quarter&apikey={FMP_API_KEY}",
        "cashflow": f"https://financialmodelingprep.com/stable/cash-flow-statement?symbol={ticker}&period=quarter&apikey={FMP_API_KEY}",
        "keymetrics": f"https://financialmodelingprep.com/stable/key-metrics?symbol={ticker}&period=annual&apikey={FMP_API_KEY}",
        "income_growth": f"https://financialmodelingprep.com/stable/income-statement-growth?symbol={ticker}&period=quarter&apikey={FMP_API_KEY}",
        "balance_growth": f"https://financialmodelingprep.com/stable/balance-sheet-statement-growth?symbol={ticker}&period=quarter&apikey={FMP_API_KEY}",
        "cashflow_growth": f"https://financialmodelingprep.com/stable/cash-flow-statement-growth?symbol={ticker}&period=quarter&apikey={FMP_API_KEY}",
        "financial_growth": f"https://financialmodelingprep.com/stable/financial-growth?symbol={ticker}&period=quarter&apikey={FMP_API_KEY}",
    }
    dfs = []
    for name, url in endpoints.items():
        try:
            data = requests.get(url, timeout=15).json()
            df = pd.DataFrame(data)
            if not df.empty:
                date_col = [c for c in df.columns if "date" in c.lower() or "period" in c.lower()]
                if date_col:
                    df["Date"] = pd.to_datetime(df[date_col[0]], errors="coerce")
                df = df.drop_duplicates(subset="Date").set_index("Date").sort_index()
                df.columns = [f"{name}_{c}" for c in df.columns]
                dfs.append(df)
        except Exception as e:
            print(f"⚠️ {name}: {e}")
    fundamentals = (pd.concat(dfs, axis=1) if dfs else pd.DataFrame()).ffill().reset_index().rename(columns={"index":"Date"}) if dfs else pd.DataFrame({"Date": df_prices["Date"]})
    print(f"✅ Fundamentales: {fundamentals.shape[0]} filas, {fundamentals.shape[1]} cols")

    print("📊 S&P500 local (sp500.csv)...")
    sp500_path = "sp500.csv"
    with open(sp500_path, "r", encoding="utf-8-sig") as f:
        first = f.readline(); sep = ";" if first.count(";") > first.count(",") else ","
    sp500 = pd.read_csv(sp500_path, sep=sep, encoding="utf-8-sig")
    sp500.columns = [c.strip().lower() for c in sp500.columns]
    date_col = next((c for c in sp500.columns if "date" in c or "fecha" in c), None)
    close_col = next((c for c in sp500.columns if "close" in c or "cierre" in c), None)
    sp500 = sp500[[date_col, close_col]].rename(columns={date_col:"Date", close_col:"sp500_close"})
    sp500["Date"] = pd.to_datetime(sp500["Date"], errors="coerce", dayfirst=True)
    sp500["sp500_close"] = sp500["sp500_close"].astype(str).str.replace(",", ".").astype(float)
    sp500 = sp500.dropna().sort_values("Date").reset_index(drop=True)

    print("💵 Treasury (rf)...")
    tdata = requests.get(f"https://financialmodelingprep.com/stable/treasury-rates?apikey={FMP_API_KEY}").json()
    df_treasury = pd.DataFrame(tdata)
    df_treasury["Date"] = pd.to_datetime(df_treasury["date"], errors="coerce")
    df_treasury = df_treasury.sort_values("Date")
    ten_cols = [c for c in df_treasury.columns if "10" in c and ("year" in c.lower() or "y" in c.lower())]
    if ten_cols:
        df_treasury["rf"] = pd.to_numeric(df_treasury[ten_cols[0]], errors="coerce") / 100.0
    else:
        numc = df_treasury.select_dtypes(include=[np.number]).columns
        df_treasury["rf"] = (df_treasury[numc].mean(axis=1) / 100.0) if len(numc) else 0.0

    print("🔗 Merge total...")
    merged = pd.merge_asof(df_prices.sort_values("Date"), fundamentals.sort_values("Date"), on="Date", direction="backward")
    merged = pd.merge_asof(merged.sort_values("Date"), sp500.sort_values("Date"), on="Date", direction="backward")
    merged = pd.merge_asof(merged.sort_values("Date"), df_treasury[["Date","rf"]].sort_values("Date"), on="Date", direction="backward")

    print("📈 Indicadores técnicos...")
    merged["r_nvda"] = merged["close"].pct_change()
    merged["r_mkt"]  = merged["sp500_close"].pct_change()
    merged["excess_nvda"] = merged["r_nvda"] - merged["rf"] / 252.0
    merged["excess_mkt"]  = merged["r_mkt"]  - merged["rf"] / 252.0
    cov = merged["excess_nvda"].rolling(30).cov(merged["excess_mkt"])
    var = merged["excess_mkt"].rolling(30).var()
    merged["beta_30d"] = cov / var

    high, low, close = merged["high"], merged["low"], merged["close"]
    plus_dm = np.where((high - high.shift(1)) > (low.shift(1) - low), np.maximum(high - high.shift(1), 0), 0)
    minus_dm = np.where((low.shift(1) - low) > (high - high.shift(1)), np.maximum(low.shift(1) - low, 0), 0)
    tr = pd.concat([(high - low), (high - close.shift(1)).abs(), (low - close.shift(1)).abs()], axis=1).max(axis=1)
    tr14 = tr.rolling(14).sum()
    plus_dm14 = pd.Series(plus_dm).rolling(14).sum()
    minus_dm14 = pd.Series(minus_dm).rolling(14).sum()
    plus_di14 = 100 * (plus_dm14 / tr14)
    minus_di14 = 100 * (minus_dm14 / tr14)
    dx = (abs(plus_di14 - minus_di14) / (plus_di14 + minus_di14)) * 100
    merged["ADX_14"] = dx.rolling(14).mean()

    merged["PVT"] = (merged["close"].pct_change() * merged["volume"]).cumsum()
    merged["TR"] = tr.values
    merged["ATR_14"] = tr.rolling(14).mean().values

        # 📰 Noticias → FinBERT → señal diaria (sin NewsAPI). Orden: Finnhub → GoogleNews RSS → Yahoo RSS → dummy
    import math
    from datetime import timedelta
    from urllib.parse import quote_plus


    def _env(key: str) -> str | None:
        v = os.getenv(key)
        return v.strip() if v else None

    def _chunk_dates(dstart: pd.Timestamp, dend: pd.Timestamp, days: int = 30):
        cur = dstart
        while cur <= dend:
            nxt = min(cur + pd.Timedelta(days=days-1), dend)
            yield cur, nxt
            cur = nxt + pd.Timedelta(days=1)
    def _normalize_rows(rows: list[dict], src: str) -> pd.DataFrame:
        if not rows:
            return pd.DataFrame(columns=["Date","title","summary","url","source","publisher"])
        out = []
        for r in rows:
            if src == "finnhub":
                dt = r.get("datetime")
                if isinstance(dt, (int,float)): dt = pd.to_datetime(dt, unit="s", errors="coerce")
                else: dt = pd.to_datetime(dt, errors="coerce")
                out.append({
                    "Date": dt.normalize() if pd.notna(dt) else pd.NaT,
                    "title": r.get("headline") or r.get("title") or "",
                    "summary": r.get("summary") or "",
                    "url": r.get("url") or "",
                    "source": "finnhub",
                    "publisher": (r.get("source") or "").lower()
                })
            else:
                dt = pd.to_datetime(r.get("published"), errors="coerce")
                out.append({
                    "Date": dt.normalize() if pd.notna(dt) else pd.NaT,
                    "title": r.get("title") or "",
                    "summary": r.get("summary") or "",
                    "url": r.get("url") or "",
                    "source": src,
                    "publisher": (r.get("publisher") or r.get("source") or "").lower()
                })
        df = pd.DataFrame(out).dropna(subset=["Date"]).reset_index(drop=True)
        if not df.empty:
            if "url" in df.columns:
                df = df.drop_duplicates(subset=["url"]).reset_index(drop=True)
            df = df.drop_duplicates(subset=["title"]).reset_index(drop=True)
        return df

    def _finnhub_fetch(symbol: str, start: pd.Timestamp, end: pd.Timestamp, api_key: str) -> pd.DataFrame:
        base = "https://finnhub.io/api/v1/company-news"
        all_rows = []
        for s,e in _chunk_dates(start, end, days=30):
            try:
                resp = requests.get(base, params={"symbol":symbol,"from":s.date().isoformat(),"to":e.date().isoformat(),"token":api_key}, timeout=15)
                if resp.status_code == 200:
                    rows = resp.json()
                    if isinstance(rows, list) and rows:
                        for x in rows:
                            # normalizar campo publisher (cuando exista)
                            if "source" in x and isinstance(x["source"], str):
                                x["source"] = x["source"].strip()
                        all_rows.extend(rows)
            except Exception:
                pass
        return _normalize_rows(all_rows, "finnhub")

    # —— RSS helpers (sin API keys) ——
    def _rss_parse(url: str) -> list[dict]:
        try:
            import feedparser
            d = feedparser.parse(url)
            rows = []
            for e in d.entries:
                rows.append({
                    "title": getattr(e, "title", "") or "",
                    "summary": getattr(e, "summary", "") or getattr(e, "description", "") or "",
                    "url": getattr(e, "link", "") or "",
                    "published": getattr(e, "published", "") or getattr(e, "updated", "") or "",
                    "publisher": getattr(e, "source", {}).get("title", "").lower() if hasattr(e, "source") else (getattr(e, "publisher", "") or getattr(e, "author", "")).lower()
                })
            return rows
        except Exception:
            # fallback mínimo
            try:
                import re
                r = requests.get(url, timeout=15); r.raise_for_status()
                text = r.text
                items = re.findall(r"<item>(.*?)</item>", text, flags=re.DOTALL|re.IGNORECASE)
                rows = []
                for it in items:
                    def _tag(t):
                        m = re.search(fr"<{t}>(.*?)</{t}>", it, flags=re.DOTALL|re.IGNORECASE)
                        return re.sub("<.*?>", "", m.group(1)).strip() if m else ""
                    rows.append({
                        "title": _tag("title"),
                        "summary": _tag("description"),
                        "url": _tag("link"),
                        "published": _tag("pubDate"),
                        "publisher": ""
                    })
                return rows
            except Exception:
                return []

    def _gnews_fetch_multi(queries: list[str], days: int = 365) -> pd.DataFrame:
        # Combina varias queries; filtra duplicados
        frames = []
        for q in queries:
            q2 = quote_plus(q + f" when:{days}d")
            url = f"https://news.google.com/rss/search?q={q2}&hl=en-US&gl=US&ceid=US:en"
            rows = _rss_parse(url)
            for r in rows:
                r["published"] = str(r.get("published",""))
            frames.append(_normalize_rows(rows, "gnews"))
        if frames:
            df = pd.concat(frames, axis=0, ignore_index=True)
            if "url" in df.columns: df = df.drop_duplicates(subset=["url"]).reset_index(drop=True)
            df = df.drop_duplicates(subset=["title"]).reset_index(drop=True)
            return df
        return pd.DataFrame(columns=["Date","title","summary","url","source","publisher"])

    def _yahoo_rss_nvda() -> pd.DataFrame:
        url = "https://feeds.finance.yahoo.com/rss/2.0/headline?s=NVDA&region=US&lang=en-US"
        rows = _rss_parse(url)
        for r in rows:
            r["published"] = str(r.get("published",""))
        return _normalize_rows(rows, "yahoo")

    # Ponderación por fuente (simple)
    SOURCE_WEIGHTS = {
        "reuters": 1.5, "bloomberg": 1.5, "wall street journal": 1.4, "wsj": 1.4,
        "financial times": 1.3, "ft": 1.3, "the verge": 1.1, "the information": 1.2,
        "yahoo": 0.9, "seeking alpha": 0.9, "investing.com": 0.9, "motley fool": 0.9,
    }
    def _weight_for_publisher(p: str) -> float:
        if not p: return 1.0
        p = p.lower()
        for k,v in SOURCE_WEIGHTS.items():
            if k in p: return float(v)
        return 1.0
    def _score_finbert_daily_weighted(df_news: pd.DataFrame) -> pd.DataFrame:
        if df_news.empty:
            return pd.DataFrame(columns=["Date","neg","neu","pos","news_count","neg_ewm3","neu_ewm3","pos_ewm3","news_pos_interact"])
        texts = (df_news["title"].fillna("") + ". " + df_news["summary"].fillna("")).tolist()
        weights = df_news["publisher"].apply(_weight_for_publisher).to_numpy()
        batch = 16
        probs_all, w_all = [], []
        for i in range(0, len(texts), batch):
            chunk = texts[i:i+batch]
            w_chunk = weights[i:i+batch]
            try:
                inputs = tokenizer(chunk, return_tensors="pt", truncation=True, padding=True, max_length=256).to(device)
                with torch.no_grad():
                    outputs = model_sent(**inputs)
                    probs = torch.nn.functional.softmax(outputs.logits, dim=-1).detach().cpu().numpy()
                probs_all.append(probs); w_all.append(w_chunk)
            except Exception:
                continue
        if not probs_all: 
            return pd.DataFrame(columns=["Date","neg","neu","pos","news_count","neg_ewm3","neu_ewm3","pos_ewm3","news_pos_interact"])
        arr = np.vstack(probs_all); wv = np.concatenate(w_all)
        df_scored = df_news.iloc[:arr.shape[0]].copy()
        df_scored[["neg","neu","pos"]] = arr
        df_scored["w"] = wv

        # agregación diaria ponderada
        def _wmean(g, col):
            a = g[col].to_numpy(); w = g["w"].to_numpy()
            s = np.sum(w); 
            return float(np.sum(a*w)/s) if s > 0 else float(np.mean(a))
        grouped = []
        for d, g in df_scored.groupby("Date"):
            grouped.append({
                "Date": d,
                "neg": _wmean(g, "neg"),
                "neu": _wmean(g, "neu"),
                "pos": _wmean(g, "pos"),
                "news_count": float(g.shape[0])
            })
        daily = pd.DataFrame(grouped).sort_values("Date").reset_index(drop=True)
        for c in ["neg","neu","pos"]:
            daily[f"{c}_ewm3"] = daily[c].ewm(span=3, adjust=False).mean()
        # interacción volumen × tono
        daily["news_pos_interact"] = daily["news_count"] * daily["pos"]
        return daily

    # Orquestación con fallback
    date_min, date_max = merged["Date"].min(), merged["Date"].max()
    recent_days = 365  # ↑ de 90 → 365
    df_news_all = pd.DataFrame(columns=["Date","title","summary","url","source","publisher"])
    used_source = "dummy"

    # 1) Finnhub si hay clave
    finnhub_key = _env("FINNHUB_API_KEY")
    if finnhub_key:
        print("📰 FinBERT con noticias reales (Finnhub)…")
        df_fh = _finnhub_fetch(ticker, date_min, date_max, finnhub_key)
        if not df_fh.empty:
            df_news_all = df_fh; used_source = "finnhub"

    # 2) Google News RSS multi-query
    if df_news_all.empty:
        print("📰 FinBERT con noticias reales (Google News RSS, 365d, multi-query)…")
        queries = ["NVDA OR Nvidia", "NVIDIA Corporation", "Nvidia earnings"]
        df_gn = _gnews_fetch_multi(queries, days=recent_days)
        if not df_gn.empty:
            df_gn = df_gn[(df_gn["Date"] >= date_min) & (df_gn["Date"] <= date_max)]
        if not df_gn.empty:
            df_news_all = df_gn; used_source = "gnews"

    # 3) Yahoo RSS
    if df_news_all.empty:
        print("📰 FinBERT con noticias (Yahoo Finance RSS)…")
        df_yh = _yahoo_rss_nvda()
        if not df_yh.empty:
            df_yh = df_yh[(df_yh["Date"] >= date_min) & (df_yh["Date"] <= date_max)]
        if not df_yh.empty:
            df_news_all = df_yh; used_source = "yahoo"
    # 4) Dummy
    if df_news_all.empty:
        print("🧠 FinBERT dummy (fallback)…")
        samples = [
            f"{ticker} strong demand for AI chips",
            f"{ticker} faces competition in GPU and data center markets",
            f"{ticker} market outlook remains positive",
            f"{ticker} growth prospects show investor optimism",
        ]
        dates_dummy = pd.date_range(start=start_date, periods=len(samples), freq="D")
        df_news_all = pd.DataFrame({
            "Date": dates_dummy, "title": samples, "summary": "", "url": "", "source": "dummy", "publisher": "dummy"
        })
        used_source = "dummy"

    daily_sent = _score_finbert_daily_weighted(df_news_all)
    if daily_sent.empty:
        daily_sent = pd.DataFrame({
            "Date": merged["Date"].copy(),
            "neg": 0.0, "neu": 1.0, "pos": 0.0, "news_count": 0.0,
            "neg_ewm3": 0.0, "neu_ewm3": 1.0, "pos_ewm3": 0.0,
            "news_pos_interact": 0.0
        })

    merged = pd.merge_asof(
        merged.sort_values("Date"),
        daily_sent.sort_values("Date"),
        on="Date", direction="backward"
    )
    print(f"✅ Sentimiento integrado desde: {used_source} — días con noticias: {int((merged['news_count']>0).sum())}")


    print("🏛️ FRED...")
    fred_ids = ["CPIAUCSL","UNRATE","GDPC1","PCE","INDPRO","PAYEMS","GNPCA","M2SL","HOUST"]
    def fred(series_id):
        try:
            url = f"https://api.stlouisfed.org/fred/series/observations?series_id={series_id}&api_key={FRED_API_KEY}&file_type=json"
            r = requests.get(url, timeout=20).json()
            df = pd.DataFrame(r["observations"])[["date", "value"]]
            df["date"] = pd.to_datetime(df["date"], errors="coerce")
            df["value"] = pd.to_numeric(df["value"], errors="coerce")
            return df.dropna().set_index("date").rename(columns={"value": series_id})
        except:
            return pd.DataFrame()
    fred_valid = [fred(fid) for fid in fred_ids]
    fred_valid = [d for d in fred_valid if not d.empty]
    if fred_valid:
        fred_merged = pd.concat(fred_valid, axis=1).ffill().reset_index().rename(columns={"date":"Date"})
        merged = pd.merge_asof(merged.sort_values("Date"), fred_merged.sort_values("Date"), on="Date", direction="backward")

    print("📈 Features + target (lag1, anti-leak)...")
    merged = merged.sort_values("Date").reset_index(drop=True)
    merged["ema10"] = merged["close"].ewm(span=10).mean()
    merged["ema20"] = merged["close"].ewm(span=20).mean()
    merged["ema50"] = merged["close"].ewm(span=50).mean()
    merged["ema100"] = merged["close"].ewm(span=100).mean()
    merged["ema10_above_ema50"] = (merged["ema10"] > merged["ema50"]).astype(int)
    merged["ema20_above_ema50"] = (merged["ema20"] > merged["ema50"]).astype(int)
    merged["slope_20"] = merged["close"].diff(20)
    merged["slope_50"] = merged["close"].diff(50)
    merged["roc_5"]  = _safe_pct_change(merged["close"], 5)
    merged["roc_10"] = _safe_pct_change(merged["close"], 10)
    merged["roc_20"] = _safe_pct_change(merged["close"], 20)
    merged["roc_50"] = _safe_pct_change(merged["close"], 50)
    ret = merged["close"].pct_change()
    merged["vol_10"] = ret.rolling(10).std()
    merged["vol_20"] = ret.rolling(20).std()
    merged["vol_50"] = ret.rolling(50).std()
    merged["ewma_vol"] = ret.ewm(span=20).std()
    merged["sp_slope_20"] = merged["sp500_close"].diff(20)

    merged["bull_t"] = ((merged["slope_20"] > 0) & (merged["close"] > merged["ema50"])).astype(int)
    merged["target"] = merged["bull_t"].shift(-1)
    
    # 1) Señal trimestral (Q1..Q4) + codificación cíclica
    merged["quarter"] = merged["Date"].dt.quarter.astype("Int64")
    q_dummies = pd.get_dummies(merged["quarter"], prefix="q", dtype=int)  # q_1..q_4
    merged = pd.concat([merged, q_dummies], axis=1)

    # codificación cíclica (estacionalidad suave)
    # Por 'qué': capta continuidad Q4→Q1 sin saltos artificiales.
    q = merged["quarter"].astype(float)
    merged["q_sin"] = np.sin(2 * np.pi * (q - 1) / 4)
    merged["q_cos"] = np.cos(2 * np.pi * (q - 1) / 4)

    # 2) >>> ACTUALIZA el bloque que define 'raw_cols' para incluir quarter y dummies <<<
    #    añade al final de tu lista:
    #        "quarter", "q_1", "q_2", "q_3", "q_4", "q_sin", "q_cos"
    #    ejemplo:
    raw_cols = [
        "open","high","low","close","volume","ema10","ema20","ema50","ema100",
        "ema10_above_ema50","ema20_above_ema50","slope_20","slope_50",
        "roc_5","roc_10","roc_20","roc_50","vol_10","vol_20","vol_50","ewma_vol",
        "r_nvda","r_mkt","excess_nvda","excess_mkt","beta_30d","ADX_14","PVT","TR","ATR_14",
        "sp500_close","sp_slope_20","rf","neg","neu","pos",
        "quarter","q_1","q_2","q_3","q_4","q_sin","q_cos",
        # Noticias (sin repetir "pos")
        "news_count","pos_ewm3","neg_ewm3","news_pos_interact"
    ]
    raw_cols = [c for c in raw_cols if c in merged.columns]

    # deja solo las que existen y elimina duplicados de nombres
    raw_cols = _unique_list([c for c in raw_cols if c in merged.columns])

    # 3) Lags anti-leak como siempre
    feats_lag1 = _lag_cols(merged, raw_cols, lag=1)

    # Desfase flexible (lag=2) para pos_ewm3
    pos_ewm3_lag2 = merged[["pos_ewm3"]].shift(2) if "pos_ewm3" in merged.columns else pd.DataFrame(index=merged.index)
    if not pos_ewm3_lag2.empty:
        pos_ewm3_lag2.columns = ["pos_ewm3_lag2"]

    # Lags para el resto (macro, etc.)
    macro_cols = [c for c in merged.columns if c not in (["Date","target","bull_t"] + raw_cols)]
    feats_macro_lag1 = _lag_cols(merged, macro_cols, lag=1) if macro_cols else pd.DataFrame(index=merged.index)

    # Ensamble de todas las features
    X_full = pd.concat([feats_lag1, feats_macro_lag1, pos_ewm3_lag2], axis=1)
    X_full = _unique_cols(X_full)  # ← evita duplicados tras concat

    df = pd.concat([merged[["Date","target"]], X_full], axis=1)
    df = df.dropna(subset=["target"]).reset_index(drop=True)

    # Mantener solo variables con _lag1 y pos_ewm3_lag2 (si existe)
    non_lag = [c for c in df.columns if c not in ["Date","target"] and not c.endswith("_lag1") and c != "pos_ewm3_lag2"]
    df = df.drop(columns=non_lag) if non_lag else df

    # 🔒 deduplicación final del dataset
    df = _unique_cols(df)

    print(f"✅ Dataset final: {df.shape[0]} filas, {df.shape[1]-2} features (lag1).")
    
    # 4) (Opcional) Mini-EDA: impacto por trimestre (solo para presentación; no afecta al modelo)
    #    Puedes colocarlo al final de build_dataset_all o en tu script de EDA.
    try:
        tmp = merged.copy()
        tmp["target_t"] = merged["bull_t"].shift(-1)  # misma definición del target (t+1)
        eda_q = (
            tmp.groupby("quarter", dropna=True)["target_t"]
            .mean().rename("bull_rate_next_day")
            .to_frame().reset_index()
            .sort_values("quarter")
        )
        print("\n📊 Bull-rate (t+1) por trimestre:")
        print(eda_q.to_string(index=False))
    except Exception as _:
        pass
    return df

# =============================================
#   Purged + Embargo TimeSeries Split
# =============================================
class PurgedTimeSeriesSplit:
    def __init__(self, n_splits=5, purge=5, embargo=5):
        self.n_splits=n_splits; self.purge=purge; self.embargo=embargo
    def split(self, X: pd.DataFrame):
        n = len(X)
        fold = n // (self.n_splits + 1)
        for i in range(self.n_splits):
            test_start = (i+1)*fold
            train_end = max(0, test_start - self.purge)
            test_end = min(n, test_start + fold)
            yield np.arange(0, train_end), np.arange(test_start, test_end)
    def get_n_splits(self, X=None, y=None, groups=None): return self.n_splits

# =============================================
#  Modelos y pipeline
# =============================================
def build_models_dict():
    xgb_params = dict(
        n_estimators=500, max_depth=6, learning_rate=0.03,
        subsample=0.8, colsample_bytree=0.8, reg_alpha=0.5, reg_lambda=3.0,
        objective="binary:logistic", eval_metric="logloss",
        random_state=42, n_jobs=-1, scale_pos_weight=0.9, tree_method="hist",
    )
    lgbm_params = dict(
        boosting_type="gbdt", num_leaves=48, max_depth=-1, learning_rate=0.03,
        n_estimators=600, subsample=0.8, colsample_bytree=0.8,
        reg_alpha=0.2, reg_lambda=0.4, min_child_samples=25, scale_pos_weight=0.9,
        random_state=42, n_jobs=-1, verbosity=-1, device_type="gpu" if USE_GPU else "cpu",
    )
    return {
        "RandomForest": RandomForestClassifier(
            n_estimators=300, max_depth=12, min_samples_split=10, min_samples_leaf=20,
            max_features="sqrt", bootstrap=True, criterion="gini",
            random_state=42, n_jobs=-1, class_weight="balanced_subsample",
        ),
        "XGBoost": XGBClassifier(**xgb_params),
        "Logistic": LogisticRegression(
            penalty="l2", C=0.5, class_weight="balanced", solver="lbfgs", max_iter=1000, random_state=42,
        ),
        "LightGBM": LGBMClassifier(**lgbm_params),
    }

def make_pipeline(model):
    return Pipeline(steps=[
        ("align", ColumnAligner()),
        ("allnan", AllNaNToConstant(constant=0.0)),
        ("imputer", DataFrameSimpleImputer(strategy="median")),
        ("winsor", OutlierClipper(0.01, 0.99)),
        ("scaler", DataFrameRobustScaler(with_centering=True, with_scaling=True)),
        ("selector", CorrelationSelector(threshold=0.90)),
        ("varth", VarianceThreshold(1e-12)),
        ("model", model),
    ])

# =============================================
#  Entrenamiento por periodo (con calibración isotónica + umbral por época)
# =============================================
def train_period_model(df_all: pd.DataFrame, start: str, end: str, label: str):
    print(f"\n🗓️ Entrenando periodo {label}: {start} → {end}")
    msk = (df_all["Date"] >= pd.to_datetime(start)) & (df_all["Date"] <= pd.to_datetime(end))
    df = df_all.loc[msk].reset_index(drop=True)
    assert len(df) > 0, f"Periodo {label} vacío"

    feats_all = [c for c in df.columns if c not in ["Date","target"]]
    feats_num = [c for c in feats_all if pd.api.types.is_numeric_dtype(df[c])]
    low_var = [c for c in feats_num if df[c].nunique() <= 1]
    high_na = [c for c in feats_num if df[c].isna().mean() > 0.5]
    base_feats = [c for c in feats_num if c not in low_var + high_na]

    X = df[base_feats].copy()
    y = df["target"].astype(int).values

    models = build_models_dict()
    tscv = PurgedTimeSeriesSplit(n_splits=5, purge=5, embargo=5)

    best_name, best_cv = None, -np.inf
    for name, base_model in tqdm(models.items(), desc=f"🔁 Modelos {label}"):
        f1s=[]
        for tr_idx, te_idx in tscv.split(X):
            X_tr, X_te = X.iloc[tr_idx], X.iloc[te_idx]
            y_tr, y_te = y[tr_idx], y[te_idx]
            pipe = make_pipeline(clone(base_model))
            pipe.fit(X_tr, y_tr)
            pred = pipe.predict(X_te)
            f1s.append(f1_score(y_te, pred))
        avg = float(np.mean(f1s))
        print(f"   → {name:12s} F1-CV: {avg:.4f}")
        if avg > best_cv: best_cv, best_name = avg, name

    print(f"🏆 Mejor {label}: {best_name} (F1-CV={best_cv:.4f})")

    # Split interno para calibración + umbral
    split_inner = int(len(X) * 0.9)
    X_tr, y_tr = X.iloc[:split_inner], y[:split_inner]
    X_val, y_val = X.iloc[split_inner:], y[split_inner:]

    base_pipe = make_pipeline(models[best_name])
    base_pipe.fit(X_tr, y_tr)

    # Calibración isotónica sobre el pipeline ya entrenado
    calib = CalibratedClassifierCV(base_pipe, method="isotonic", cv="prefit")
    calib.fit(X_val, y_val)  # aprende mapeo de probas → calibradas

    val_prob_cal = calib.predict_proba(X_val)[:, 1]
    thr = _best_f1_threshold(y_val, val_prob_cal)

    # Guardar artefacto del periodo (NO reentrenar base; mantener consistencia calibrada)
    artifact = {
        "calibrator": calib,
        "features": _unique_list(base_feats),  # ← lista de features sin duplicados,
        "threshold": float(thr),
        "cv_f1": float(best_cv),
        "period": [start, end],
        "model_name": best_name,
    }
    joblib.dump(artifact, f"NVDA_{label}_model.joblib")
    print(f"💾 Guardado NVDA_{label}_model.joblib (thr={thr:.3f})")

    return {
        "label": label, "artifact": artifact,
        "cv_best": best_cv, "features": base_feats, "dates": (start, end)
    }

# =============================================
#  Helpers de predicción calibrada
# =============================================
def _predict_proba_calibrated(df_features: pd.DataFrame, artifact: dict) -> np.ndarray:
    feats = _unique_list(list(artifact["features"]))         # lista única
    Xf = _unique_cols(df_features)                           # DF con columnas únicas
    # Asegura que solo pides columnas que existen
    feats_final = [c for c in feats if c in Xf.columns]
    # Si faltara alguna (por nuevos merges/ediciones), rellena con NaN
    X = Xf.reindex(columns=feats_final, fill_value=np.nan)
    return artifact["calibrator"].predict_proba(X)[:, 1]


# =============================================
#  Ensemble + Stacking + Reportes
# =============================================
def train_two_models_and_ensemble(ticker="NVDA", start_date="2000-01-01"):
    df_all = build_dataset_all(ticker, start_date)

    # === Entrena periodos con calibración isotónica + umbral por época
    p1 = train_period_model(df_all, "2000-01-01", "2019-12-31", "2000_2019")
    p2 = train_period_model(df_all, "2020-01-01", "2025-11-10", "2020_2025")

    art1, art2 = p1["artifact"], p2["artifact"]
    w1, w2 = p1["cv_best"], p2["cv_best"]

    # === Probabilidades calibradas en TODO el historial
    X_all = df_all[[c for c in df_all.columns if c not in ["Date","target"]]].copy()
    X_all = _unique_cols(X_all)  # ← columnas únicas

    proba1 = _predict_proba_calibrated(X_all, art1)
    proba2 = _predict_proba_calibrated(X_all, art2)
    proba_w = _weighted_average(proba1, proba2, w1, w2)
    
    # === Meta features: agregamos news_count_lag1 (si no existe, 0)
    news_count_lag1 = (X_all["news_count_lag1"].to_numpy() if "news_count_lag1" in X_all.columns else np.zeros(len(X_all)))
    X_meta_all = np.c_[proba1, proba2, news_count_lag1]

    # === Voto por periodo (umbral por época)
    dates = pd.to_datetime(df_all["Date"])
    start1, end1 = pd.to_datetime(art1["period"][0]), pd.to_datetime(art1["period"][1])
    start2, end2 = pd.to_datetime(art2["period"][0]), pd.to_datetime(art2["period"][1])
    thr1, thr2 = float(art1["threshold"]), float(art2["threshold"])

    pred_period = np.zeros_like(proba_w, dtype=int)
    m1 = (dates >= start1) & (dates <= end1)
    m2 = (dates >= start2) & (dates <= end2)
    pred_period[m1] = (proba1[m1] >= thr1).astype(int)
    pred_period[m2] = (proba2[m2] >= thr2).astype(int)
    # fuera de rango: fallback al ensemble ponderado con umbral medio simple
    thr_mid = (thr1 + thr2) / 2.0
    rest = ~(m1 | m2)
    pred_period[rest] = (proba_w[rest] >= thr_mid).astype(int)

    # === Ensemble ponderado (probas)
    thr_ens = (w1 * thr1 + w2 * thr2) / (w1 + w2 if (w1 + w2) > 0 else 1.0)
    pred_weighted = (proba_w >= thr_ens).astype(int)

    # === Stacking (meta-Logistic) sobre [proba1, proba2] con CV rolling
    X_meta_all = np.c_[proba1, proba2]
    y_all = df_all["target"].astype(int).values

    def _rolling_splits(n: int, n_splits: int = 5, purge: int = 5):
        fold = n // (n_splits + 1)
        for i in range(n_splits):
            test_start = (i + 1) * fold
            train_end = max(0, test_start - purge)
            test_end = min(n, test_start + fold)
            yield np.arange(0, train_end), np.arange(test_start, test_end)

    f1s_meta = []
    for tr_idx, te_idx in _rolling_splits(len(df_all), n_splits=5, purge=5):
        X_tr, X_te = X_meta_all[tr_idx], X_meta_all[te_idx]
        y_tr, y_te = y_all[tr_idx], y_all[te_idx]
        meta = LogisticRegression(penalty="l2", C=1.0, solver="lbfgs", max_iter=1000)
        meta.fit(X_tr, y_tr)
        pred_te = meta.predict(X_te)
        f1s_meta.append(f1_score(y_te, pred_te))
    print(f"🧱 Meta-Logistic F1-CV (rolling): {np.mean(f1s_meta):.4f}")

    split_inner = int(0.9 * len(df_all))
    X_tr_m, y_tr_m = X_meta_all[:split_inner], y_all[:split_inner]
    X_val_m, y_val_m = X_meta_all[split_inner:], y_all[split_inner:]
    meta = LogisticRegression(penalty="l2", C=1.0, solver="lbfgs", max_iter=1000)
    meta.fit(X_tr_m, y_tr_m)
    val_prob_m = meta.predict_proba(X_val_m)[:, 1]
    thr_meta = _best_f1_threshold(y_val_m, val_prob_m)
    joblib.dump({"meta_clf": meta, "threshold": float(thr_meta)}, "NVDA_stack.joblib")
    print(f"💾 Guardado NVDA_stack.joblib (thr_meta={thr_meta:.3f})")

    proba_stack = meta.predict_proba(X_meta_all)[:, 1]
    pred_stack = (proba_stack >= thr_meta).astype(int)

    # === Evaluación en holdout (último 20%)
    split_ho = int(len(df_all) * 0.8)
    y_ho = y_all[split_ho:]
    d_ho = df_all["Date"].iloc[split_ho:]

    def _report(tag, y_true, pred):
        f1 = f1_score(y_true, pred); acc = accuracy_score(y_true, pred)
        print(f"🎯 {tag} HOLDOUT → F1={f1:.4f}, ACC={acc:.4f}")
        print(confusion_matrix(y_true, pred))
        print(classification_report(y_true, pred, digits=4))
        return f1, acc

    f1_p, acc_p = _report("Period-Vote", y_ho, pred_period[split_ho:])
    f1_w, acc_w = _report("Weighted",    y_ho, pred_weighted[split_ho:])
    f1_s, acc_s = _report("Stacking",    y_ho, pred_stack[split_ho:])

    # === Guardar ensemble config (sin umbral global fijo; guardamos ambos)
    ensemble_art = {
        "model_p1": "NVDA_2000_2019_model.joblib",
        "model_p2": "NVDA_2020_2025_model.joblib",
        "weights": {"p1": float(w1), "p2": float(w2)},
        "thresholds": {"p1": float(thr1), "p2": float(thr2), "weighted": float(thr_ens)}
    }
    with open("NVDA_ensemble.json", "w", encoding="utf-8") as f:
        json.dump(ensemble_art, f, indent=2)
    print("💾 Ensemble guardado en: NVDA_ensemble.json")

    # ======= Gráficos (holdout) =======
    plt.figure(figsize=(12,5))
    plt.plot(d_ho, proba_stack[split_ho:], label="Stack proba")
    plt.title(f"{ticker} — Stacking Prob. Bull (holdout)")
    plt.legend(); plt.tight_layout(); plt.show()

    plt.figure(figsize=(12,5))
    plt.plot(d_ho, proba_w[split_ho:], label="Weighted proba")
    plt.axhline(thr_ens, linestyle="--", label=f"Umbral weighted {thr_ens:.2f}")
    plt.title(f"{ticker} — Weighted Prob. Bull (holdout)")
    plt.legend(); plt.tight_layout(); plt.show()

    # Meta resumen
    with open("NVDA_results_summary.json","w",encoding="utf-8") as f:
        json.dump({
            "period_vote": {"f1": float(f1_p), "acc": float(acc_p)},
            "weighted":    {"f1": float(f1_w), "acc": float(acc_w)},
            "stacking":    {"f1": float(f1_s), "acc": float(acc_s)}
        }, f, indent=2)
    print("💾 Resultados guardados en: NVDA_results_summary.json")

    return {
        "period_vote": (f1_p, acc_p),
        "weighted": (f1_w, acc_w),
        "stacking": (f1_s, acc_s),
        "thr_ens": float(thr_ens),
        "thr_meta": float(thr_meta)
    }

# Ejecutar
if __name__ == "__main__":
    summary = train_two_models_and_ensemble(ticker="NVDA", start_date="2000-01-01")
```

```python
# forward_eval_nvda.py
import os, json, joblib, numpy as np, pandas as pd, requests, math
from datetime import datetime, timedelta
from urllib.parse import quote_plus
from pathlib import Path
from sklearn.metrics import f1_score, accuracy_score, confusion_matrix, classification_report

import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification

# =======================
# CONFIG
# =======================
TICKER = "NVDA"
END_TRAIN = pd.to_datetime("2025-11-10")   # fin del set de entrenamiento
END_FWD   = pd.to_datetime("2025-11-27")   # hasta dónde evaluar
LOOKBACK_DAYS = 220                        # para EMAs/ROCs
SP500_CSV = "sp500_test.csv"               # tu CSV local con días nuevos
FMP_API_KEY   = os.getenv("FMP_API_KEY",   "")
FRED_API_KEY  = os.getenv("FRED_API_KEY",  "")

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"🧠 Torch device: {device}")

# FinBERT para noticias (sin NewsAPI, usamos RSS)
print("🔄 Cargando FinBERT...")
tokenizer = AutoTokenizer.from_pretrained("yiyanghkust/finbert-tone", local_files_only=False)
model_sent = AutoModelForSequenceClassification.from_pretrained(
    "yiyanghkust/finbert-tone", local_files_only=False
).to(device)
print("✅ FinBERT cargado")

# =======================
# Cargar artefactos
# =======================
art_p1 = joblib.load("NVDA_2000_2019_model.joblib")  # {'calibrator','features','threshold','period',...}
art_p2 = joblib.load("NVDA_2020_2025_model.joblib")
stack_art = joblib.load("NVDA_stack.joblib")         # {'meta_clf','threshold'}
with open("NVDA_ensemble.json","r",encoding="utf-8") as f:
    ens_cfg = json.load(f)

w1, w2 = ens_cfg["weights"]["p1"], ens_cfg["weights"]["p2"]
thr1, thr2 = ens_cfg["thresholds"]["p1"], ens_cfg["thresholds"]["p2"]
thr_weighted = ens_cfg["thresholds"]["weighted"]
thr_meta = stack_art["threshold"]
meta_clf = stack_art["meta_clf"]

# =======================
# Helpers
# =======================
def _safe_pct_change(s: pd.Series, periods: int) -> pd.Series:
    return s.pct_change(periods)

def _lag_cols(df: pd.DataFrame, cols, lag=1):
    lagged = df[cols].shift(lag)
    lagged.columns = [f"{c}_lag{lag}" for c in cols]
    return lagged

def _weighted_average(p1, p2, w1, w2):
    s = (w1+w2) if (w1+w2)>0 else 1.0
    return (w1*p1 + w2*p2)/s

def _predict_proba_calibrated(df_features: pd.DataFrame, artifact: dict) -> np.ndarray:
    feats = pd.Index(artifact["features"]).unique()
    X = df_features.loc[:, ~df_features.columns.duplicated()]
    X = X.reindex(columns=feats, fill_value=np.nan)
    return artifact["calibrator"].predict_proba(X)[:, 1]

# ---- Datos base: precios NVDA (FMP), S&P500 (CSV), rf (FMP) ----
def fetch_prices_fmp(ticker, start_date, end_date, api_key):
    url = (f"https://financialmodelingprep.com/api/v3/historical-price-full/"
           f"{ticker}?from={start_date.date()}&to={end_date.date()}&apikey={api_key}")
    r = requests.get(url, timeout=20).json()
    if "historical" not in r:
        raise RuntimeError("No se pudieron obtener precios (FMP).")
    dfp = pd.DataFrame(r["historical"])
    dfp["Date"] = pd.to_datetime(dfp["date"], errors="coerce")
    dfp = dfp.sort_values("Date").reset_index(drop=True)
    keep = ["Date","open","high","low","close","volume"]
    if "close" not in dfp.columns and "adjClose" in dfp.columns:
        dfp["close"] = pd.to_numeric(dfp["adjClose"], errors="coerce")
    for k in keep:
        if k not in dfp.columns and k!="Date": dfp[k] = np.nan
    return dfp[keep]

def load_sp500_csv(path):
    with open(path, "r", encoding="utf-8-sig") as f:
        first = f.readline()
        sep = ";" if first.count(";") > first.count(",") else ","
    sp = pd.read_csv(path, sep=sep, encoding="utf-8-sig")
    sp.columns = [c.strip().lower() for c in sp.columns]
    date_col = next((c for c in sp.columns if "date" in c or "fecha" in c), None)
    close_col = next((c for c in sp.columns if "close" in c or "cierre" in c), None)
    if date_col is None or close_col is None:
        raise RuntimeError("sp500 CSV debe tener columnas de fecha y close/cierre.")
    sp = sp[[date_col, close_col]].rename(columns={date_col:"Date", close_col:"sp500_close"})
    sp["Date"] = pd.to_datetime(sp["Date"], errors="coerce", dayfirst=True)
    sp["sp500_close"] = pd.to_numeric(sp["sp500_close"].astype(str).str.replace(",", "."), errors="coerce")
    sp = sp.dropna().sort_values("Date").reset_index(drop=True)
    return sp

def fetch_treasury_10y(api_key):
    url = f"https://financialmodelingprep.com/stable/treasury-rates?apikey={api_key}"
    t = requests.get(url, timeout=20).json()
    dt = pd.DataFrame(t)
    dt["Date"] = pd.to_datetime(dt["date"], errors="coerce")
    dt = dt.sort_values("Date")
    ten_cols = [c for c in dt.columns if "10" in c and ("year" in c.lower() or "y" in c.lower())]
    if ten_cols:
        dt["rf"] = pd.to_numeric(dt[ten_cols[0]], errors="coerce")/100.0
    else:
        numc = dt.select_dtypes(include=[np.number]).columns
        dt["rf"] = (dt[numc].mean(axis=1) / 100.0) if len(numc) else 0.0
    return dt[["Date","rf"]]

# ---- FRED series (último valor conocido a END_TRAIN) ----
FRED_IDS = ["CPIAUCSL","UNRATE","GDPC1","PCE","INDPRO","PAYEMS","GNPCA","M2SL","HOUST"]
def fetch_fred_last_values(series_ids, end_train, api_key):
    last_vals = {}
    for sid in series_ids:
        try:
            url = f"https://api.stlouisfed.org/fred/series/observations?series_id={sid}&api_key={api_key}&file_type=json"
            r = requests.get(url, timeout=20).json()
            df = pd.DataFrame(r["observations"])[["date","value"]]
            df["date"] = pd.to_datetime(df["date"], errors="coerce")
            df["value"] = pd.to_numeric(df["value"], errors="coerce")
            df = df.dropna().sort_values("date")
            df = df[df["date"] <= end_train]
            if not df.empty:
                last_vals[sid] = float(df.iloc[-1]["value"])
        except Exception:
            pass
    return last_vals  # dict sid -> value

# ---- Fundamentales FMP (último valor conocido por columna) ----
def fetch_fundamentals_last(ticker, end_train, api_key):
    endpoints = {
        "marketcap": f"https://financialmodelingprep.com/stable/historical-market-capitalization?symbol={ticker}&apikey={api_key}",
        "income": f"https://financialmodelingprep.com/stable/income-statement?symbol={ticker}&period=quarter&apikey={api_key}",
        "balance": f"https://financialmodelingprep.com/stable/balance-sheet-statement?symbol={ticker}&period=quarter&apikey={api_key}",
        "cashflow": f"https://financialmodelingprep.com/stable/cash-flow-statement?symbol={ticker}&period=quarter&apikey={api_key}",
        "keymetrics": f"https://financialmodelingprep.com/stable/key-metrics?symbol={ticker}&period=annual&apikey={api_key}",
        "income_growth": f"https://financialmodelingprep.com/stable/income-statement-growth?symbol={ticker}&period=quarter&apikey={api_key}",
        "balance_growth": f"https://financialmodelingprep.com/stable/balance-sheet-statement-growth?symbol={ticker}&period=quarter&apikey={api_key}",
        "cashflow_growth": f"https://financialmodelingprep.com/stable/cash-flow-statement-growth?symbol={ticker}&period=quarter&apikey={api_key}",
        "financial_growth": f"https://financialmodelingprep.com/stable/financial-growth?symbol={ticker}&period=quarter&apikey={api_key}",
    }
    col_last = {}
    for name, url in endpoints.items():
        try:
            data = requests.get(url, timeout=15).json()
            df = pd.DataFrame(data)
            if df.empty: continue
            date_col = next((c for c in df.columns if "date" in c.lower() or "period" in c.lower()), None)
            if date_col is None: continue
            df["Date"] = pd.to_datetime(df[date_col], errors="coerce")
            df = df.dropna(subset=["Date"])
            df = df[df["Date"] <= end_train].sort_values("Date")
            if df.empty: continue
            # Para cada columna numérica, guardo último valor
            numc = [c for c in df.columns if c not in ["Date"] and pd.api.types.is_numeric_dtype(df[c])]
            for c in numc:
                col_last[f"{name}_{c}"] = float(df.iloc[-1][c]) if pd.notna(df.iloc[-1][c]) else np.nan
        except Exception:
            continue
    return col_last  # dict fully_qualified_col -> value

# ---- Noticias (Google News RSS multi-query) + FinBERT ----
def rss_parse(url: str) -> list[dict]:
    try:
        import feedparser
        d = feedparser.parse(url)
        rows = []
        for e in d.entries:
            rows.append({
                "title": getattr(e, "title", "") or "",
                "summary": getattr(e, "summary", "") or getattr(e, "description", "") or "",
                "url": getattr(e, "link", "") or "",
                "published": getattr(e, "published", "") or getattr(e, "updated", "") or "",
                "publisher": getattr(e, "source", {}).get("title", "").lower() if hasattr(e, "source") else (getattr(e, "publisher", "") or getattr(e, "author", "")).lower()
            })
        return rows
    except Exception:
        # fallback simple
        try:
            import re
            r = requests.get(url, timeout=15); r.raise_for_status()
            text = r.text
            items = re.findall(r"<item>(.*?)</item>", text, flags=re.DOTALL|re.IGNORECASE)
            rows=[]
            for it in items:
                def _tag(t):
                    m = re.search(fr"<{t}>(.*?)</{t}>", it, flags=re.DOTALL|re.IGNORECASE)
                    return re.sub("<.*?>","", m.group(1)).strip() if m else ""
                rows.append({
                    "title": _tag("title"), "summary": _tag("description"), "url": _tag("link"),
                    "published": _tag("pubDate"), "publisher": ""
                })
            return rows
        except Exception:
            return []

def gnews_fetch_multi(queries, start_dt, end_dt, days_window=365):
    frames = []
    for q in queries:
        q2 = quote_plus(q + f" when:{days_window}d")
        url = f"https://news.google.com/rss/search?q={q2}&hl=en-US&gl=US&ceid=US:en"
        rows = rss_parse(url)
        for r in rows:
            r["published"] = str(r.get("published",""))
        df = pd.DataFrame(rows)
        if df.empty: continue
        df["Date"] = pd.to_datetime(df["published"], errors="coerce").dt.normalize()
        df = df.dropna(subset=["Date"])
        df = df[(df["Date"] >= start_dt) & (df["Date"] <= end_dt)]
        df["publisher"] = df["publisher"].astype(str).str.lower()
        frames.append(df[["Date","title","summary","url","publisher"]])
    if not frames:
        return pd.DataFrame(columns=["Date","title","summary","url","publisher"])
    out = pd.concat(frames, axis=0, ignore_index=True)
    if "url" in out.columns: out = out.drop_duplicates(subset=["url"]).reset_index(drop=True)
    out = out.drop_duplicates(subset=["title"]).reset_index(drop=True)
    return out

SOURCE_WEIGHTS = {
    "reuters": 1.5, "bloomberg": 1.5, "wall street journal": 1.4, "wsj": 1.4,
    "financial times": 1.3, "ft": 1.3, "the information": 1.2, "the verge": 1.1,
    "yahoo": 0.9, "seeking alpha": 0.9, "investing.com": 0.9, "motley fool": 0.9,
}
def weight_for_pub(p: str) -> float:
    if not p: return 1.0
    p = p.lower()
    for k,v in SOURCE_WEIGHTS.items():
        if k in p: return float(v)
    return 1.0

def score_finbert_daily_weighted(df_news: pd.DataFrame) -> pd.DataFrame:
    if df_news.empty:
        return pd.DataFrame(columns=["Date","neg","neu","pos","news_count","neg_ewm3","neu_ewm3","pos_ewm3","news_pos_interact"])
    texts = (df_news["title"].fillna("") + ". " + df_news["summary"].fillna("")).tolist()
    weights = df_news["publisher"].apply(weight_for_pub).to_numpy()
    probs_all, w_all = [], []
    batch = 16
    for i in range(0, len(texts), batch):
        chunk = texts[i:i+batch]; w_chunk = weights[i:i+batch]
        inputs = tokenizer(chunk, return_tensors="pt", truncation=True, padding=True, max_length=256).to(device)
        with torch.no_grad():
            outputs = model_sent(**inputs)
            probs = torch.nn.functional.softmax(outputs.logits, dim=-1).detach().cpu().numpy()
        probs_all.append(probs); w_all.append(w_chunk)
    arr = np.vstack(probs_all); wv = np.concatenate(w_all)
    df_scored = df_news.iloc[:arr.shape[0]].copy()
    df_scored[["neg","neu","pos"]] = arr
    df_scored["w"] = wv

    rows=[]
    for d,g in df_scored.groupby("Date"):
        a = g[["neg","neu","pos"]].to_numpy(); w = g["w"].to_numpy()
        s = np.sum(w); ww = w/ s if s>0 else np.ones_like(w)/len(w)
        mean = (a*ww[:,None]).sum(axis=0)
        rows.append({"Date":d, "neg":mean[0], "neu":mean[1], "pos":mean[2], "news_count": float(g.shape[0])})
    daily = pd.DataFrame(rows).sort_values("Date").reset_index(drop=True)
    for c in ["neg","neu","pos"]:
        daily[f"{c}_ewm3"] = daily[c].ewm(span=3, adjust=False).mean()
    daily["news_pos_interact"] = daily["news_count"] * daily["pos"]
    return daily

# =======================
# 1) Construcción minimal forward
# =======================
start_need = (END_TRAIN - pd.Timedelta(days=LOOKBACK_DAYS)).normalize()

prices = fetch_prices_fmp(TICKER, start_need, END_FWD, FMP_API_KEY)
sp500  = load_sp500_csv(SP500_CSV)
treas  = fetch_treasury_10y(FMP_API_KEY)

merged = pd.merge_asof(prices.sort_values("Date"), sp500.sort_values("Date"), on="Date", direction="backward")
merged = pd.merge_asof(merged.sort_values("Date"), treas.sort_values("Date"), on="Date", direction="backward")

# Técnicos mínimos
ret = merged["close"].pct_change()
merged["r_nvda"] = ret
merged["r_mkt"]  = merged["sp500_close"].pct_change()
merged["excess_nvda"] = merged["r_nvda"] - merged["rf"] / 252.0
merged["excess_mkt"]  = merged["r_mkt"]  - merged["rf"] / 252.0

high, low, close = merged["high"], merged["low"], merged["close"]
tr = pd.concat([(high - low), (high - close.shift(1)).abs(), (low - close.shift(1)).abs()], axis=1).max(axis=1)
merged["TR"] = tr.values
merged["ATR_14"] = tr.rolling(14).mean().values

plus_dm = np.where((high - high.shift(1)) > (low.shift(1) - low), np.maximum(high - high.shift(1), 0), 0)
minus_dm = np.where((low.shift(1) - low) > (high - high.shift(1)), np.maximum(low - low.shift(1), 0), 0)
tr14 = tr.rolling(14).sum()
plus_di14 = 100 * (pd.Series(plus_dm).rolling(14).sum() / tr14)
minus_di14 = 100 * (pd.Series(minus_dm).rolling(14).sum() / tr14)
dx = (abs(plus_di14 - minus_di14) / (plus_di14 + minus_di14)) * 100
merged["ADX_14"] = dx.rolling(14).mean()

merged["ema10"] = merged["close"].ewm(span=10).mean()
merged["ema20"] = merged["close"].ewm(span=20).mean()
merged["ema50"] = merged["close"].ewm(span=50).mean()
merged["ema100"] = merged["close"].ewm(span=100).mean()
merged["ema10_above_ema50"] = (merged["ema10"] > merged["ema50"]).astype(int)
merged["ema20_above_ema50"] = (merged["ema20"] > merged["ema50"]).astype(int)
merged["slope_20"] = merged["close"].diff(20)
merged["slope_50"] = merged["close"].diff(50)
merged["roc_5"]  = _safe_pct_change(merged["close"], 5)
merged["roc_10"] = _safe_pct_change(merged["close"],10)
merged["roc_20"] = _safe_pct_change(merged["close"],20)
merged["roc_50"] = _safe_pct_change(merged["close"],50)
merged["vol_10"] = ret.rolling(10).std()
merged["vol_20"] = ret.rolling(20).std()
merged["vol_50"] = ret.rolling(50).std()
merged["ewma_vol"] = ret.ewm(span=20).std()
merged["sp_slope_20"] = merged["sp500_close"].diff(20)
merged["PVT"] = (merged["close"].pct_change() * merged["volume"]).cumsum()

# Quarter & dummies
merged["quarter"] = merged["Date"].dt.quarter.astype("Int64")
q = merged["quarter"].astype(float)
merged["q_sin"] = np.sin(2*np.pi*(q-1)/4)
merged["q_cos"] = np.cos(2*np.pi*(q-1)/4)
qdum = pd.get_dummies(merged["quarter"], prefix="q", dtype=int)
merged = pd.concat([merged, qdum], axis=1)

# Target igual que en entrenamiento (para evaluación)
merged["bull_t"] = ((merged["slope_20"] > 0) & (merged["close"] > merged["ema50"])).astype(int)
merged["target"] = merged["bull_t"].shift(-1)

# =======================
# 2) Inyección de FRED/fundamentales con último valor conocido
# =======================
fred_last = fetch_fred_last_values(FRED_IDS, END_TRAIN, FRED_API_KEY)
for sid, val in fred_last.items():
    merged[sid] = float(val)

fund_last = fetch_fundamentals_last(TICKER, END_TRAIN, FMP_API_KEY)
for col, val in fund_last.items():
    merged[col] = float(val)

# =======================
# 3) Noticias forward (Google News RSS) + FinBERT
# =======================
queries = ["NVDA OR Nvidia", "NVIDIA Corporation", "Nvidia earnings"]
news_df = gnews_fetch_multi(queries, start_dt=start_need, end_dt=END_FWD, days_window=365)
daily_sent = score_finbert_daily_weighted(news_df)
if daily_sent.empty:
    # columna dummy neutra (por si tus artefactos incluyen neg/neu/pos)
    daily_sent = pd.DataFrame({"Date": merged["Date"].copy(),
                               "neg":0.0,"neu":1.0,"pos":0.0,
                               "news_count":0.0,
                               "neg_ewm3":0.0,"neu_ewm3":1.0,"pos_ewm3":0.0,
                               "news_pos_interact":0.0})
merged = pd.merge_asof(merged.sort_values("Date"), daily_sent.sort_values("Date"), on="Date", direction="backward")

# =======================
# 4) Construcción de features EXACTAMENTE como p1/p2 los esperan
#    - Tomo los nombres del artefacto, saco el sufijo "_lag1" → base_cols
#    - Creo lag1 solo para esas base_cols (si existen)
# =======================
feats_p1 = pd.Index(art_p1["features"]).unique().tolist()
feats_p2 = pd.Index(art_p2["features"]).unique().tolist()
# union para construir sin sorpresas
feats_union = sorted(set(feats_p1) | set(feats_p2))
base_cols_needed = sorted(set([c[:-5] for c in feats_union if c.endswith("_lag1")]))

# crea lag1 SOLO de las que existan en merged
base_cols_present = [c for c in base_cols_needed if c in merged.columns]
lag1 = _lag_cols(merged, base_cols_present, lag=1)

# Ensambla dataset forward
df_feat = pd.concat([merged[["Date","target"]], lag1], axis=1).dropna(subset=["target"]).reset_index(drop=True)

# Slice forward (después del fin de entrenamiento)
mask_fwd = df_feat["Date"] > END_TRAIN
df_fwd = df_feat.loc[mask_fwd].copy()
if df_fwd.empty:
    raise RuntimeError("No hay días posteriores al fin de entrenamiento en el dataset construido.")

X_fwd_all = df_fwd[[c for c in df_fwd.columns if c not in ["Date","target"]]].copy()
X_fwd_all = X_fwd_all.loc[:, ~X_fwd_all.columns.duplicated()]
y_fwd = df_fwd["target"].astype(int).values
D = pd.to_datetime(df_fwd["Date"])

# =======================
# 5) Probabilidades calibradas + ensemble + stacking (sin tocar umbrales)
# =======================
p1 = _predict_proba_calibrated(X_fwd_all, art_p1)
p2 = _predict_proba_calibrated(X_fwd_all, art_p2)
pw = _weighted_average(p1, p2, w1, w2)

# si entrenaste el meta solo con [p1,p2]
X_meta = np.c_[p1, p2]
ps = meta_clf.predict_proba(X_meta)[:,1]

pred_p1 = (p1 >= thr1).astype(int)
pred_p2 = (p2 >= thr2).astype(int)
pred_w  = (pw >= thr_weighted).astype(int)
pred_s  = (ps >= thr_meta).astype(int)

def _report(tag, y_true, pred):
    f1  = f1_score(y_true, pred)
    acc = accuracy_score(y_true, pred)
    print(f"\n📈 {tag} FORWARD ({D.min().date()} → {D.max().date()}), N={len(y_true)}")
    print(f"F1={f1:.4f} | ACC={acc:.4f}")
    print(confusion_matrix(y_true, pred))
    print(classification_report(y_true, pred, digits=4))

_report("Period p1", y_fwd, pred_p1)
_report("Period p2", y_fwd, pred_p2)
_report("Weighted",  y_fwd, pred_w)
_report("Stacking",  y_fwd, pred_s)

# Tabla día a día
out = pd.DataFrame({
    "Date": D,
    "y_true": y_fwd,
    "proba_p1": p1, "pred_p1": pred_p1,
    "proba_p2": p2, "pred_p2": pred_p2,
    "proba_weighted": pw, "pred_weighted": pred_w,
    "proba_stacking": ps, "pred_stacking": pred_s,
})
print("\n🔎 Detalle forward:")
print(out.to_string(index=False))

fname = f"NVDA_forward_{D.min().date()}_to_{D.max().date()}.csv"
out.to_csv(fname, index=False)
print("\n💾 Guardado:", fname)
```

```python
# =============================================
#  NVDA BULL/BEAR — Dos modelos por época + Ensemble + Calibración + Stacking
# =============================================
import os
import json
from datetime import datetime, UTC

import joblib
import numpy as np
import pandas as pd
import requests
import torch
import matplotlib.pyplot as plt

from tqdm import tqdm
from transformers import AutoTokenizer, AutoModelForSequenceClassification

from sklearn.base import BaseEstimator, TransformerMixin, clone
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import RobustScaler
from sklearn.feature_selection import VarianceThreshold
from sklearn.metrics import (
    f1_score, classification_report, confusion_matrix,
    accuracy_score, precision_recall_curve
)
from sklearn.calibration import CalibratedClassifierCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

# =============================================
#              CONFIG
# =============================================
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"🧠 Device Torch: {device}")

print("🔄 Cargando FinBERT (esto puede tardar unos segundos)...")
tokenizer = AutoTokenizer.from_pretrained("yiyanghkust/finbert-tone", local_files_only=False)
model_sent = AutoModelForSequenceClassification.from_pretrained(
    "yiyanghkust/finbert-tone",
    local_files_only=False
).to(device)
print("✅ FinBERT cargado")

FRED_API_KEY = os.getenv("FRED_API_KEY", "")
FMP_API_KEY = os.getenv("FMP_API_KEY", "")
FINNHUB_API_KEY = os.getenv("FINNHUB_API_KEY", "")
USE_GPU = torch.cuda.is_available()

# === Fechas clave: ENTRENAMIENTO hasta 30-set-2025; FORWARD del 1 oct al 27-nov-2025 ===
END_TRAIN = "2025-9-30"
FWD_START = "2025-10-01"
FWD_END   = "2025-11-27"
SP500_TRAIN_CSV = "sp500.csv"        # S&P500 para entrenamiento (cortado a set-2025)
SP500_TEST_CSV  = "sp500_test.csv"   # S&P500 con días de oct y nov

# =============================================
#  Utils
# =============================================
def _safe_pct_change(s: pd.Series, periods: int) -> pd.Series:
    return s.pct_change(periods)

def _lag_cols(df: pd.DataFrame, cols: list[str], lag: int = 1) -> pd.DataFrame:
    lagged = df[cols].shift(lag)
    lagged.columns = [f"{c}_lag{lag}" for c in cols]
    return lagged

def _best_f1_threshold(y_true: np.ndarray, y_prob: np.ndarray) -> float:
    p, r, t = precision_recall_curve(y_true, y_prob)
    f1 = np.where((p + r) > 0, 2 * p * r / (p + r), 0.0)
    if len(t) == 0:
        return 0.5
    best = int(np.nanargmax(f1[:-1]))
    return float(np.clip(t[best], 0.01, 0.99))

def _weighted_average(p1: np.ndarray, p2: np.ndarray, w1: float, w2: float) -> np.ndarray:
    s = (w1 + w2) if (w1 + w2) > 0 else 1.0
    return (w1 * p1 + w2 * p2) / s

# === Helpers de deduplicación ===
def _unique_cols(df: pd.DataFrame) -> pd.DataFrame:
    if not isinstance(df, pd.DataFrame):
        return df
    return df.loc[:, ~df.columns.duplicated()].copy()

def _unique_list(seq):
    seen = set()
    out = []
    for x in seq:
        if x not in seen:
            seen.add(x)
            out.append(x)
    return out

# =============================================
#  Transformadores que preservan nombres/shape
# =============================================
class ColumnAligner(BaseEstimator, TransformerMixin):
    def fit(self, X, y=None):
        self.columns_ = list(X.columns) if isinstance(X, pd.DataFrame) else [f"f_{i}" for i in range(np.shape(X)[1])]
        return self
    def transform(self, X):
        df = X if isinstance(X, pd.DataFrame) else pd.DataFrame(X, columns=self.columns_)
        return df.reindex(columns=self.columns_)

class DataFrameSimpleImputer(SimpleImputer):
    def fit(self, X, y=None):
        self._cols = list(X.columns)
        return super().fit(X, y)
    def transform(self, X):
        arr = super().transform(X)
        return pd.DataFrame(arr, index=X.index, columns=self._cols)

class DataFrameRobustScaler(RobustScaler):
    def fit(self, X, y=None):
        self._cols = list(X.columns)
        return super().fit(X, y)
    def transform(self, X):
        arr = super().transform(X)
        return pd.DataFrame(arr, index=X.index, columns=self._cols)

class OutlierClipper(BaseEstimator, TransformerMixin):
    def __init__(self, lower_q: float = 0.01, upper_q: float = 0.99):
        self.lower_q = float(lower_q); self.upper_q = float(upper_q)
    def fit(self, X, y=None):
        X_df = X if isinstance(X, pd.DataFrame) else pd.DataFrame(X)
        self.columns_ = list(X_df.columns)
        self.lower_ = X_df.quantile(self.lower_q)
        self.upper_ = X_df.quantile(self.upper_q)
        return self
    def transform(self, X):
        X_df = X if isinstance(X, pd.DataFrame) else pd.DataFrame(X, columns=self.columns_)
        return X_df.clip(lower=self.lower_, upper=self.upper_, axis=1)

class CorrelationSelector(BaseEstimator, TransformerMixin):
    def __init__(self, threshold: float = 0.90):
        self.threshold = float(threshold)
    def fit(self, X, y=None):
        X_df = X if isinstance(X, pd.DataFrame) else pd.DataFrame(X)
        corr = X_df.corr().abs().fillna(0.0)
        upper = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))
        to_drop = [c for c in upper.columns if (upper[c] > self.threshold).any()]
        self.keep_ = [c for c in X_df.columns if c not in to_drop]
        return self
    def transform(self, X):
        X_df = X if isinstance(X, pd.DataFrame) else pd.DataFrame(X, columns=self.keep_)
        return X_df[self.keep_]
    def get_feature_names_out(self, input_features=None):
        return np.array(self.keep_)

class AllNaNToConstant(BaseEstimator, TransformerMixin):
    """Evita que el Imputer salte columnas 100% NaN (shape mismatch)."""
    def __init__(self, constant: float = 0.0):
        self.constant = constant
        self.all_nan_cols_: list[str] = []
    def fit(self, X, y=None):
        X_df = X if isinstance(X, pd.DataFrame) else pd.DataFrame(X)
        self.all_nan_cols_ = [c for c in X_df.columns if not X_df[c].notna().any()]
        return self
    def transform(self, X):
        X_df = X if isinstance(X, pd.DataFrame) else pd.DataFrame(X, columns=self.all_nan_cols_)
        if self.all_nan_cols_:
            X_df = X_df.copy()
            X_df[self.all_nan_cols_] = X_df[self.all_nan_cols_].fillna(self.constant)
        return X_df

# =============================================
#  Descarga/transform de fundamentales + FRED (serie y snapshot)
# =============================================
def _fetch_fundamentals_series(ticker: str) -> pd.DataFrame:
    endpoints = {
        "marketcap": f"https://financialmodelingprep.com/stable/historical-market-capitalization?symbol={ticker}&apikey={FMP_API_KEY}",
        "income": f"https://financialmodelingprep.com/stable/income-statement?symbol={ticker}&period=quarter&apikey={FMP_API_KEY}",
        "balance": f"https://financialmodelingprep.com/stable/balance-sheet-statement?symbol={ticker}&period=quarter&apikey={FMP_API_KEY}",
        "cashflow": f"https://financialmodelingprep.com/stable/cash-flow-statement?symbol={ticker}&period=quarter&apikey={FMP_API_KEY}",
        "keymetrics": f"https://financialmodelingprep.com/stable/key-metrics?symbol={ticker}&period=annual&apikey={FMP_API_KEY}",
        "income_growth": f"https://financialmodelingprep.com/stable/income-statement-growth?symbol={ticker}&period=quarter&apikey={FMP_API_KEY}",
        "balance_growth": f"https://financialmodelingprep.com/stable/balance-sheet-statement-growth?symbol={ticker}&period=quarter&apikey={FMP_API_KEY}",
        "cashflow_growth": f"https://financialmodelingprep.com/stable/cash-flow-statement-growth?symbol={ticker}&period=quarter&apikey={FMP_API_KEY}",
        "financial_growth": f"https://financialmodelingprep.com/stable/financial-growth?symbol={ticker}&period=quarter&apikey={FMP_API_KEY}",
    }
    dfs = []
    for name, url in endpoints.items():
        try:
            data = requests.get(url, timeout=20).json()
            df = pd.DataFrame(data)
            if not df.empty:
                date_col = [c for c in df.columns if "date" in c.lower() or "period" in c.lower()]
                if date_col:
                    df["Date"] = pd.to_datetime(df[date_col[0]], errors="coerce")
                df = df.drop_duplicates(subset="Date").set_index("Date").sort_index()
                df.columns = [f"{name}_{c}" for c in df.columns]
                dfs.append(df)
        except Exception as e:
            print(f"⚠️ fundamentals {name}: {e}")
    if not dfs:
        return pd.DataFrame()
    out = pd.concat(dfs, axis=1).ffill().reset_index().rename(columns={"index":"Date"})
    return out

def _fetch_fred_series() -> pd.DataFrame:
    fred_ids = ["CPIAUCSL","UNRATE","GDPC1","PCE","INDPRO","PAYEMS","GNPCA","M2SL","HOUST"]
    def fred(series_id):
        try:
            url = f"https://api.stlouisfed.org/fred/series/observations?series_id={series_id}&api_key={FRED_API_KEY}&file_type=json"
            r = requests.get(url, timeout=25).json()
            df = pd.DataFrame(r["observations"])[["date", "value"]]
            df["date"] = pd.to_datetime(df["date"], errors="coerce")
            df["value"] = pd.to_numeric(df["value"], errors="coerce")
            return df.dropna().set_index("date").rename(columns={"value": series_id})
        except Exception:
            return pd.DataFrame()
    parts = [fred(fid) for fid in fred_ids]
    parts = [p for p in parts if not p.empty]
    if not parts:
        return pd.DataFrame()
    out = pd.concat(parts, axis=1).ffill().reset_index().rename(columns={"date":"Date"})
    return out

def fundamentals_snapshot_at(asof_date: pd.Timestamp) -> pd.DataFrame:
    """Devuelve una fila (Date=asof_date) con el último valor conocido de cada fundamental/FRED hasta esa fecha."""
    fnds = _fetch_fundamentals_series("NVDA")
    fred = _fetch_fred_series()
    base = pd.DataFrame({"Date":[pd.to_datetime(asof_date).normalize()]})
    for block in [fnds, fred]:
        if block is not None and not block.empty:
            tmp = pd.merge_asof(base.sort_values("Date"), block.sort_values("Date"), on="Date", direction="backward")
            base = pd.merge(base, tmp, on="Date", how="left")
    return base.fillna(method="ffill").fillna(method="bfill")

# =============================================
#  Noticias (Finnhub → GNews RSS → Yahoo → dummy) + FinBERT
# =============================================
def _chunk_dates(dstart: pd.Timestamp, dend: pd.Timestamp, days: int = 365):
    cur = dstart
    while cur <= dend:
        nxt = min(cur + pd.Timedelta(days=days-1), dend)
        yield cur, nxt
        cur = nxt + pd.Timedelta(days=1)

def _rss_parse(url: str) -> list[dict]:
    try:
        import feedparser
        d = feedparser.parse(url)
        rows = []
        for e in d.entries:
            rows.append({
                "title": getattr(e, "title", "") or "",
                "summary": getattr(e, "summary", "") or getattr(e, "description", "") or "",
                "url": getattr(e, "link", "") or "",
                "published": getattr(e, "published", "") or getattr(e, "updated", "") or "",
                "publisher": getattr(e, "source", {}).get("title", "").lower() if hasattr(e, "source") else (getattr(e, "publisher", "") or getattr(e, "author", "")).lower()
            })
        return rows
    except Exception:
        try:
            import re
            r = requests.get(url, timeout=15); r.raise_for_status()
            text = r.text
            items = re.findall(r"<item>(.*?)</item>", text, flags=re.DOTALL|re.IGNORECASE)
            rows = []
            for it in items:
                def _tag(t):
                    m = re.search(fr"<{t}>(.*?)</{t}>", it, flags=re.DOTALL|re.IGNORECASE)
                    return re.sub("<.*?>", "", m.group(1)).strip() if m else ""
                rows.append({
                    "title": _tag("title"),
                    "summary": _tag("description"),
                    "url": _tag("link"),
                    "published": _tag("pubDate"),
                    "publisher": ""
                })
            return rows
        except Exception:
            return []

def _normalize_rows(rows: list[dict], src: str) -> pd.DataFrame:
    if not rows:
        return pd.DataFrame(columns=["Date","title","summary","url","source","publisher"])
    out = []
    for r in rows:
        if src == "finnhub":
            dt = r.get("datetime")
            if isinstance(dt, (int,float)): dt = pd.to_datetime(dt, unit="s", errors="coerce")
            else: dt = pd.to_datetime(dt, errors="coerce")
            out.append({
                "Date": dt.normalize() if pd.notna(dt) else pd.NaT,
                "title": r.get("headline") or r.get("title") or "",
                "summary": r.get("summary") or "",
                "url": r.get("url") or "",
                "source": "finnhub",
                "publisher": (r.get("source") or "").lower()
            })
        else:
            dt = pd.to_datetime(r.get("published"), errors="coerce")
            out.append({
                "Date": dt.normalize() if pd.notna(dt) else pd.NaT,
                "title": r.get("title") or "",
                "summary": r.get("summary") or "",
                "url": r.get("url") or "",
                "source": src,
                "publisher": (r.get("publisher") or r.get("source") or "").lower()
            })
    df = pd.DataFrame(out).dropna(subset=["Date"]).reset_index(drop=True)
    if not df.empty:
        if "url" in df.columns:
            df = df.drop_duplicates(subset=["url"]).reset_index(drop=True)
        df = df.drop_duplicates(subset=["title"]).reset_index(drop=True)
    return df

def _finnhub_fetch(symbol: str, start: pd.Timestamp, end: pd.Timestamp, api_key: str) -> pd.DataFrame:
    base = "https://finnhub.io/api/v1/company-news"
    all_rows = []
    for s,e in _chunk_dates(start, end, days=30):
        try:
            resp = requests.get(base, params={"symbol":symbol,"from":s.date().isoformat(),"to":e.date().isoformat(),"token":api_key}, timeout=20)
            if resp.status_code == 200:
                rows = resp.json()
                if isinstance(rows, list) and rows:
                    all_rows.extend(rows)
        except Exception:
            pass
    return _normalize_rows(all_rows, "finnhub")

def _gnews_fetch_multi(queries: list[str], start: pd.Timestamp, end: pd.Timestamp) -> pd.DataFrame:
    """Mejor esfuerzo: usa when:Nd con el total de días (si el motor respeta >365 genial; si no, traerá lo más reciente)."""
    from urllib.parse import quote_plus
    days = int((end - start).days) + 1
    frames = []
    for q in queries:
        q2 = quote_plus(q + f" when:{days}d")
        url = f"https://news.google.com/rss/search?q={q2}&hl=en-US&gl=US&ceid=US:en"
        rows = _rss_parse(url)
        for r in rows:
            r["published"] = str(r.get("published",""))
        frames.append(_normalize_rows(rows, "gnews"))
    if frames:
        df = pd.concat(frames, axis=0, ignore_index=True)
        if "url" in df.columns: df = df.drop_duplicates(subset=["url"]).reset_index(drop=True)
        df = df.drop_duplicates(subset=["title"]).reset_index(drop=True)
        # filtra exacto por ventana
        df = df[(df["Date"] >= start.normalize()) & (df["Date"] <= end.normalize())]
        return df
    return pd.DataFrame(columns=["Date","title","summary","url","source","publisher"])

def _yahoo_rss_nvda(start: pd.Timestamp, end: pd.Timestamp) -> pd.DataFrame:
    url = "https://feeds.finance.yahoo.com/rss/2.0/headline?s=NVDA&region=US&lang=en-US"
    rows = _rss_parse(url)
    for r in rows:
        r["published"] = str(r.get("published",""))
    df = _normalize_rows(rows, "yahoo")
    if not df.empty:
        df = df[(df["Date"] >= start.normalize()) & (df["Date"] <= end.normalize())]
    return df

# Ponderación simple por fuente
SOURCE_WEIGHTS = {
    "reuters": 1.5, "bloomberg": 1.5, "wall street journal": 1.4, "wsj": 1.4,
    "financial times": 1.3, "ft": 1.3, "the verge": 1.1, "the information": 1.2,
    "yahoo": 0.9, "seeking alpha": 0.9, "investing.com": 0.9, "motley fool": 0.9,
}
def _weight_for_publisher(p: str) -> float:
    if not p: return 1.0
    p = p.lower()
    for k,v in SOURCE_WEIGHTS.items():
        if k in p: return float(v)
    return 1.0

def _score_finbert_daily_weighted(df_news: pd.DataFrame) -> pd.DataFrame:
    if df_news.empty:
        return pd.DataFrame(columns=["Date","neg","neu","pos","news_count","neg_ewm3","neu_ewm3","pos_ewm3","news_pos_interact"])
    texts = (df_news["title"].fillna("") + ". " + df_news["summary"].fillna("")).tolist()
    weights = df_news["publisher"].apply(_weight_for_publisher).to_numpy()
    batch = 16
    probs_all, w_all = [], []
    for i in range(0, len(texts), batch):
        chunk = texts[i:i+batch]
        w_chunk = weights[i:i+batch]
        try:
            inputs = tokenizer(chunk, return_tensors="pt", truncation=True, padding=True, max_length=256).to(device)
            with torch.no_grad():
                outputs = model_sent(**inputs)
                probs = torch.nn.functional.softmax(outputs.logits, dim=-1).detach().cpu().numpy()
            probs_all.append(probs); w_all.append(w_chunk)
        except Exception:
            continue
    if not probs_all: 
        return pd.DataFrame(columns=["Date","neg","neu","pos","news_count","neg_ewm3","neu_ewm3","pos_ewm3","news_pos_interact"])
    arr = np.vstack(probs_all); wv = np.concatenate(w_all)
    df_scored = df_news.iloc[:arr.shape[0]].copy()
    df_scored[["neg","neu","pos"]] = arr
    df_scored["w"] = wv

    # agregación diaria ponderada
    def _wmean(g, col):
        a = g[col].to_numpy(); w = g["w"].to_numpy()
        s = np.sum(w); 
        return float(np.sum(a*w)/s) if s > 0 else float(np.mean(a))
    grouped = []
    for d, g in df_scored.groupby("Date"):
        grouped.append({
            "Date": d,
            "neg": _wmean(g, "neg"),
            "neu": _wmean(g, "neu"),
            "pos": _wmean(g, "pos"),
            "news_count": float(g.shape[0])
        })
    daily = pd.DataFrame(grouped).sort_values("Date").reset_index(drop=True)
    for c in ["neg","neu","pos"]:
        daily[f"{c}_ewm3"] = daily[c].ewm(span=3, adjust=False).mean()
    daily["news_pos_interact"] = daily["news_count"] * daily["pos"]
    return daily

def _collect_news_with_fallback(start: pd.Timestamp, end: pd.Timestamp) -> tuple[pd.DataFrame, str]:
    """Finnhub (si hay key) → Google News RSS multi-query (máximo posible) → Yahoo → Dummy."""
    used_source = "dummy"
    df_news_all = pd.DataFrame(columns=["Date","title","summary","url","source","publisher"])

    # 1) Finnhub (cubre todo el rango exacto por chunks)
    try:
        api_key = FINNHUB_API_KEY if FINNHUB_API_KEY else None
        if api_key:
            df_fh = _finnhub_fetch("NVDA", start, end, api_key)
            if not df_fh.empty:
                df_news_all = df_fh; used_source = "finnhub"
    except Exception as e:
        print("⚠️ Finnhub:", e)

    # 2) Google News RSS con ventana = días totales (mejor esfuerzo)
    if df_news_all.empty:
        try:
            queries = ["NVDA OR Nvidia", "NVIDIA Corporation", "Nvidia earnings"]
            df_gn = _gnews_fetch_multi(queries, start, end)
            if not df_gn.empty:
                df_news_all = df_gn; used_source = "gnews"
        except Exception as e:
            print("⚠️ GNews:", e)

    # 3) Yahoo Finance RSS
    if df_news_all.empty:
        try:
            df_yh = _yahoo_rss_nvda(start, end)
            if not df_yh.empty:
                df_news_all = df_yh; used_source = "yahoo"
        except Exception as e:
            print("⚠️ Yahoo RSS:", e)

    # 4) Dummy
    if df_news_all.empty:
        print("🧠 FinBERT dummy (fallback)…")
        samples = [
            f"NVDA strong demand for AI chips",
            f"NVDA faces competition in GPU and data center markets",
            f"NVDA market outlook remains positive",
            f"NVDA growth prospects show investor optimism",
        ]
        dates_dummy = pd.date_range(start=start, end=end, freq="7D")
        df_news_all = pd.DataFrame({
            "Date": dates_dummy, "title": samples*(len(dates_dummy)//len(samples)+1),
            "summary": "", "url": "", "source": "dummy", "publisher": "dummy"
        }).iloc[:len(dates_dummy)]
        used_source = "dummy"

    return df_news_all, used_source

# =============================================
#  Dataset (anti-leak; target t+1 | X_t lag1)
# =============================================
def build_dataset_all(ticker: str, start_date: str = "2000-01-01", end_date: str = END_TRAIN):
    print(f"📡 Descargando datos diarios de {ticker} desde FMP... {start_date} → {end_date}")
    url_price = (
        f"https://financialmodelingprep.com/api/v3/historical-price-full/"
        f"{ticker}?from={start_date}&to={end_date}&apikey={FMP_API_KEY}"
    )
    r = requests.get(url_price).json()
    if "historical" not in r:
        raise ValueError("❌ No se pudieron obtener precios históricos desde FMP.")
    df_prices = pd.DataFrame(r["historical"])
    df_prices["Date"] = pd.to_datetime(df_prices["date"], errors="coerce")
    df_prices = df_prices.sort_values("Date").reset_index(drop=True)
    df_prices = df_prices[(df_prices["Date"] >= pd.to_datetime(start_date)) & (df_prices["Date"] <= pd.to_datetime(end_date))]
    print(f"✅ {len(df_prices)} días {df_prices['Date'].min().date()} → {df_prices['Date'].max().date()}")

    print("🏦 Fundamentales FMP...")
    fundamentals = _fetch_fundamentals_series(ticker)
    if fundamentals is None or fundamentals.empty:
        fundamentals = pd.DataFrame({"Date": df_prices["Date"]})
    print(f"✅ Fundamentales: {fundamentals.shape[0]} filas, {fundamentals.shape[1]} cols")

    print("📊 S&P500 local (sp500.csv)...")
    with open(SP500_TRAIN_CSV, "r", encoding="utf-8-sig") as f:
        first = f.readline(); sep = ";" if first.count(";") > first.count(",") else ","
    sp500 = pd.read_csv(SP500_TRAIN_CSV, sep=sep, encoding="utf-8-sig")
    sp500.columns = [c.strip().lower() for c in sp500.columns]
    date_col = next((c for c in sp500.columns if "date" in c or "fecha" in c), None)
    close_col = next((c for c in sp500.columns if "close" in c or "cierre" in c), None)
    sp500 = sp500[[date_col, close_col]].rename(columns={date_col:"Date", close_col:"sp500_close"})
    sp500["Date"] = pd.to_datetime(sp500["Date"], errors="coerce", dayfirst=True)
    sp500["sp500_close"] = sp500["sp500_close"].astype(str).str.replace(",", ".").astype(float)
    sp500 = sp500.dropna().sort_values("Date").reset_index(drop=True)

    print("💵 Treasury (rf)...")
    tdata = requests.get(f"https://financialmodelingprep.com/stable/treasury-rates?apikey={FMP_API_KEY}").json()
    df_treasury = pd.DataFrame(tdata)
    df_treasury["Date"] = pd.to_datetime(df_treasury["date"], errors="coerce")
    df_treasury = df_treasury.sort_values("Date")
    ten_cols = [c for c in df_treasury.columns if "10" in c and ("year" in c.lower() or "y" in c.lower())]
    if ten_cols:
        df_treasury["rf"] = pd.to_numeric(df_treasury[ten_cols[0]], errors="coerce") / 100.0
    else:
        numc = df_treasury.select_dtypes(include=[np.number]).columns
        df_treasury["rf"] = (df_treasury[numc].mean(axis=1) / 100.0) if len(numc) else 0.0

    print("🔗 Merge total...")
    merged = pd.merge_asof(df_prices.sort_values("Date"), fundamentals.sort_values("Date"), on="Date", direction="backward")
    merged = pd.merge_asof(merged.sort_values("Date"), sp500.sort_values("Date"), on="Date", direction="backward")
    merged = pd.merge_asof(merged.sort_values("Date"), df_treasury[["Date","rf"]].sort_values("Date"), on="Date", direction="backward")

    print("📈 Indicadores técnicos...")
    merged["r_nvda"] = merged["close"].pct_change()
    merged["r_mkt"]  = merged["sp500_close"].pct_change()
    merged["excess_nvda"] = merged["r_nvda"] - merged["rf"] / 252.0
    merged["excess_mkt"]  = merged["r_mkt"]  - merged["rf"] / 252.0
    cov = merged["excess_nvda"].rolling(30).cov(merged["excess_mkt"])
    var = merged["excess_mkt"].rolling(30).var()
    merged["beta_30d"] = cov / var

    high, low, close = merged["high"], merged["low"], merged["close"]
    plus_dm = np.where((high - high.shift(1)) > (low.shift(1) - low), np.maximum(high - high.shift(1), 0), 0)
    minus_dm = np.where((low.shift(1) - low) > (high - high.shift(1)), np.maximum(low - low.shift(1), 0), 0)
    tr = pd.concat([(high - low), (high - close.shift(1)).abs(), (low - close.shift(1)).abs()], axis=1).max(axis=1)
    tr14 = tr.rolling(14).sum()
    plus_dm14 = pd.Series(plus_dm).rolling(14).sum()
    minus_dm14 = pd.Series(minus_dm).rolling(14).sum()
    plus_di14 = 100 * (plus_dm14 / tr14)
    minus_di14 = 100 * (minus_dm14 / tr14)
    dx = (abs(plus_di14 - minus_di14) / (plus_di14 + minus_di14)) * 100
    merged["ADX_14"] = dx.rolling(14).mean()

    merged["PVT"] = (merged["close"].pct_change() * merged["volume"]).cumsum()
    merged["TR"] = tr.values
    merged["ATR_14"] = tr.rolling(14).mean().values

    # 📰 Noticias entrenamiento (máximo posible) en TODO el rango de entrenamiento
    print("📰 Buscando noticias para ENTRENAMIENTO (máxima cobertura)…")
    train_start, train_end = merged["Date"].min().normalize(), merged["Date"].max().normalize()
    news_train, src_train = _collect_news_with_fallback(train_start, train_end)
    daily_sent_train = _score_finbert_daily_weighted(news_train) if not news_train.empty else pd.DataFrame()
    if daily_sent_train.empty:
        daily_sent_train = pd.DataFrame({"Date": merged["Date"].copy(),
                                         "neg":0.0,"neu":1.0,"pos":0.0,"news_count":0.0,
                                         "neg_ewm3":0.0,"neu_ewm3":1.0,"pos_ewm3":0.0,"news_pos_interact":0.0})
    merged = pd.merge_asof(merged.sort_values("Date"), daily_sent_train.sort_values("Date"), on="Date", direction="backward")
    print(f"✅ Sentimiento integrado (train) desde: {src_train} — días con noticias: {int((merged['news_count']>0).sum())}")

    print("📈 Features + target (lag1, anti-leak)...")
    merged = merged.sort_values("Date").reset_index(drop=True)
    merged["ema10"] = merged["close"].ewm(span=10).mean()
    merged["ema20"] = merged["close"].ewm(span=20).mean()
    merged["ema40"] = merged["close"].ewm(span=40).mean() 
    merged["ema50"] = merged["close"].ewm(span=50).mean()
    merged["ema100"] = merged["close"].ewm(span=100).mean()
    merged["ema10_above_ema50"] = (merged["ema10"] > merged["ema50"]).astype(int)
    merged["ema20_above_ema50"] = (merged["ema20"] > merged["ema50"]).astype(int)

    # momentum
    merged["slope_10"] = merged["close"].diff(10)
    merged["slope_20"] = merged["close"].diff(20)
    merged["slope_50"] = merged["close"].diff(50)

    merged["roc_5"]  = _safe_pct_change(merged["close"], 5)
    merged["roc_10"] = _safe_pct_change(merged["close"], 10)
    merged["roc_20"] = _safe_pct_change(merged["close"], 20)
    merged["roc_50"] = _safe_pct_change(merged["close"], 50)
    ret = merged["close"].pct_change()
    merged["vol_10"] = ret.rolling(10).std()
    merged["vol_20"] = ret.rolling(20).std()
    merged["vol_50"] = ret.rolling(50).std()
    merged["ewma_vol"] = ret.ewm(span=20).std()
    merged["sp_slope_20"] = merged["sp500_close"].diff(20)

    # === NUEVO target (EMA20 + slope_5)
    merged["bull_t"] = ((merged["slope_10"] > 0) & (merged["close"] > merged["ema40"])).astype(int)
    merged["target"] = merged["bull_t"].shift(-1)

    # Señal trimestral y codificación cíclica
    merged["quarter"] = merged["Date"].dt.quarter.astype("Int64")
    q_dummies = pd.get_dummies(merged["quarter"], prefix="q", dtype=int)
    merged = pd.concat([merged, q_dummies], axis=1)
    q = merged["quarter"].astype(float)
    merged["q_sin"] = np.sin(2 * np.pi * (q - 1) / 4)
    merged["q_cos"] = np.cos(2 * np.pi * (q - 1) / 4)

    # Features base
    raw_cols = [
        "open","high","low","close","volume","ema10","ema20","ema40","ema50","ema100",
        "ema10_above_ema50","ema20_above_ema50","slope_10","slope_20","slope_50",
        "roc_5","roc_10","roc_20","roc_50","vol_10","vol_20","vol_50","ewma_vol",
        "r_nvda","r_mkt","excess_nvda","excess_mkt","beta_30d","ADX_14","PVT","TR","ATR_14",
        "sp500_close","sp_slope_20","rf",
        # Noticias
        "neg","neu","pos","news_count","pos_ewm3","neg_ewm3","news_pos_interact",
        # Trimestre
        "quarter","q_1","q_2","q_3","q_4","q_sin","q_cos",
    ]
    raw_cols = _unique_list([c for c in raw_cols if c in merged.columns])

    # Lags anti-leak + lag flexible de pos_ewm3
    feats_lag1 = _lag_cols(merged, raw_cols, lag=1)
    pos_ewm3_lag2 = merged[["pos_ewm3"]].shift(2) if "pos_ewm3" in merged.columns else pd.DataFrame(index=merged.index)
    if not pos_ewm3_lag2.empty:
        pos_ewm3_lag2.columns = ["pos_ewm3_lag2"]

    macro_cols = [c for c in merged.columns if c not in (["Date","target","bull_t"] + raw_cols)]
    feats_macro_lag1 = _lag_cols(merged, macro_cols, lag=1) if macro_cols else pd.DataFrame(index=merged.index)

    X_full = pd.concat([feats_lag1, feats_macro_lag1, pos_ewm3_lag2], axis=1)
    X_full = _unique_cols(X_full)

    df = pd.concat([merged[["Date","target"]], X_full], axis=1)
    df = df.dropna(subset=["target"]).reset_index(drop=True)

    non_lag = [c for c in df.columns if c not in ["Date","target"] and not c.endswith("_lag1") and c != "pos_ewm3_lag2"]
    df = df.drop(columns=non_lag) if non_lag else df
    df = _unique_cols(df)

    print(f"✅ Dataset final: {df.shape[0]} filas, {df.shape[1]-2} features (lag1).")

    # Mini-EDA por trimestre (opcional)
    try:
        tmp = merged.copy()
        tmp["target_t"] = merged["bull_t"].shift(-1)
        eda_q = (
            tmp.groupby("quarter", dropna=True)["target_t"]
            .mean().rename("bull_rate_next_day")
            .to_frame().reset_index()
            .sort_values("quarter")
        )
        print("\n📊 Bull-rate (t+1) por trimestre:")
        print(eda_q.to_string(index=False))
    except Exception:
        pass
    return df

# =============================================
#   Purged + Embargo TimeSeries Split
# =============================================
class PurgedTimeSeriesSplit:
    def __init__(self, n_splits=5, purge=5, embargo=5):
        self.n_splits=n_splits; self.purge=purge; self.embargo=embargo
    def split(self, X: pd.DataFrame):
        n = len(X)
        fold = n // (self.n_splits + 1)
        for i in range(self.n_splits):
            test_start = (i+1)*fold
            train_end = max(0, test_start - self.purge)
            test_end = min(n, test_start + fold)
            yield np.arange(0, train_end), np.arange(test_start, test_end)
    def get_n_splits(self, X=None, y=None, groups=None): return self.n_splits

# =============================================
#  Modelos y pipeline
# =============================================
def build_models_dict():
    xgb_params = dict(
        n_estimators=500, max_depth=6, learning_rate=0.03,
        subsample=0.8, colsample_bytree=0.8, reg_alpha=0.5, reg_lambda=3.0,
        objective="binary:logistic", eval_metric="logloss",
        random_state=42, n_jobs=-1, scale_pos_weight=0.9, tree_method="hist",
    )
    lgbm_params = dict(
        boosting_type="gbdt", num_leaves=48, max_depth=-1, learning_rate=0.03,
        n_estimators=600, subsample=0.8, colsample_bytree=0.8,
        reg_alpha=0.2, reg_lambda=0.4, min_child_samples=25, scale_pos_weight=0.9,
        random_state=42, n_jobs=-1, verbosity=-1, device_type="gpu" if USE_GPU else "cpu",
    )
    return {
        "RandomForest": RandomForestClassifier(
            n_estimators=300, max_depth=12, min_samples_split=10, min_samples_leaf=20,
            max_features="sqrt", bootstrap=True, criterion="gini",
            random_state=42, n_jobs=-1, class_weight="balanced_subsample",
        ),
        "XGBoost": XGBClassifier(**xgb_params),
        "Logistic": LogisticRegression(
            penalty="l2", C=0.5, class_weight="balanced", solver="lbfgs", max_iter=1000, random_state=42,
        ),
        "LightGBM": LGBMClassifier(**lgbm_params),
    }

def make_pipeline(model):
    return Pipeline(steps=[
        ("align", ColumnAligner()),
        ("allnan", AllNaNToConstant(constant=0.0)),
        ("imputer", DataFrameSimpleImputer(strategy="median")),
        ("winsor", OutlierClipper(0.01, 0.99)),
        ("scaler", DataFrameRobustScaler(with_centering=True, with_scaling=True)),
        ("selector", CorrelationSelector(threshold=0.90)),
        ("varth", VarianceThreshold(1e-12)),
        ("model", model),
    ])

# =============================================
#  Entrenamiento por periodo (con calibración isotónica + umbral por época)
# =============================================
def train_period_model(df_all: pd.DataFrame, start: str, end: str, label: str):
    print(f"\n🗓️ Entrenando periodo {label}: {start} → {end}")
    msk = (df_all["Date"] >= pd.to_datetime(start)) & (df_all["Date"] <= pd.to_datetime(end))
    df = df_all.loc[msk].reset_index(drop=True)
    assert len(df) > 0, f"Periodo {label} vacío"

    feats_all = [c for c in df.columns if c not in ["Date","target"]]
    feats_num = [c for c in feats_all if pd.api.types.is_numeric_dtype(df[c])]
    low_var = [c for c in feats_num if df[c].nunique() <= 1]
    high_na = [c for c in feats_num if df[c].isna().mean() > 0.5]
    base_feats = [c for c in feats_num if c not in low_var + high_na]

    X = df[base_feats].copy()
    y = df["target"].astype(int).values

    models = build_models_dict()
    tscv = PurgedTimeSeriesSplit(n_splits=5, purge=5, embargo=5)

    best_name, best_cv = None, -np.inf
    for name, base_model in tqdm(models.items(), desc=f"🔁 Modelos {label}"):
        f1s=[]
        for tr_idx, te_idx in tscv.split(X):
            X_tr, X_te = X.iloc[tr_idx], X.iloc[te_idx]
            y_tr, y_te = y[tr_idx], y[te_idx]
            pipe = make_pipeline(clone(base_model))
            pipe.fit(X_tr, y_tr)
            pred = pipe.predict(X_te)
            f1s.append(f1_score(y_te, pred))
        avg = float(np.mean(f1s))
        print(f"   → {name:12s} F1-CV: {avg:.4f}")
        if avg > best_cv: best_cv, best_name = avg, name

    print(f"🏆 Mejor {label}: {best_name} (F1-CV={best_cv:.4f})")

    # Split interno para calibración + umbral
    split_inner = int(len(X) * 0.9)
    X_tr, y_tr = X.iloc[:split_inner], y[:split_inner]
    X_val, y_val = X.iloc[split_inner:], y[split_inner:]

    base_pipe = make_pipeline(models[best_name])
    base_pipe.fit(X_tr, y_tr)

    # Calibración isotónica sobre el pipeline ya entrenado
    calib = CalibratedClassifierCV(base_pipe, method="isotonic", cv="prefit")
    calib.fit(X_val, y_val)

    val_prob_cal = calib.predict_proba(X_val)[:, 1]
    thr = _best_f1_threshold(y_val, val_prob_cal)

    artifact = {
        "calibrator": calib,
        "features": _unique_list(base_feats),
        "threshold": float(thr),
        "cv_f1": float(best_cv),
        "period": [start, end],
        "model_name": best_name,
    }
    joblib.dump(artifact, f"NVDA_{label}_model.joblib")
    print(f"💾 Guardado NVDA_{label}_model.joblib (thr={thr:.3f})")

    return {
        "label": label, "artifact": artifact,
        "cv_best": best_cv, "features": base_feats, "dates": (start, end)
    }

# =============================================
#  Helpers de predicción calibrada
# =============================================
def _predict_proba_calibrated(df_features: pd.DataFrame, artifact: dict) -> np.ndarray:
    feats = _unique_list(list(artifact["features"]))
    Xf = _unique_cols(df_features)
    feats_final = [c for c in feats if c in Xf.columns]
    X = Xf.reindex(columns=feats_final, fill_value=np.nan)
    return artifact["calibrator"].predict_proba(X)[:, 1]

# =============================================
#  Ensemble + Stacking + Reportes
# =============================================
def train_two_models_and_ensemble(ticker="NVDA", start_date="2000-01-01"):
    df_all = build_dataset_all(ticker, start_date, END_TRAIN)

    p1 = train_period_model(df_all, "2000-01-01", "2019-12-31", "2000_2019")
    p2 = train_period_model(df_all, "2020-01-01", END_TRAIN, "2020_2025")

    art1, art2 = p1["artifact"], p2["artifact"]
    w1, w2 = p1["cv_best"], p2["cv_best"]

    X_all = df_all[[c for c in df_all.columns if c not in ["Date","target"]]].copy()
    X_all = _unique_cols(X_all)

    proba1 = _predict_proba_calibrated(X_all, art1)
    proba2 = _predict_proba_calibrated(X_all, art2)
    proba_w = _weighted_average(proba1, proba2, w1, w2)

    # Stacking basado en [proba1, proba2]
    y_all = df_all["target"].astype(int).values
    X_meta_all = np.c_[proba1, proba2]

    def _rolling_splits(n: int, n_splits: int = 5, purge: int = 5):
        fold = n // (n_splits + 1)
        for i in range(n_splits):
            test_start = (i + 1) * fold
            train_end = max(0, test_start - purge)
            test_end = min(n, test_start + fold)
            yield np.arange(0, train_end), np.arange(test_start, test_end)

    f1s_meta = []
    for tr_idx, te_idx in _rolling_splits(len(df_all), n_splits=5, purge=5):
        X_tr, X_te = X_meta_all[tr_idx], X_meta_all[te_idx]
        y_tr, y_te = y_all[tr_idx], y_all[te_idx]
        meta = LogisticRegression(penalty="l2", C=1.0, solver="lbfgs", max_iter=1000)
        meta.fit(X_tr, y_tr)
        pred_te = meta.predict(X_te)
        f1s_meta.append(f1_score(y_te, pred_te))
    print(f"🧱 Meta-Logistic F1-CV (rolling): {np.mean(f1s_meta):.4f}")

    split_inner = int(0.9 * len(df_all))
    X_tr_m, y_tr_m = X_meta_all[:split_inner], y_all[:split_inner]
    X_val_m, y_val_m = X_meta_all[split_inner:], y_all[split_inner:]
    meta = LogisticRegression(penalty="l2", C=1.0, solver="lbfgs", max_iter=1000)
    meta.fit(X_tr_m, y_tr_m)
    val_prob_m = meta.predict_proba(X_val_m)[:, 1]
    thr_meta = _best_f1_threshold(y_val_m, val_prob_m)
    joblib.dump({"meta_clf": meta, "threshold": float(thr_meta)}, "NVDA_stack.joblib")
    print(f"💾 Guardado NVDA_stack.joblib (thr_meta={thr_meta:.3f})")

    proba_stack = meta.predict_proba(X_meta_all)[:, 1]
    thr1, thr2 = float(art1["threshold"]), float(art2["threshold"])
    thr_ens = (w1 * thr1 + w2 * thr2) / (w1 + w2 if (w1 + w2) > 0 else 1.0)

    # Voto por periodo (umbral por época)
    dates = pd.to_datetime(df_all["Date"])
    start1, end1 = pd.to_datetime(art1["period"][0]), pd.to_datetime(art1["period"][1])
    start2, end2 = pd.to_datetime(art2["period"][0]), pd.to_datetime(art2["period"][1])
    pred_period = np.zeros_like(proba_stack, dtype=int)
    m1 = (dates >= start1) & (dates <= end1)
    m2 = (dates >= start2) & (dates <= end2)
    pred_period[m1] = (proba1[m1] >= thr1).astype(int)
    pred_period[m2] = (proba2[m2] >= thr2).astype(int)
    rest = ~(m1 | m2)
    pred_period[rest] = (proba_w[rest] >= (thr1+thr2)/2.0).astype(int)

    pred_weighted = (proba_w >= thr_ens).astype(int)
    pred_stack = (proba_stack >= thr_meta).astype(int)

    # Evaluación holdout (último 20%)
    split_ho = int(len(df_all) * 0.8)
    y_ho = y_all[split_ho:]
    d_ho = df_all["Date"].iloc[split_ho:]

    def _report(tag, y_true, pred):
        f1 = f1_score(y_true, pred); acc = accuracy_score(y_true, pred)
        print(f"🎯 {tag} HOLDOUT → F1={f1:.4f}, ACC={acc:.4f}")
        print(confusion_matrix(y_true, pred))
        print(classification_report(y_true, pred, digits=4))
        return f1, acc

    f1_p, acc_p = _report("Period-Vote", y_ho, pred_period[split_ho:])
    f1_w, acc_w = _report("Weighted",    y_ho, pred_weighted[split_ho:])
    f1_s, acc_s = _report("Stacking",    y_ho, pred_stack[split_ho:])

    ensemble_art = {
        "model_p1": "NVDA_2000_2019_model.joblib",
        "model_p2": "NVDA_2020_2025_model.joblib",
        "weights": {"p1": float(w1), "p2": float(w2)},
        "thresholds": {"p1": float(thr1), "p2": float(thr2), "weighted": float(thr_ens)}
    }
    with open("NVDA_ensemble.json", "w", encoding="utf-8") as f:
        json.dump(ensemble_art, f, indent=2)
    print("💾 Ensemble guardado en: NVDA_ensemble.json")

    # Gráficos (holdout)
    plt.figure(figsize=(12,5))
    plt.plot(d_ho, proba_stack[split_ho:], label="Stack proba")
    plt.title(f"{ticker} — Stacking Prob. Bull (holdout)")
    plt.legend(); plt.tight_layout(); plt.show()

    plt.figure(figsize=(12,5))
    plt.plot(d_ho, proba_w[split_ho:], label="Weighted proba")
    plt.axhline(thr_ens, linestyle="--", label=f"Umbral weighted {thr_ens:.2f}")
    plt.title(f"{ticker} — Weighted Prob. Bull (holdout)")
    plt.legend(); plt.tight_layout(); plt.show()

    with open("NVDA_results_summary.json","w",encoding="utf-8") as f:
        json.dump({
            "period_vote": {"f1": float(f1_p), "acc": float(acc_p)},
            "weighted":    {"f1": float(f1_w), "acc": float(acc_w)},
            "stacking":    {"f1": float(f1_s), "acc": float(acc_s)}
        }, f, indent=2)
    print("💾 Resultados guardados en: NVDA_results_summary.json")

    return {
        "period_vote": (f1_p, acc_p),
        "weighted": (f1_w, acc_w),
        "stacking": (f1_s, acc_s),
        "thr_ens": float(thr_ens),
        "thr_meta": float(thr_meta)
    }

# =============================================
#  Forward test integrado (1 octubre – 27 noviembre 2025)
#  - Usa snapshot de fundamentales/FRED a 30-set y ffill
#  - Usa noticias FinBERT de octubre y noviembre
#  - Usa exactamente las features que cada período entrenó
# =============================================
def run_forward_eval_nov(ticker="NVDA"):
    from sklearn.metrics import f1_score, accuracy_score, confusion_matrix, classification_report

    fwd_start = pd.to_datetime(FWD_START); fwd_end = pd.to_datetime(FWD_END)
    lookback = 200
    start_need = (pd.to_datetime(END_TRAIN) - pd.Timedelta(days=lookback)).normalize()

    # Artefactos
    art_p1 = joblib.load("NVDA_2000_2019_model.joblib")
    art_p2 = joblib.load("NVDA_2020_2025_model.joblib")
    stack_art = joblib.load("NVDA_stack.joblib")
    with open("NVDA_ensemble.json","r",encoding="utf-8") as f:
        ens_cfg = json.load(f)
    w1, w2 = ens_cfg["weights"]["p1"], ens_cfg["weights"]["p2"]
    thr1, thr2 = ens_cfg["thresholds"]["p1"], ens_cfg["thresholds"]["p2"]
    thr_weighted = ens_cfg["thresholds"]["weighted"]
    thr_meta = stack_art["threshold"]; meta_clf = stack_art["meta_clf"]

    FEATS_P1 = pd.Index(art_p1["features"]).unique()
    FEATS_P2 = pd.Index(art_p2["features"]).unique()

    # — Datos base: precios + S&P500 (oct y nov) + rf —
    def fetch_prices_fmp(ticker, start_dt, end_dt, api_key):
        url = (f"https://financialmodelingprep.com/api/v3/historical-price-full/"
               f"{ticker}?from={start_dt.date()}&to={end_dt.date()}&apikey={api_key}")
        r = requests.get(url, timeout=25).json()
        if "historical" not in r: raise RuntimeError("No prices from FMP")
        dfp = pd.DataFrame(r["historical"])
        dfp["Date"] = pd.to_datetime(dfp["date"], errors="coerce")
        dfp = dfp.sort_values("Date").reset_index(drop=True)
        keep = ["Date","open","high","low","close","volume"]
        for k in keep:
            if k not in dfp.columns and k!="Date":
                if k=="close" and "adjClose" in dfp.columns:
                    dfp["close"] = pd.to_numeric(dfp["adjClose"], errors="coerce")
                else:
                    dfp[k] = np.nan
        return dfp[keep]

    def load_sp500_csv(path):
        with open(path,"r",encoding="utf-8-sig") as f:
            first = f.readline(); sep = ";" if first.count(";") > first.count(",") else ","
        sp = pd.read_csv(path, sep=sep, encoding="utf-8-sig")
        sp.columns = [c.strip().lower() for c in sp.columns]
        date_col  = next((c for c in sp.columns if "date" in c or "fecha" in c), None)
        close_col = next((c for c in sp.columns if "close" in c or "cierre" in c), None)
        sp = sp[[date_col, close_col]].rename(columns={date_col:"Date", close_col:"sp500_close"})
        sp["Date"] = pd.to_datetime(sp["Date"], errors="coerce", dayfirst=True)
        sp["sp500_close"] = pd.to_numeric(sp["sp500_close"].astype(str).str.replace(",", "."), errors="coerce")
        sp = sp.dropna().sort_values("Date").reset_index(drop=True)
        return sp

    def fetch_treasury_10y(api_key):
        url = f"https://financialmodelingprep.com/stable/treasury-rates?apikey={api_key}"
        t = requests.get(url, timeout=25).json()
        dt = pd.DataFrame(t)
        dt["Date"] = pd.to_datetime(dt["date"], errors="coerce")
        dt = dt.sort_values("Date")
        ten_cols = [c for c in dt.columns if "10" in c and ("year" in c.lower() or "y" in c.lower())]
        if ten_cols:
            dt["rf"] = pd.to_numeric(dt[ten_cols[0]], errors="coerce")/100.0
        else:
            numc = dt.select_dtypes(include=[np.number]).columns
            dt["rf"] = (dt[numc].mean(axis=1) / 100.0) if len(numc) else 0.0
        return dt[["Date","rf"]]

    prices = fetch_prices_fmp(ticker, start_need, fwd_end, FMP_API_KEY)
    sp500  = load_sp500_csv(SP500_TEST_CSV)
    treas  = fetch_treasury_10y(FMP_API_KEY)

    merged = pd.merge_asof(prices.sort_values("Date"), sp500.sort_values("Date"), on="Date", direction="backward")
    merged = pd.merge_asof(merged.sort_values("Date"), treas.sort_values("Date"), on="Date", direction="backward")

    # — Técnicos mínimos coherentes —
    ret = merged["close"].pct_change()
    merged["r_nvda"] = ret
    merged["r_mkt"]  = merged["sp500_close"].pct_change()
    merged["excess_nvda"] = merged["r_nvda"] - merged["rf"]/252.0
    merged["excess_mkt"]  = merged["r_mkt"]  - merged["rf"]/252.0

    high, low, close = merged["high"], merged["low"], merged["close"]
    tr = pd.concat([(high-low), (high-close.shift(1)).abs(), (low-close.shift(1)).abs()], axis=1).max(axis=1)
    merged["TR"] = tr.values
    merged["ATR_14"] = tr.rolling(14).mean().values

    plus_dm = np.where((high - high.shift(1)) > (low.shift(1) - low), np.maximum(high - high.shift(1), 0), 0)
    minus_dm = np.where((low.shift(1) - low) > (high - high.shift(1)), np.maximum(low - low.shift(1), 0), 0)
    tr14 = tr.rolling(14).sum()
    plus_di14 = 100 * (pd.Series(plus_dm).rolling(14).sum() / tr14)
    minus_di14 = 100 * (pd.Series(minus_dm).rolling(14).sum() / tr14)
    dx = (abs(plus_di14 - minus_di14) / (plus_di14 + minus_di14)) * 100
    merged["ADX_14"] = dx.rolling(14).mean()

    merged["ema10"]  = merged["close"].ewm(span=10).mean()
    merged["ema20"]  = merged["close"].ewm(span=20).mean()
    merged["ema40"] = merged["close"].ewm(span=40).mean() 
    merged["ema50"]  = merged["close"].ewm(span=50).mean()
    merged["ema100"] = merged["close"].ewm(span=100).mean()
    merged["ema10_above_ema50"] = (merged["ema10"] > merged["ema50"]).astype(int)
    merged["ema20_above_ema50"] = (merged["ema20"] > merged["ema50"]).astype(int)
    merged["slope_10"]  = merged["close"].diff(5)
    merged["slope_20"] = merged["close"].diff(20)
    merged["slope_50"] = merged["close"].diff(50)
    merged["roc_5"]  = _safe_pct_change(merged["close"], 5)
    merged["roc_10"] = _safe_pct_change(merged["close"],10)
    merged["roc_20"] = _safe_pct_change(merged["close"],20)
    merged["roc_50"] = _safe_pct_change(merged["close"],50)
    merged["vol_10"] = ret.rolling(10).std()
    merged["vol_20"] = ret.rolling(20).std()
    merged["vol_50"] = ret.rolling(50).std()
    merged["ewma_vol"] = ret.ewm(span=20).std()
    merged["sp_slope_20"] = merged["sp500_close"].diff(20)

    merged["quarter"] = merged["Date"].dt.quarter.astype("Int64")
    q = merged["quarter"].astype(float)
    merged["q_sin"] = np.sin(2*np.pi*(q-1)/4)
    merged["q_cos"] = np.cos(2*np.pi*(q-1)/4)
    qdum = pd.get_dummies(merged["quarter"], prefix="q", dtype=int)
    merged = pd.concat([merged, qdum], axis=1)

    # — Snapshot de fundamentales/FRED a END_TRAIN + forward fill —
    snap = fundamentals_snapshot_at(pd.to_datetime(END_TRAIN))
    if not snap.empty:
        merged = pd.merge_asof(merged.sort_values("Date"), snap.sort_values("Date"), on="Date", direction="backward")
        for c in snap.columns:
            if c != "Date" and c in merged.columns:
                merged[c] = merged[c].ffill()

    # — Noticias de noviembre (FinBERT) —
    print("📰 Buscando noticias para FORWARD (noviembre)…")
    news_fwd, src_fwd = _collect_news_with_fallback(fwd_start, fwd_end)
    daily_news = _score_finbert_daily_weighted(news_fwd) if not news_fwd.empty else pd.DataFrame()
    if not daily_news.empty:
        merged = pd.merge_asof(merged.sort_values("Date"), daily_news.sort_values("Date"), on="Date", direction="backward")
    else:
        for c,v in [("neg",0.0),("neu",1.0),("pos",0.0),("news_count",0.0),("neg_ewm3",0.0),("neu_ewm3",1.0),("pos_ewm3",0.0),("news_pos_interact",0.0)]:
            merged[c]=v
    print(f"✅ Sentimiento integrado (forward) desde: {src_fwd}")

    # — lag1 de todo y recorte a ventana forward —
    all_cols = [c for c in merged.columns if c!="Date"]
    feats_lag1 = _lag_cols(merged, all_cols, lag=1)
    df_all = pd.concat([merged[["Date"]], feats_lag1], axis=1)
    X_fwd = df_all.set_index("Date").loc[fwd_start:fwd_end].reset_index()

    def _select_feats(df, feats_required):
        cols = ["Date"] + [c for c in df.columns if c!="Date" and c in feats_required]
        X = df.reindex(columns=cols, fill_value=np.nan)
        return X.loc[:, ~X.columns.duplicated()]

    X_p1 = _select_feats(X_fwd, pd.Index(art_p1["features"]).unique())
    X_p2 = _select_feats(X_fwd, pd.Index(art_p2["features"]).unique())

    proba1 = _predict_proba_calibrated(X_p1.drop(columns=["Date"]), art_p1)
    proba2 = _predict_proba_calibrated(X_p2.drop(columns=["Date"]), art_p2)
    proba_w = _weighted_average(proba1, proba2, w1, w2)
    X_meta = np.c_[proba1, proba2]
    proba_s = meta_clf.predict_proba(X_meta)[:,1]

    D = pd.to_datetime(X_p1["Date"])
    pred_p1 = (proba1 >= thr1).astype(int)
    pred_p2 = (proba2 >= thr2).astype(int)
    pred_w  = (proba_w >= thr_weighted).astype(int)
    pred_s  = (proba_s >= thr_meta).astype(int)

    # target forward con la misma lógica (EMA40 + slope_10)
    close = merged.set_index("Date")["close"]
    ema40 = merged.set_index("Date")["ema40"]
    slope10 = merged.set_index("Date")["slope_10"]
    bull_t = ((slope10>0) & (close>ema40)).astype(int)
    y = bull_t.shift(-1).loc[D.min():D.max()].reindex(D).fillna(0).astype(int).values

    def _report(tag, y_true, pred):
        f1 = f1_score(y_true, pred) if len(np.unique(y_true))>1 else float("nan")
        acc = accuracy_score(y_true, pred)
        print(f"\n📈 {tag} FORWARD ({D.min().date()} → {D.max().date()}), N={len(y_true)}")
        print(f"F1={f1:.4f} | ACC={acc:.4f}")
        print(confusion_matrix(y_true, pred))
        print(classification_report(y_true, pred, digits=4, zero_division=0))

    _report("Period p1", y, pred_p1)
    _report("Period p2", y, pred_p2)
    _report("Weighted",  y, pred_w)
    _report("Stacking",  y, pred_s)

    out = pd.DataFrame({
        "Date": D, "y_true": y,
        "proba_p1": proba1, "pred_p1": pred_p1,
        "proba_p2": proba2, "pred_p2": pred_p2,
        "proba_weighted": proba_w, "pred_weighted": pred_w,
        "proba_stacking": proba_s, "pred_stacking": pred_s,
    })
    fname = f"NVDA_forward_{D.min().date()}_to_{D.max().date()}.csv"
    out.to_csv(fname, index=False)
    print("\n🔎 Detalle forward:")
    print(out.to_string(index=False))
    print("\n💾 Guardado:", fname)

# =============================================
#  Main
# =============================================
if __name__ == "__main__":
    # Entrena y calibra hasta 2025-09-30 (con target EMA40+slope_10 + noticias en todo el histórico)
    summary = train_two_models_and_ensemble(ticker="NVDA", start_date="2000-01-01")
    # Prueba out-of-sample del primero de 1 octubre al 27 de noviembre (con noticias de noviembre y octubre)
    run_forward_eval_nov(ticker="NVDA")
```

```python
import pandas as pd
import numpy as np
import joblib

# --- 1) Cargar dataset base (el mismo que usaste para entrenar) ---
# Si ya tienes 'df_all' (o 'base') con ['Date','target', features_lag1], úsalo.
# Si no, reconstruye con tu función:
# df_all = build_dataset_all("NVDA", start_date="2000-01-01")
# Supondré que se llama df_all:
df_all = df_all.copy()
df_all["Date"] = pd.to_datetime(df_all["Date"], errors="coerce")
df_all = df_all.dropna(subset=["Date"]).sort_values("Date").reset_index(drop=True)

# --- 2) Helpers ---


def ensure_feature_alignment(df_feats: pd.DataFrame, feat_names: list[str]) -> pd.DataFrame:
    """
    Alinea columnas a las que espera el modelo y garantiza dtype numérico (float).
    Soporta dtypes 'Int64' (nullable) y convierte todo a float.
    """
    X = df_feats.copy()
    X = X.replace([np.inf, -np.inf], np.nan)

    # Asegurar que TODAS las columnas (excepto 'Date' si estuviera) sean numéricas
    for c in X.columns:
        if c == "Date":
            continue
        if not pd.api.types.is_numeric_dtype(X[c]):
            X[c] = pd.to_numeric(X[c], errors="coerce")

    # Castear a float (evita Int64/booleanos que rompen con ciertos estimadores)
    num_cols = [c for c in X.columns if c != "Date"]
    X[num_cols] = X[num_cols].astype(float)

    # Alinear al set exacto de features del modelo (faltantes = NaN → los imputa el pipeline)
    X = X.reindex(columns=feat_names)

    return X


def make_period_df(df_all, start, end, artifact):
    """Toma df_all, recorta por fechas y calcula proba con el calibrador guardado."""
    mask = (df_all["Date"] >= pd.to_datetime(start)) & (df_all["Date"] <= pd.to_datetime(end))
    d = df_all.loc[mask].reset_index(drop=True)
    # columnas de features = todas menos Date/target
    feat_cols_all = [c for c in d.columns if c not in ["Date","target"]]
    # las que el modelo realmente usó:
    model_feats = list(artifact["features"])
    X = ensure_feature_alignment(d[feat_cols_all], model_feats)
    proba = artifact["calibrator"].predict_proba(X)[:, 1]
    out = pd.DataFrame({
        "Date": d["Date"].values,
        "y_true": d["target"].astype(int).values,
        "proba_model": proba
    })
    return out

# --- 3) Cargar artefactos p1 y p2 ---
art_p1 = joblib.load("NVDA_2000_2019_model.joblib")  # contiene calibrator, features, threshold, period
art_p2 = joblib.load("NVDA_2020_2025_model.joblib")

thr_p1 = float(art_p1["threshold"])  # p.ej. 0.429
thr_p2 = float(art_p2["threshold"])  # p.ej. 0.500
p1_start, p1_end = art_p1["period"]  # ["2000-01-01","2019-12-31"]
p2_start, p2_end = art_p2["period"]  # ["2020-01-01","2025-09-30"]

# --- 4) Reconstruir probabilidades por período con el calibrador guardado ---
df_p1 = make_period_df(df_all, p1_start, p1_end, art_p1)
df_p2 = make_period_df(df_all, p2_start, p2_end, art_p2)


# ==========================================================
# 6) PERIOD-VOTE (modelo base) EN HOLDOUT GLOBAL (último 20%)
#     → matriz de confusión alineada con tus logs
# ==========================================================
import matplotlib.pyplot as plt  # por si no estaba importado

def _align_numeric(df, cols):
    """Alinea columnas a las esperadas por el calibrador y castea a float."""
    X = df.reindex(columns=cols)
    for c in X.columns:
        X[c] = pd.to_numeric(X[c], errors="coerce")
    return X.astype(float)

def period_vote_predictions(df_all, art_p1, art_p2):
    """
    Calcula y_pred (Period-Vote): p1 decide en 2000–2019 con su umbral,
    p2 decide en 2020–2025 con su umbral.
    Devuelve fechas, y_true y y_pred para TODO df_all.
    """
    dates = pd.to_datetime(df_all["Date"])
    y_true = df_all["target"].astype(int).values

    feats1, feats2 = list(art_p1["features"]), list(art_p2["features"])
    thr1, thr2 = float(art_p1["threshold"]), float(art_p2["threshold"])
    s1, e1 = map(pd.to_datetime, art_p1["period"])
    s2, e2 = map(pd.to_datetime, art_p2["period"])

    # Probabilidades calibradas por época
    X1 = _align_numeric(df_all, feats1)
    X2 = _align_numeric(df_all, feats2)
    proba1 = art_p1["calibrator"].predict_proba(X1)[:, 1]
    proba2 = art_p2["calibrator"].predict_proba(X2)[:, 1]

    # Period-Vote: cada época con su umbral
    y_pred = np.zeros(len(df_all), dtype=int)
    m1 = (dates >= s1) & (dates <= e1)
    m2 = (dates >= s2) & (dates <= e2)
    y_pred[m1] = (proba1[m1] >= thr1).astype(int)
    y_pred[m2] = (proba2[m2] >= thr2).astype(int)

    return dates, y_true, y_pred

# --- Ejecutar Period-Vote sobre TODO el set y cortar HOLDOUT global (último 20%) ---
dates_all, y_all, yhat_all = period_vote_predictions(df_all, art_p1, art_p2)
cut = int(len(df_all) * 0.8)  # mismo criterio que en los logs
y_ho, yhat_ho = y_all[cut:], yhat_all[cut:]

# Para reutilizar tu función 'plot_confusion_with_metrics' (que espera 'proba' + umbral),
# armamos un df con 'y_pred' 0/1 y usamos threshold=0.5 (equivale).
df_period_vote_ho = pd.DataFrame({
    "y_true": y_ho,
    "y_pred": yhat_ho
})

plot_confusion_with_metrics(
    df_period_vote_ho,
    proba_col="y_pred",          # tratamos 0/1 como "proba" binaria
    truth_col="y_true",
    threshold=0.5,               # 0/1 con corte 0.5 → mismas predicciones
    title=f"Matriz de confusión — Period-Vote (HOLDOUT global, N={len(df_period_vote_ho)})",
    normalize="true",
    outfile="cm_period_vote_holdout.png"
)



# --- 5) Graficar matrices de confusión para cada período ---
# Usa tu función ya creada:
# plot_confusion_with_metrics(df_fwd, proba_col="proba_model", truth_col="y_true", threshold=..., title=..., normalize=...)

plot_confusion_with_metrics(
    df_p1,
    proba_col="proba_model",
    truth_col="y_true",
    threshold=thr_p1,
    title="Matriz de confusión — Periodo 2000–2019 (model p1)",
    normalize="true",
    outfile="cm_p1_2000_2019.png"
)

plot_confusion_with_metrics(
    df_p2,
    proba_col="proba_model",
    truth_col="y_true",
    threshold=thr_p2,
    title="Matriz de confusión — Periodo 2020–2025 (model p2)",
    normalize="true",
    outfile="cm_p2_2020_2025.png"
)
```

```python

# Asumiendo que el archivo guardado tiene este nombre
FORWARD_RESULTS_FILE = "NVDA_forward_2025-10-01_to_2025-11-26.csv"

# =============================================
# Funciones auxiliares
# =============================================

def _best_f1_threshold(y_true: np.ndarray, y_proba: np.ndarray, steps: int = 100) -> float:
    """Busca el umbral que maximiza el F1-Score."""
    thresholds = np.linspace(0.01, 0.99, steps)
    f1s = [f1_score(y_true, (y_proba >= t).astype(int), zero_division=0) for t in thresholds]
    best_idx = np.argmax(f1s)
    return thresholds[best_idx]

# =============================================
# Análisis de errores y optimización de umbral
# =============================================

def analyze_forward_errors_and_optimize_threshold(file_path: str):
    """
    Carga los resultados del forward test, lista los errores y optimiza el umbral
    para la probabilidad de Stacking.
    """
    print(f"📊 Analizando el archivo: {file_path}")
    df = pd.read_csv(file_path)
    df["Date"] = pd.to_datetime(df["Date"])
    df.set_index("Date", inplace=True)
    y_true = df["y_true"].values

    print("\n--- 🔍 Análisis de Errores (Stacking / Period p1) ---")
    
    # Ambos modelos tienen la misma predicción y errores
    pred_stack = df["pred_stacking"]
    
    # Días de Falso Positivo (FP): Se predice 1 (Bull) pero es 0 (Bear/Lateral)
    fp_mask = (pred_stack == 1) & (df["y_true"] == 0)
    fp_dates = df[fp_mask].index.date
    print(f"\n❌ Días de Falso Positivo (FP = {len(fp_dates)}):")
    print(", ".join(map(str, fp_dates)))
    print(df[fp_mask][["y_true", "proba_stacking", "pred_stacking"]])

    # Días de Falso Negativo (FN): Se predice 0 (Bear/Lateral) pero es 1 (Bull)
    fn_mask = (pred_stack == 0) & (df["y_true"] == 1)
    fn_dates = df[fn_mask].index.date
    print(f"\n⚠️ Días de Falso Negativo (FN = {len(fn_dates)}):")
    print(", ".join(map(str, fn_dates)))
    print(df[fn_mask][["y_true", "proba_stacking", "pred_stacking"]])

    print("\n--- 💡 Optimización de Umbral para Stacking ---")
    proba_stack = df["proba_stacking"].values
    
    # Umbral entrenado vs. Umbral óptimo para este periodo
    thr_trained = 0.404 # Se obtiene del output anterior 'thr_meta=0.404'
    thr_optimal = _best_f1_threshold(y_true, proba_stack)
    
    f1_trained = f1_score(y_true, (proba_stack >= thr_trained).astype(int))
    f1_optimal = f1_score(y_true, (proba_stack >= thr_optimal).astype(int))
    
    print(f"Umbral entrenado (thr_meta): {thr_trained:.3f} → F1-Score: {f1_trained:.4f}")
    print(f"Umbral óptimo (en este OOS): {thr_optimal:.3f} → F1-Score: {f1_optimal:.4f}")
    
    if thr_optimal != thr_trained:
        pred_optimal = (proba_stack >= thr_optimal).astype(int)
        conf_matrix = confusion_matrix(y_true, pred_optimal)
        print("\nMatriz de Confusión con Umbral Óptimo:")
        print(conf_matrix)
        print(f"Mejora potencial en F1: {f1_optimal - f1_trained:.4f}")

# =============================================
# Ejecución
# =============================================
if __name__ == "__main__":
    try:
        analyze_forward_errors_and_optimize_threshold(FORWARD_RESULTS_FILE)
    except FileNotFoundError:
        print(f"Error: No se encontró el archivo {FORWARD_RESULTS_FILE}. Asegúrate de que el código anterior se ejecutó y guardó el archivo correctamente.")
```

```python
def build_df_fwd(forward_csv: str,
                 base: pd.DataFrame | None = None,
                 proba_preference=("proba_stacking","proba_weighted","proba_p2","proba_p1"),
                 allow_no_truth: bool = True) -> pd.DataFrame:
    import pandas as pd
    fwd = pd.read_csv(forward_csv)
    fwd["Date"] = pd.to_datetime(fwd["Date"], errors="coerce")

    # Elegir probabilidad disponible
    proba_col = next((c for c in proba_preference if c in fwd.columns), None)
    if proba_col is None:
        raise ValueError(f"No encuentro columnas de proba en {proba_preference}")

    df = fwd[["Date", proba_col]].rename(columns={proba_col: "proba_model"}).copy()

    # y_true desde el CSV si existe
    if "y_true" in fwd.columns:
        df["y_true"] = fwd["y_true"].astype("Int64")
    else:
        # intentar cruzar con base si fue pasada
        if base is not None:
            b = base[["Date","target"]].copy()
            b["Date"] = pd.to_datetime(b["Date"], errors="coerce")
            df = df.merge(b, on="Date", how="left").rename(columns={"target":"y_true"})
            df["y_true"] = df["y_true"].astype("Int64")
        elif not allow_no_truth:
            raise ValueError("No hay 'y_true' en CSV ni 'base[target]'. Pasa allow_no_truth=True o incluye y_true.")

    return df.dropna(subset=["Date"]).sort_values("Date").reset_index(drop=True)

```

```python


def build_df_fwd(forward_csv: str,
                 base: pd.DataFrame | None = None,
                 proba_preference=("proba_stacking","proba_weighted","proba_p2","proba_p1"),
                 allow_no_truth: bool = True) -> pd.DataFrame:
    """
    Devuelve un df con ['Date','proba_model'] y opcionalmente 'y_true' si está disponible.
    - Prioriza 'y_true' del CSV. Si no existe, intenta cruzar con base['target'].
    - Si allow_no_truth=True, sigue sin 'y_true' (para graficar probas).
    """
    fwd = pd.read_csv(forward_csv)
    fwd["Date"] = pd.to_datetime(fwd["Date"], errors="coerce")

    # Elegir probabilidad disponible
    proba_col = next((c for c in proba_preference if c in fwd.columns), None)
    if proba_col is None:
        raise ValueError(f"No encuentro columnas de proba en {proba_preference}")

    out_cols = ["Date", proba_col]
    df = fwd[out_cols].rename(columns={proba_col: "proba_model"}).copy()

    # y_true: del CSV si existe
    if "y_true" in fwd.columns:
        df["y_true"] = fwd["y_true"].astype("Int64")
    else:
        # intentar cruzar con base si se pasó
        if base is not None and "target" in base.columns:
            b = base[["Date","target"]].copy()
            b["Date"] = pd.to_datetime(b["Date"], errors="coerce")
            df = df.merge(b, on="Date", how="left")
            df = df.rename(columns={"target":"y_true"})
            df["y_true"] = df["y_true"].astype("Int64")
        elif not allow_no_truth:
            raise ValueError("No hay 'y_true' en CSV ni 'base[target]'. Pasa allow_no_truth=True o incluye y_true.")

    # NO tiramos filas por NaN de y_true: dejamos que el plot decida qué hacer
    df = df.dropna(subset=["Date"]).sort_values("Date").reset_index(drop=True)
    return df

def plot_proba_vs_truth(df_fwd: pd.DataFrame,
                        proba_col: str = "proba_model",
                        truth_col: str = "y_true",
                        threshold: float = 0.5,
                        title: str = "Probabilidad 'Bull mañana' vs etiqueta real (forward)",
                        smooth_window: int = 0):
    if "Date" not in df_fwd.columns or proba_col not in df_fwd.columns:
        raise ValueError("df_fwd debe contener 'Date' y la columna de probabilidad.")

    d = df_fwd.copy().sort_values("Date").reset_index(drop=True)
    d["Date"] = pd.to_datetime(d["Date"], errors="coerce")
    d = d.dropna(subset=["Date"])

    fig, ax = plt.subplots(figsize=(13, 5))
    ax.plot(d["Date"], d[proba_col], label="Prob. bull (modelo)", linewidth=1.6)

    if smooth_window and smooth_window > 1:
        smooth = d[proba_col].rolling(smooth_window, min_periods=1).mean()
        ax.plot(d["Date"], smooth, linestyle="--", linewidth=1, label=f"Media móvil {smooth_window}d")

    ax.axhline(threshold, linestyle="--", linewidth=1, label=f"Umbral = {threshold:.2f}")

    # Si hay verdad terreno, sombrear “bull real” y mostrar métricas
    has_truth = truth_col in d.columns and d[truth_col].notna().any()
    if has_truth:
        y = d[truth_col].fillna(-1).astype(int).values
        proba = d[proba_col].values
        yhat = (proba >= float(threshold)).astype(int)

        valid = y >= 0  # usa solo filas con y_true conocido
        yv, yhatv, probav = y[valid], yhat[valid], proba[valid]

        tp_idx = (yv == 1) & (yhatv == 1)
        fp_idx = (yv == 0) & (yhatv == 1)
        fn_idx = (yv == 1) & (yhatv == 0)

        def _safe_div(a,b): return (a/b) if b>0 else 0.0
        tp, fp, fn = int(tp_idx.sum()), int(fp_idx.sum()), int(fn_idx.sum())
        prec = _safe_div(tp, (tp + fp))
        rec  = _safe_div(tp, (tp + fn))
        f1   = _safe_div(2 * prec * rec, (prec + rec))

        # sombreo bull real en el eje completo (no solo “valid”)
        is_bull = (d[truth_col] == 1).fillna(False).values
        start = None
        first_bull = np.where(is_bull)[0]
        first_bull = first_bull[0] if len(first_bull) else None
        for i in range(len(d)):
            if is_bull[i] and start is None:
                start = i
            if (not is_bull[i] or i == len(d)-1) and start is not None:
                end = i if not is_bull[i] else i
                ax.axvspan(d["Date"].iloc[start], d["Date"].iloc[end],
                           alpha=0.08, label="Bull real" if i == first_bull else None)
                start = None

        # puntos (solo en válidos)
        ax.scatter(d.loc[valid].loc[tp_idx, "Date"], d.loc[valid].loc[tp_idx, proba_col],
                   marker="o", s=35, label="TP (acierto bull)")
        ax.scatter(d.loc[valid].loc[fp_idx, "Date"], d.loc[valid].loc[fp_idx, proba_col],
                   marker="x", s=45, label="FP (falso bull)")
        ax.scatter(d.loc[valid].loc[fn_idx, "Date"], d.loc[valid].loc[fn_idx, proba_col],
                   marker="v", s=45, label="FN (bull no detectado)")

        text_metrics = (f"Precisión: {prec:.2f}\n"
                        f"Recall:    {rec:.2f}\n"
                        f"F1:        {f1:.2f}\n"
                        f"N válida={valid.sum()}  TP={tp}  FP={fp}  FN={fn}")
        ax.text(0.995, 0.03, text_metrics, transform=ax.transAxes,
                ha="right", va="bottom", fontsize=9,
                bbox=dict(boxstyle="round,pad=0.35", alpha=0.15))
    else:
        ax.text(0.995, 0.03, "Sin etiquetas reales en este rango.\nSe muestra solo la probabilidad.",
                transform=ax.transAxes, ha="right", va="bottom", fontsize=9,
                bbox=dict(boxstyle="round,pad=0.35", alpha=0.15))

    ax.set_ylim(-0.05, 1.05)
    ax.set_title(title)
    ax.set_ylabel("Probabilidad de bull (t+1)")
    ax.xaxis.set_major_formatter(DateFormatter("%Y-%m-%d"))
    fig.autofmt_xdate()
    ax.legend(loc="upper left", ncols=2)
    ax.grid(alpha=0.25)
    plt.tight_layout()
    plt.show()
```

```python

from sklearn.inspection import permutation_importance
from sklearn.metrics import roc_auc_score, f1_score

# ========= Helpers =========
def _unique_cols(df: pd.DataFrame) -> pd.DataFrame:
    return df.loc[:, ~df.columns.duplicated()].copy()

def ensure_feature_alignment(X: pd.DataFrame, features: list[str]) -> pd.DataFrame:
    """Alinea X a la lista de features usada por el modelo (rellena faltantes con NaN)."""
    Xu = _unique_cols(X)
    feats = [c for c in features if c in Xu.columns]
    missing = [c for c in features if c not in Xu.columns]
    if missing:
        Xu = Xu.assign(**{c: np.nan for c in missing})
        feats = features[:]  # respeta el orden original del modelo
    return Xu.reindex(columns=feats)

def base_name(col: str) -> str:
    """Quita sufijos comunes (_lag1, _lag2, etc.) para agrupar por 'feature base'."""
    for suf in ["_lag1","_lag2","_lag3","_lag4","_lag5"]:
        if col.endswith(suf):
            return col[: -len(suf)]
    return col

def plot_barh_with_errorbars(imp_df: pd.DataFrame, title: str, top_n: int = 12):
    """
    imp_df: DataFrame con columnas ['feature','mean','std'].
    """
    imp_top = imp_df.sort_values("mean", ascending=False).head(top_n).iloc[::-1]
    fig, ax = plt.subplots(figsize=(9, max(4, 0.35*len(imp_top))))
    ax.barh(imp_top["feature"], imp_top["mean"], xerr=imp_top["std"], alpha=0.9)
    ax.set_title(title)
    ax.set_xlabel("Importancia (media de caída de score)")
    for i, (m, s) in enumerate(zip(imp_top["mean"], imp_top["std"])):
        ax.text(m + (0.01 * np.nanmax(imp_top["mean"].values)), i, f"{m:.3f}±{s:.3f}", va="center", fontsize=8)
    ax.grid(axis="x", alpha=0.25)
    plt.tight_layout()
    plt.show()

# ========= Scorers (AUC y F1@umbral) =========
def scorer_auc(estimator, X, y):
    p = estimator.predict_proba(X)[:, 1]
    # Si todas las clases son iguales, ROC AUC es indefinido; devolvemos 0.5 neutro
    if len(np.unique(y)) < 2:
        return 0.5
    return roc_auc_score(y, p)

def make_scorer_f1_at_threshold(thr: float):
    def scorer(estimator, X, y):
        p = estimator.predict_proba(X)[:, 1]
        pred = (p >= thr).astype(int)
        # si y es una sola clase, F1 no está definido; devolvemos 0
        if len(np.unique(y)) < 2:
            return 0.0
        return f1_score(y, pred)
    return scorer

# ========= Permutation Importance mejorado =========
def permutation_importance_df(estimator,
                              X_val: pd.DataFrame,
                              y_val: np.ndarray,
                              scoring,
                              n_repeats: int = 30,
                              random_state: int = 42) -> pd.DataFrame:
    """
    Devuelve DataFrame con ['feature','mean','std'] de permutation importance.
    - scoring: puede ser un string de sklearn ('roc_auc'), un callable (scorer_auc),
               o un wrapper como make_scorer_f1_at_threshold(thr).
    """
    result = permutation_importance(estimator, X_val, y_val,
                                    n_repeats=n_repeats,
                                    scoring=scoring,
                                    random_state=random_state,
                                    n_jobs=-1)
    df_imp = pd.DataFrame({
        "feature": X_val.columns,
        "mean": result.importances_mean,
        "std": result.importances_std
    })
    return df_imp.sort_values("mean", ascending=False).reset_index(drop=True)

def aggregate_by_base_feature(imp_df: pd.DataFrame) -> pd.DataFrame:
    """
    Colapsa por 'feature base' (sin sufijos de lag), tomando la importancia máxima del grupo.
    Devuelve ['feature','mean','std'] ya ordenado.
    """
    if imp_df.empty: 
        return imp_df
    tmp = imp_df.copy()
    tmp["base"] = tmp["feature"].apply(base_name)
    agg = (
        tmp.sort_values("mean", ascending=False)
           .groupby("base", as_index=False)
           .agg(mean=("mean","max"), std=("std","max"))
           .rename(columns={"base":"feature"})
           .sort_values("mean", ascending=False)
           .reset_index(drop=True)
    )
    return agg

# ========= Uso típico =========
# - calibrator: tu CalibratedClassifierCV guardado (artifact['calibrator'])
# - artifact['features']: lista de columnas usadas por el periodo al entrenar
# - X_val, y_val: conjunto de validación del periodo (ojo: MISMAS columnas)

def feature_importance_workflow(calibrator,
                                X_val: pd.DataFrame,
                                y_val: np.ndarray,
                                model_features: list[str],
                                thr_for_f1: float | None = None,
                                repeats: int = 30):
    """
    Calcula y grafica:
      1) Importancia por Permutación con AUC.
      2) Importancia por Permutación con F1@umbral (si thr_for_f1 no es None).
      3) Versiones agregadas por 'feature base' para cada métrica.
    """
    # Alinea columnas a las del modelo
    Xv = ensure_feature_alignment(X_val, model_features)

    # --- (1) AUC ---
    imp_auc = permutation_importance_df(calibrator, Xv, y_val, scoring=scorer_auc, n_repeats=repeats)
    plot_barh_with_errorbars(imp_auc, "Top variables por Permutation Importance (AUC)")
    imp_auc_grp = aggregate_by_base_feature(imp_auc)
    plot_barh_with_errorbars(imp_auc_grp, "Top *grupos* de variables (AUC, colapsando lags)")

    # --- (2) F1@umbral (si lo pides) ---
    if thr_for_f1 is not None:
        scorer_f1_thr = make_scorer_f1_at_threshold(thr_for_f1)
        imp_f1 = permutation_importance_df(calibrator, Xv, y_val, scoring=scorer_f1_thr, n_repeats=repeats)
        plot_barh_with_errorbars(imp_f1, f"Top variables por Permutation (F1@{thr_for_f1:.3f})")
        imp_f1_grp = aggregate_by_base_feature(imp_f1)
        plot_barh_with_errorbars(imp_f1_grp, f"Top *grupos* (F1@{thr_for_f1:.3f}, colapsando lags)")

    # Devuelve DataFrames por si quieres exportar
    out = {"AUC": imp_auc, "AUC_grouped": imp_auc_grp}
    if thr_for_f1 is not None:
        out["F1_thr"] = imp_f1
        out["F1_thr_grouped"] = imp_f1_grp
    return out
```

```python


def plot_proba_sentiment_news(
    df_fwd: pd.DataFrame,
    proba_col: str = "proba_stacking",
    pos_col: str = "pos_ewm3",
    news_col: str = "news_count",
    truth_col: str | None = "y_true",        # si quieres sombrear días bull reales
    threshold: float = 0.5,                   # umbral de decisión del modelo
    smooth_window: int = 3,                   # suavizado (rolling) de probas y pos
    lag_pos: int = 0,                         # desfase para pos_ewm3 (ej: 1 o 2 días)
    normalize_news: bool = True,              # escalar news_count (0-1) para compararlo visualmente
    title: str = "Probabilidad (modelo) vs Sentimiento y Volumen de noticias",
):
    need = {"Date", proba_col}
    assert need.issubset(df_fwd.columns), f"Faltan columnas mínimas: {need - set(df_fwd.columns)}"
    d = df_fwd.copy()
    d["Date"] = pd.to_datetime(d["Date"], errors="coerce")
    d = d.sort_values("Date").reset_index(drop=True)

    # Series base
    proba = d[proba_col].astype(float)
    proba_s = proba.rolling(smooth_window, min_periods=1).mean() if smooth_window > 1 else proba

    # Sentimiento positivo (opcional + lag + suavizado)
    has_pos = pos_col in d.columns
    if has_pos:
        pos = d[pos_col].astype(float)
        if lag_pos != 0:
            pos = pos.shift(lag_pos)
        pos_s = pos.rolling(smooth_window, min_periods=1).mean() if smooth_window > 1 else pos
    else:
        pos_s = None

    # Volumen de noticias (opcional + normalización a 0..1 para escala visual)
    has_news = news_col in d.columns
    if has_news:
        news = d[news_col].astype(float).fillna(0.0)
        if normalize_news and news.max() > 0:
            news_plot = (news - news.min()) / (news.max() - news.min())
        else:
            news_plot = news
    else:
        news_plot = None

    # Figura principal
    fig, ax = plt.subplots(figsize=(12, 5))

    # Probabilidades (+ versión suavizada)
    ax.plot(d["Date"], proba, label="Prob. bull (modelo)")
    if smooth_window > 1:
        ax.plot(d["Date"], proba_s, label=f"Prob. bull (rolling {smooth_window})", linewidth=2)

    # Umbral y zonas por encima/debajo
    ax.axhline(threshold, linestyle="--", linewidth=1, label=f"Umbral = {threshold:.2f}")
    # sombreado cuando proba_s >= threshold
    above = (proba_s >= threshold)
    # rellenar por tramos para no dibujar cada día
    if above.any():
        # encontrar segmentos contiguos
        idx = np.where(above.values)[0]
        if idx.size > 0:
            starts = [idx[0]]
            ends = []
            for i in range(1, len(idx)):
                if idx[i] != idx[i-1] + 1:
                    ends.append(idx[i-1])
                    starts.append(idx[i])
            ends.append(idx[-1])
            for s, e in zip(starts, ends):
                ax.axvspan(d["Date"].iloc[s], d["Date"].iloc[e], alpha=0.10)

    ax.set_ylabel("Probabilidad (0–1)")

    # Barras de noticias en el mismo eje (visual, escala 0–1 si normalize_news)
    if has_news:
        # barras semi-transparentes
        ax.bar(d["Date"], news_plot, width=1.0, alpha=0.2, label=("Noticias (norm.)" if normalize_news else "Noticias"))

    # Sentimiento en eje secundario
    if has_pos:
        ax2 = ax.twinx()
        ax2.plot(d["Date"], pos_s, label=("Sent. positivo" + (f" (lag {lag_pos})" if lag_pos else "")) , linewidth=1.5)
        ax2.set_ylabel("Sentimiento (0–1)")
        # unimos leyendas de ambos ejes
        lines1, labels1 = ax.get_legend_handles_labels()
        lines2, labels2 = ax2.get_legend_handles_labels()
        ax2.legend(lines1 + lines2, labels1 + labels2, loc="upper left")
    else:
        ax.legend(loc="upper left")

    ax.set_title(title)
    ax.grid(True, linestyle="--", alpha=0.4)
    plt.tight_layout()
    plt.show()

    # ======= Métricas rápidas y mini-panel de correlaciones =======
    # Correlación simple y rolling con sentimiento (si está)
    if has_pos:
        # Alineamos
        df_corr = pd.DataFrame({
            "proba": proba_s,
            "pos": pos_s
        }).dropna()
        if not df_corr.empty:
            corr_simple = df_corr["proba"].corr(df_corr["pos"])
            # Rolling correl (ventana 10 días)
            win = min(10, max(5, len(df_corr)//6))  # ventana prudente
            roll_corr = df_corr["proba"].rolling(win).corr(df_corr["pos"])

            # Scan de lag (pos vs proba) para ver en qué desfase correlaciona mejor
            max_lag = 5
            stats = []
            for L in range(-max_lag, max_lag+1):
                a = df_corr["pos"].shift(L)
                tmp = pd.concat([df_corr["proba"], a], axis=1).dropna()
                if len(tmp) > 5:
                    stats.append((L, tmp.iloc[:,0].corr(tmp.iloc[:,1])))
            best_lag, best_corr = max(stats, key=lambda x: x[1]) if stats else (0, np.nan)

            # Panel pequeño
            fig, ax = plt.subplots(figsize=(12, 3))
            ax.plot(df_corr.index, roll_corr, label=f"Correlación rolling (win={win})")
            ax.axhline(0, linestyle="--", linewidth=1)
            ax.set_title(f"Correlación proba ↔ sentimiento | simple={corr_simple:.2f} | mejor lag={best_lag} (corr={best_corr:.2f})")
            ax.legend(loc="upper left")
            ax.grid(True, linestyle="--", alpha=0.4)
            plt.tight_layout()
            plt.show()

    # ======= Si quieres sombrear días ‘bull’ reales =======
    if truth_col and truth_col in d.columns:
        fig, ax = plt.subplots(figsize=(12, 1.6))
        ax.set_title("Etiqueta real (bull=1, bear=0)")
        ax.plot(d["Date"], d[truth_col].astype(int), drawstyle="steps-pre")
        ax.set_ylim(-0.1, 1.1); ax.set_yticks([0,1])
        ax.grid(True, linestyle="--", alpha=0.4)
        plt.tight_layout()
        plt.show()
```

```python


def plot_proba_vs_truth_with_labels(
    df_fwd: pd.DataFrame,
    proba_col: str = "proba_stacking",
    truth_col: str = "y_true",
    threshold: float = 0.5,
    title: str = "Probabilidad 'Bull mañana' vs Realidad",
    smooth_window: int = 0
):
    # --- checks mínimos ---
    need = {"Date", proba_col, truth_col}
    if not need.issubset(df_fwd.columns):
        raise ValueError(f"Faltan columnas: {need - set(df_fwd.columns)}")

    d = df_fwd.copy().sort_values("Date")
    d["Date"] = pd.to_datetime(d["Date"])
    d = d.dropna(subset=[proba_col, truth_col])

    # suavizado opcional (rolling mean centrado)
    if smooth_window and smooth_window > 1:
        d["proba_smooth"] = d[proba_col].rolling(smooth_window, center=True, min_periods=1).mean()
        line_col = "proba_smooth"
        label_line = f"Prob. bull (rolling {smooth_window})"
    else:
        line_col = proba_col
        label_line = "Prob. bull (modelo)"

    # predicción binaria y tipos de acierto/error
    d["pred"] = (d[proba_col] >= threshold).astype(int)
    tp = (d["pred"] == 1) & (d[truth_col] == 1)
    tn = (d["pred"] == 0) & (d[truth_col] == 0)
    fp = (d["pred"] == 1) & (d[truth_col] == 0)  # falso positivo
    fn = (d["pred"] == 0) & (d[truth_col] == 1)  # falso negativo

    # --- figura: dos paneles ---
    fig, axes = plt.subplots(nrows=2, ncols=1, figsize=(12, 5.5), sharex=True,
                             gridspec_kw={"height_ratios":[3, 1]})

    ax = axes[0]
    ax.plot(d["Date"], d[line_col], label=label_line)
    ax.axhline(threshold, linestyle="--", linewidth=1, label=f"Umbral = {threshold:.2f}")

    # marcar FP/FN con símbolos distintos (para entender dónde falla)
    ax.scatter(d.loc[fp, "Date"], d.loc[fp, proba_col], marker="x", s=40, label="Falso positivo")
    ax.scatter(d.loc[fn, "Date"], d.loc[fn, proba_col], marker="o", facecolors="none", s=40, label="Falso negativo")

    # sombreado suave por zonas (bull/bear reales)
    # (ligero relleno para guiar la vista; si no te gusta, comenta estas 2 líneas)
    ax.fill_between(d["Date"], 0, 1, where=(d[truth_col] == 1), transform=ax.get_xaxis_transform(),
                    alpha=0.06, label="Zona bull real")

    ax.set_ylabel("Probabilidad")
    ax.set_title(title)
    ax.grid(True, linestyle=":", alpha=0.5)
    ax.legend(loc="upper right")

    # --- panel inferior: tira binaria bull/bear real + spans de rachas bull ---
    ax2 = axes[1]
    ax2.step(d["Date"], d[truth_col].astype(int), where="post")
    ax2.set_ylim(-0.15, 1.15)
    ax2.set_yticks([0, 1])
    ax2.set_yticklabels(["0", "1"])
    ax2.set_xlabel("Fecha")
    ax2.set_title("Etiqueta real (bull=1, bear=0)")
    ax2.grid(True, linestyle=":", alpha=0.4)

    # resaltar rachas bull continuas como spans
    y = d[truth_col].astype(int).to_numpy()
    dates = d["Date"].to_numpy()
    starts, ends = [], []
    in_bull = False
    for i in range(len(y)):
        if y[i] == 1 and not in_bull:
            in_bull = True
            starts.append(i)
        if in_bull and (i == len(y)-1 or y[i+1] == 0):
            in_bull = False
            ends.append(i)
    for s, e in zip(starts, ends):
        ax2.axvspan(dates[s], dates[e], alpha=0.15)

    plt.tight_layout()
    plt.show()



```

```python


def _wilson_interval(k, n, z=1.96):
    """IC binomial de Wilson (k éxitos sobre n). Devuelve (low, high)."""
    if n == 0:
        return (np.nan, np.nan)
    p = k / n
    denom = 1 + z**2 / n
    center = (p + z**2/(2*n)) / denom
    half = (z * np.sqrt((p*(1-p) + z**2/(4*n)) / n)) / denom
    return (max(0.0, center - half), min(1.0, center + half))

def plot_quarter_bull_rate(
    df: pd.DataFrame,
    date_col: str = "Date",
    label_col: str = "y_true",
    title: str = "Tasa histórica de días ‘Bull mañana’ por trimestre",
    period: tuple[str, str] | None = None  # ej. ("2000-01-01","2025-09-30")
):
    # ---- Validaciones mínimas
    need = {date_col, label_col}
    if not need.issubset(df.columns):
        raise ValueError(f"Faltan columnas: {need - set(df.columns)}")

    d = df[[date_col, label_col]].copy()
    d[date_col] = pd.to_datetime(d[date_col], errors="coerce")
    d = d.dropna(subset=[date_col, label_col])
    d[label_col] = d[label_col].astype(int)

    # ---- Slice temporal opcional
    if period is not None:
        start, end = pd.to_datetime(period[0]), pd.to_datetime(period[1])
        d = d[(d[date_col] >= start) & (d[date_col] <= end)].copy()

    # ---- Quarter
    d["quarter"] = d[date_col].dt.quarter.astype(int)
    grp = d.groupby("quarter", dropna=True)[label_col]
    rate = grp.mean()
    n = grp.size()
    k = grp.sum()

    # Orden Q1..Q4 (por si falta alguno)
    q_order = [1, 2, 3, 4]
    rate = rate.reindex(q_order, fill_value=np.nan)
    n = n.reindex(q_order, fill_value=0)
    k = k.reindex(q_order, fill_value=0)

    # ---- Intervalos de confianza (Wilson 95%)
    ci_low, ci_high = [], []
    for qi in q_order:
        lo, hi = _wilson_interval(k[qi], n[qi], z=1.96)
        ci_low.append(lo)
        ci_high.append(hi)

    # Altura y barras de error (asimétricas)
    heights = rate.values
    yerr = np.vstack([heights - np.array(ci_low), np.array(ci_high) - heights])

    # ---- Plot
    fig, ax = plt.subplots(figsize=(7, 4))
    ax.bar(np.arange(len(q_order)), heights, yerr=yerr, capsize=4)
    ax.set_xticks(np.arange(len(q_order)))
    ax.set_xticklabels([f"Q{q}" for q in q_order])
    ax.set_ylim(0, min(1.0, np.nanmax(ci_high) * 1.15 if np.isfinite(np.nanmax(ci_high)) else 1.0))
    ax.yaxis.set_major_formatter(PercentFormatter(1.0))
    ax.set_title(title)
    ax.set_xlabel("Trimestre")
    ax.set_ylabel("Proporción de días bull (t+1)")
    ax.grid(axis="y", linestyle=":", alpha=0.5)

    # Anotaciones (porcentaje y N)
    for i, qi in enumerate(q_order):
        if np.isnan(heights[i]):
            ax.text(i, 0.02, "sin datos", ha="center", va="bottom", fontsize=9)
            continue
        ax.text(i, heights[i] + (ax.get_ylim()[1]*0.02),
                f"{heights[i]*100:.1f}%\nN={n[qi]}",
                ha="center", va="bottom", fontsize=9)

    plt.tight_layout()
    plt.show()



```

```python

import matplotlib.pyplot as plt
from sklearn.metrics import (
    confusion_matrix, precision_recall_fscore_support,
    accuracy_score
)

def plot_confusion_with_metrics(
    df_fwd: pd.DataFrame,
    proba_col: str = "proba_model",
    truth_col: str = "y_true",
    threshold: float = 0.5,
    title: str = "Matriz de confusión (forward)",
    normalize: str | None = None,   # None | "true" | "pred" | "all" (solo para la imagen de fondo)
    labels: tuple[str,str] = ("Bear (0)","Bull (1)"),
    outfile: str | None = None
):
    """
    df_fwd: DataFrame con columnas [truth_col, proba_col]
    threshold: umbral para convertir probas en clase (usa tu calibrado)
    normalize: normalización del heatmap (sklearn): None/"true"/"pred"/"all"
    """
    need = {proba_col, truth_col}
    missing = need - set(df_fwd.columns)
    if missing:
        raise ValueError(f"Faltan columnas: {missing}")

    y_true = df_fwd[truth_col].astype(int).to_numpy()
    y_prob = df_fwd[proba_col].astype(float).to_numpy()
    y_pred = (y_prob >= float(threshold)).astype(int)

    # Matrices: cruda + (opcional) normalizada para el fondo
    cm_raw = confusion_matrix(y_true, y_pred, labels=[0,1])
    cm_norm = (confusion_matrix(y_true, y_pred, labels=[0,1], normalize=normalize)
               if normalize else None)

    # Métricas
    acc = accuracy_score(y_true, y_pred)
    prec, rec, f1, sup = precision_recall_fscore_support(
        y_true, y_pred, labels=[0,1], zero_division=0
    )
    macro_p = prec.mean(); macro_r = rec.mean(); macro_f1 = f1.mean()

    # Figura
    fig, ax = plt.subplots(figsize=(6.5, 5.2))

    # Fondo (heatmap simple con imshow); no dependemos de seaborn
    data_to_show = cm_norm if normalize else cm_raw
    im = ax.imshow(data_to_show, aspect="auto")

    # Ejes y ticks
    ax.set_title(f"{title}\n(thr = {threshold:.3f})", pad=10)
    ax.set_xlabel("Predicción")
    ax.set_ylabel("Real")
    ax.set_xticks([0,1]); ax.set_yticks([0,1])
    ax.set_xticklabels(labels); ax.set_yticklabels(labels)

    # Anotaciones por celda: conteo + porcentaje por fila (recall local)
    totals_row = cm_raw.sum(axis=1).clip(min=1)  # evitar /0
    for i in range(2):
        for j in range(2):
            count = cm_raw[i, j]
            pct_row = 100.0 * count / totals_row[i]
            txt = f"{count:d}\n{pct_row:.1f}%"
            ax.text(j, i, txt,
                    ha="center", va="center", fontsize=11, fontweight="bold")

    # Cuadro de métricas al costado
    right_txt = (
        f"Accuracy: {acc:.3f}\n"
        f"Macro P/R/F1: {macro_p:.3f} / {macro_r:.3f} / {macro_f1:.3f}\n"
        f"\nClase 0 (Bear)\n"
        f"  P: {prec[0]:.3f}  R: {rec[0]:.3f}  F1: {f1[0]:.3f}  (n={sup[0]})\n"
        f"Clase 1 (Bull)\n"
        f"  P: {prec[1]:.3f}  R: {rec[1]:.3f}  F1: {f1[1]:.3f}  (n={sup[1]})"
    )
    # caja de texto dentro del eje (a la derecha)
    ax.text(2.15, 0.0, right_txt, transform=ax.transAxes,
            fontsize=10, va="top")

    plt.tight_layout()

    if outfile:
        plt.savefig(outfile, dpi=150, bbox_inches="tight")
        print("💾 Guardado:", outfile)
    plt.show()


```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def bull_precision(df: pd.DataFrame, proba_col: str, truth_col: str, threshold: float) -> float:
    """
    Devuelve la precisión (PPV) de la clase bull (1) para un DF dado:
      precisión_bull = TP / (TP + FP)
    Si no hubo predicciones positivas, devuelve np.nan.
    """
    need = {proba_col, truth_col}
    missing = need - set(df.columns)
    if missing:
        raise ValueError(f"Faltan columnas en el DF: {missing}")
    y_true = df[truth_col].astype(int).to_numpy()
    y_pred = (df[proba_col].astype(float).to_numpy() >= float(threshold)).astype(int)

    pred_pos = (y_pred == 1)
    if pred_pos.sum() == 0:
        return float("nan")
    tp = ((y_pred == 1) & (y_true == 1)).sum()
    fp = ((y_pred == 1) & (y_true == 0)).sum()
    return tp / (tp + fp)

def plot_bull_precision_bars(
    base: pd.DataFrame,
    df_fwd: pd.DataFrame,
    proba_col_base: str = "proba_stacking",
    proba_col_fwd: str  = "proba_stacking",
    truth_col: str      = "y_true",
    thr_base: float     = 0.404,       # tu umbral calibrado p/ stacking; ajusta si usas otro modelo
    thr_fwd: float      = 0.404,
    train_split_1=("2000-01-01", "2019-12-31"),
    train_split_2=("2020-01-01", "2025-09-30"),
    title: str = "Precisión de señales BULL por tramo (TP / (TP+FP))"
):
    # --- checks y fechas
    for col in [proba_col_base, truth_col, "Date"]:
        if col not in base.columns:
            raise ValueError(f"'{col}' no está en base.columns")
    for col in [proba_col_fwd, truth_col, "Date"]:
        if col not in df_fwd.columns:
            raise ValueError(f"'{col}' no está en df_fwd.columns")

    b = base.copy()
    b["Date"] = pd.to_datetime(b["Date"])
    f = df_fwd.copy()
    f["Date"] = pd.to_datetime(f["Date"])

    # --- cortes de entrenamiento
    s1, e1 = pd.to_datetime(train_split_1[0]), pd.to_datetime(train_split_1[1])
    s2, e2 = pd.to_datetime(train_split_2[0]), pd.to_datetime(train_split_2[1])

    base_1 = b[(b["Date"] >= s1) & (b["Date"] <= e1)]
    base_2 = b[(b["Date"] >= s2) & (b["Date"] <= e2)]

    # --- precisiones
    p1 = bull_precision(base_1, proba_col_base, truth_col, thr_base)
    p2 = bull_precision(base_2, proba_col_base, truth_col, thr_base)
    p3 = bull_precision(f,       proba_col_fwd,  truth_col, thr_fwd)

    vals = [p1, p2, p3]
    labels = ["2000–2019", "2020–2025", "Forward (eval)"]

    # --- gráfico
    fig, ax = plt.subplots(figsize=(7,4))
    ax.bar(np.arange(len(vals)), vals)
    ax.set_xticks(np.arange(len(vals))); ax.set_xticklabels(labels, rotation=0)
    ax.set_ylim(0, 1)  # porcentaje 0–100%
    ax.set_title(title)
    ax.set_ylabel("Precisión Bull")

    # anotar valores arriba de cada barra
    for i, v in enumerate(vals):
        txt = "—" if (v is np.nan) or (v != v) else f"{100*v:.1f}%"
        ax.text(i, 0.02 + (0 if (v is np.nan or v != v) else v), txt,
                ha="center", va="bottom", fontsize=10)

    plt.tight_layout(); plt.show()


```

```python
import pandas as pd
import numpy as np
import joblib
from sklearn.metrics import precision_score
import matplotlib.pyplot as plt

# ========= 0) Carga artefactos =========
art_p1 = joblib.load("NVDA_2000_2019_model.joblib")
art_p2 = joblib.load("NVDA_2020_2025_model.joblib")
stack_art = joblib.load("NVDA_stack.joblib")     # {'meta_clf','threshold'}
thr_meta = float(stack_art["threshold"])
meta_clf = stack_art["meta_clf"]

with open("NVDA_ensemble.json","r",encoding="utf-8") as f:
    import json
    ens_cfg = json.load(f)
thr_weighted = float(ens_cfg["thresholds"]["weighted"])
w1, w2 = ens_cfg["weights"]["p1"], ens_cfg["weights"]["p2"]

def _weighted_average(p1, p2, w1, w2):
    s = (w1+w2) if (w1+w2)>0 else 1.0
    return (w1*p1 + w2*p2)/s

def _predict_proba_calibrated(df_features: pd.DataFrame, artifact: dict) -> np.ndarray:
    feats = pd.Index(artifact["features"]).unique()
    X = df_features.loc[:, ~df_features.columns.duplicated()]
    # Si faltan features, se rellenan con NaN (el imputer del calibrador se encarga)
    for c in feats:
        if c not in X.columns:
            X[c] = np.nan
    X = X.reindex(columns=feats)
    return artifact["calibrator"].predict_proba(X)[:, 1]

# ========= 1) Asegurar etiquetas en 'base' =========
def ensure_base_with_labels(base: pd.DataFrame) -> pd.DataFrame:
    if "y_true" in base.columns:
        return base.copy()

    if "target" in base.columns:  # ya viene del dataset de entrenamiento
        out = base.copy()
        out["y_true"] = out["target"].astype(int)
        return out

    # Reconstruir con la regla de entrenamiento: bull_t = (close > EMA40) & (slope10 > 0); y_true = bull_t.shift(-1)
    need_date_close = {"Date", "close"}
    if not need_date_close.issubset(base.columns):
        raise ValueError(f"Para reconstruir y_true necesito {need_date_close}, y me faltan: {need_date_close - set(base.columns)}")

    out = base.copy()
    out["Date"] = pd.to_datetime(out["Date"], errors="coerce")
    out = out.sort_values("Date").reset_index(drop=True)

    # Calcular EMA40 si no existe
    if "EMA40" not in out.columns:
        out["EMA40"] = out["close"].ewm(span=40, adjust=False).mean()

    # Calcular slope10 si no existe (diferencia de 10 días)
    if "slope10" not in out.columns:
        out["slope10"] = out["close"].diff(10)

    bull_t = ((out["close"] > out["EMA40"]) & (out["slope10"] > 0)).astype(int)
    out["y_true"] = bull_t.shift(-1).fillna(0).astype(int)  # última fila queda NaN → 0 por seguridad
    return out

# ========= 2) Añadir probabilidades del modelo a 'base' =========
def augment_base_with_probas(base: pd.DataFrame, truth_col="y_true") -> pd.DataFrame:
    # Features para los calibradores: usa todas las columnas menos Date/target/y_true
    feat_cols = [c for c in base.columns if c not in ["Date", "target", truth_col]]
    X = base[feat_cols].copy()

    p1 = _predict_proba_calibrated(X, art_p1)
    p2 = _predict_proba_calibrated(X, art_p2)
    pw = _weighted_average(p1, p2, w1, w2)
    ps = meta_clf.predict_proba(np.c_[p1, p2])[:, 1]

    out = base.copy()
    out["proba_p1"] = p1
    out["proba_p2"] = p2
    out["proba_weighted"] = pw
    out["proba_stacking"] = ps
    return out

# ========= 3) Construir df_fwd a partir del CSV forward =========
def build_df_fwd(forward_csv: str) -> pd.DataFrame:
    d = pd.read_csv(forward_csv, encoding="utf-8")
    d["Date"] = pd.to_datetime(d["Date"], errors="coerce")
    # Si el CSV forward no trae y_true, intentamos reconstruirlo como antes (requiere close en algún lado).
    if "y_true" not in d.columns:
        if {"close"}.issubset(d.columns):
            d = ensure_base_with_labels(d)
        elif "target" in d.columns:
            d["y_true"] = d["target"].astype(int)
        else:
            # como último recurso: marca 0 (no ideal, pero evita romper)
            d["y_true"] = 0
    # Si no trae proba_stacking pero sí proba_weighted, usa esa columna
    if "proba_stacking" not in d.columns:
        if "proba_weighted" in d.columns:
            d["proba_stacking"] = d["proba_weighted"]
        elif {"proba_p1","proba_p2"}.issubset(d.columns):
            d["proba_stacking"] = meta_clf.predict_proba(d[["proba_p1","proba_p2"]].to_numpy())[:,1]
        else:
            # Si no hay ninguna proba, no podemos graficar métricas de precisión
            pass
    return d.sort_values("Date").reset_index(drop=True)

# ========= 4) Gráfica de precisión BULL por tramo =========
def plot_bull_precision_bars(base_aug: pd.DataFrame,
                             df_fwd: pd.DataFrame,
                             proba_col_base: str = "proba_stacking",
                             proba_col_fwd: str  = "proba_stacking",
                             truth_col: str = "y_true",
                             thr_base: float = None,
                             thr_fwd: float  = None,
                             train_split_1=("2000-01-01","2019-12-31"),
                             train_split_2=("2020-01-01","2025-09-30"),
                             title="Precisión de señales 'Bull' por tramo"):
    d1s, d1e = pd.to_datetime(train_split_1[0]), pd.to_datetime(train_split_1[1])
    d2s, d2e = pd.to_datetime(train_split_2[0]), pd.to_datetime(train_split_2[1])

    if thr_base is None: thr_base = thr_meta
    if thr_fwd  is None: thr_fwd  = thr_meta

    b = base_aug.copy()
    b["Date"] = pd.to_datetime(b["Date"])
    m1 = (b["Date"] >= d1s) & (b["Date"] <= d1e)
    m2 = (b["Date"] >= d2s) & (b["Date"] <= d2e)

    def _bull_precision(df, proba_col, thr, truth_col):
        if proba_col not in df.columns:
            return np.nan
        pred_bull = (df[proba_col] >= thr).astype(int)
        if pred_bull.sum() == 0:
            return 0.0
        return float((df[truth_col][pred_bull==1] == 1).mean())

    p_2000_2019 = _bull_precision(b.loc[m1], proba_col_base, thr_base, truth_col)
    p_2020_2025 = _bull_precision(b.loc[m2], proba_col_base, thr_base, truth_col)
    p_forward   = _bull_precision(df_fwd,     proba_col_fwd,  thr_fwd,  truth_col)

    labels = ["2000–2019", "2020–2025", "Forward"]
    vals   = [p_2000_2019, p_2020_2025, p_forward]

    fig, ax = plt.subplots(figsize=(6,4))
    bars = ax.bar(labels, np.array(vals)*100)
    ax.set_ylim(0, 100)
    ax.set_ylabel("% de BULL correctos")
    ax.set_title(title)
    for r, v in zip(bars, vals):
        ax.text(r.get_x()+r.get_width()/2, r.get_height()+1, f"{v*100:.1f}%", ha="center", va="bottom", fontsize=10)
    plt.tight_layout(); plt.show()

# ========= 5) USO =========
# (a) Asegura y_true en base
base = ensure_base_with_labels(base)

# (b) Añade probabilidades
base_aug = augment_base_with_probas(base, truth_col="y_true")

# (c) Carga el forward (tu CSV de octubre–noviembre)
df_fwd = build_df_fwd("NVDA_forward_2025-10-01_to_2025-11-26.csv")

# (d) Grafica
plot_bull_precision_bars(base_aug, df_fwd,
                         proba_col_base="proba_stacking",
                         proba_col_fwd="proba_stacking",
                         truth_col="y_true",
                         thr_base=thr_meta,
                         thr_fwd=thr_meta,
                         train_split_1=("2000-01-01","2019-12-31"),
                         train_split_2=("2020-01-01","2025-09-30"),
                         title="Precisión de señales 'Bull' — 2000–2019 vs 2020–2025 vs Forward")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def _segment_mask(d, start, end):
    s, e = pd.to_datetime(start), pd.to_datetime(end)
    return (pd.to_datetime(d["Date"]) >= s) & (pd.to_datetime(d["Date"]) <= e)

def _bull_precision_and_counts(df, proba_col, thr, truth_col="y_true"):
    """Devuelve (precision, n_pred_bull, n_hits)."""
    if proba_col not in df.columns or truth_col not in df.columns:
        return np.nan, 0, 0
    pred_bull = (df[proba_col] >= thr).astype(int)
    n_pred = int(pred_bull.sum())
    if n_pred == 0:
        return 0.0, 0, 0
    hits = int((df.loc[pred_bull==1, truth_col] == 1).sum())
    prec = hits / n_pred
    return float(prec), n_pred, hits

def plot_bull_precision_and_counts(
    base_aug: pd.DataFrame,
    df_fwd: pd.DataFrame,
    proba_col_base="proba_stacking",
    proba_col_fwd="proba_stacking",
    thr_base=0.5,
    thr_fwd=0.5,
    truth_col="y_true",
    split1=("2000-01-01","2019-12-31"),
    split2=("2020-01-01","2025-09-30"),
    title="Señales ‘Bull’: Precisión (%) y Aciertos (n)"
):
    # Segmentos
    m1 = _segment_mask(base_aug, *split1)
    m2 = _segment_mask(base_aug, *split2)

    # Métricas por tramo
    p1, n1, h1 = _bull_precision_and_counts(base_aug.loc[m1], proba_col_base, thr_base, truth_col)
    p2, n2, h2 = _bull_precision_and_counts(base_aug.loc[m2], proba_col_base, thr_base, truth_col)
    p3, n3, h3 = _bull_precision_and_counts(df_fwd,            proba_col_fwd,  thr_fwd,  truth_col)

    labels = ["2000–2019", "2020–2025", "Forward"]
    precisions = np.array([p1, p2, p3]) * 100.0   # %
    hits = np.array([h1, h2, h3], dtype=float)    # cantidad

    x = np.arange(len(labels))
    width = 0.38

    fig, ax1 = plt.subplots(figsize=(7.5,4.5))

    # Barras: precisión (izquierda)
    b1 = ax1.bar(x - width/2, precisions, width, label="% BULL correctos")
    ax1.set_ylabel("% de BULL correctos")
    ax1.set_ylim(0, max(100, np.nanmax(precisions)*1.15))

    # Segundo eje: aciertos (derecha)
    ax2 = ax1.twinx()
    b2 = ax2.bar(x + width/2, hits, width, label="Aciertos BULL (n)")
    ax2.set_ylabel("Aciertos (n)")
    ax2.set_ylim(0, max(1, np.nanmax(hits)*1.25))

    # Título y eje X
    ax1.set_title(title)
    ax1.set_xticks(x, labels)

    # Etiquetas sobre cada barra
    for rect, val in zip(b1, precisions):
        ax1.annotate(f"{val:.1f}%",
                     xy=(rect.get_x()+rect.get_width()/2, rect.get_height()),
                     xytext=(0,3), textcoords="offset points",
                     ha="center", va="bottom", fontsize=9)
    for rect, val in zip(b2, hits):
        ax2.annotate(f"{int(val)}",
                     xy=(rect.get_x()+rect.get_width()/2, rect.get_height()),
                     xytext=(0,3), textcoords="offset points",
                     ha="center", va="bottom", fontsize=9)

    # Leyenda combinada
    h1_, l1_ = ax1.get_legend_handles_labels()
    h2_, l2_ = ax2.get_legend_handles_labels()
    ax1.legend(h1_+h2_, l1_+l2_, loc="upper left")

    plt.tight_layout()
    plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# --- z-score robusto (MAD) ---
def _robust_z(s: pd.Series) -> pd.Series:
    med = np.nanmedian(s)
    mad = np.nanmedian(np.abs(s - med))
    if mad == 0 or np.isnan(mad):
        return pd.Series(np.zeros(len(s)), index=s.index, dtype=float)
    return (s - med) / (1.4826 * mad)

def detect_outliers_price_volume(df: pd.DataFrame,
                                 date_col="Date",
                                 price_col="close",
                                 vol_col="volume",
                                 z_thresh=3.5,
                                 log_volume=True):
    """
    Devuelve df_out con columnas:
      Date, close, volume, z_price, z_vol, out_price, out_vol, out_any
    """
    d = df[[date_col, price_col, vol_col]].copy()
    d[date_col] = pd.to_datetime(d[date_col], errors="coerce")
    d = d.dropna(subset=[date_col])

    # z robusto para precio
    d["z_price"] = _robust_z(d[price_col].astype(float))

    # volumen (opcional log para estabilizar)
    if log_volume:
        vol = np.log1p(d[vol_col].astype(float))
    else:
        vol = d[vol_col].astype(float)
    d["z_vol"] = _robust_z(vol)

    d["out_price"] = d["z_price"].abs() > z_thresh
    d["out_vol"]   = d["z_vol"].abs()   > z_thresh
    d["out_any"]   = d["out_price"] | d["out_vol"]
    return d

def evaluate_on_outliers(d_merged: pd.DataFrame,
                         proba_col="proba_stacking",
                         truth_col="y_true",
                         thr=0.5):
    """
    Espera un DF que ya contenga columnas:
      Date, close, volume, out_price/out_vol/out_any, proba_col, truth_col
    Devuelve una tabla solo de outliers con:
      pred (0/1), correct (bool)
    """
    need = {"Date", "out_any", proba_col, truth_col}
    missing = need - set(d_merged.columns)
    if missing:
        raise ValueError(f"Faltan columnas: {missing}")

    out = d_merged[d_merged["out_any"]].copy()
    if out.empty:
        print("No hay outliers con el umbral actual.")
        return out

    out["pred"] = (out[proba_col] >= thr).astype(int)
    out["correct"] = (out["pred"] == out[truth_col].astype(int))
    return out.sort_values("Date").reset_index(drop=True)
```

```python
# 1) Detectar outliers
det = detect_outliers_price_volume(base_aug, date_col="Date", price_col="close", vol_col="volume", z_thresh=3.5)

# 2) Unir con probabilidades y etiqueta
#    (para histórico: base_aug ya tiene y_true y proba_*; para forward: df_fwd)
cols_preds = ["Date", "y_true", "proba_stacking"]  # cambia proba_* si querés
merged = det.merge(base_aug[cols_preds], on="Date", how="left")

# 3) Evaluar en outliers (umbral calibrado meta que viste en tus logs)
out_tab = evaluate_on_outliers(merged, proba_col="proba_stacking", truth_col="y_true", thr=0.404)

print("Resumen outliers:")
print(out_tab[["Date","close","volume","out_price","out_vol","proba_stacking","y_true","pred","correct"]].head(12))

# Métricas rápidas en outliers
if not out_tab.empty:
    prec = out_tab["correct"].mean()
    n = len(out_tab)
    n_bull = (out_tab["pred"]==1).sum()
    n_bull_ok = ((out_tab["pred"]==1) & (out_tab["correct"])).sum()
    print(f"\nOutliers totales: {n} | Señales bull en outliers: {n_bull} | Bull correctas: {n_bull_ok} | Precisión (global outliers): {prec:.1%}")
```

```python
def plot_outliers_with_hits(df_det_eval: pd.DataFrame,
                            date_col="Date", price_col="close",
                            proba_col="proba_stacking", thr=0.5,
                            title="Precio + outliers (precio/volumen) y aciertos del modelo"):
    """
    df_det_eval: lo que sale de mergear detect_outliers_price_volume(...) con proba + y_true
    Pone:
      • línea de precio
      • marca outliers (rombos)
      • contorno/marcado distinto si el modelo acertó (✓) o erró (✗)
    """
    need = {date_col, price_col, "out_any", "y_true", proba_col}
    if not need.issubset(df_det_eval.columns):
        raise ValueError(f"Faltan columnas: {need - set(df_det_eval.columns)}")

    d = df_det_eval.copy()
    d[date_col] = pd.to_datetime(d[date_col], errors="coerce")
    d = d.dropna(subset=[date_col, price_col])

    d["pred"] = (d[proba_col] >= thr).astype(int)
    d["correct"] = (d["pred"] == d["y_true"].astype(int))

    fig, ax = plt.subplots(figsize=(12,4))
    ax.plot(d[date_col], d[price_col], label="Precio")

    # puntos outlier
    out = d[d["out_any"]]
    ax.scatter(out[date_col], out[price_col], marker="D", s=40, label="Outlier (precio/vol)")

    # aciertos/errores sobre outliers (señal del modelo = bull/no-bull)
    ok  = out[out["correct"]]
    bad = out[~out["correct"]]
    # Para distinguir visualmente:
    ax.scatter(ok[date_col],  ok[price_col],  marker="o", s=60, linewidth=1, facecolors="none", label="✓ Acierto (en outliers)")
    ax.scatter(bad[date_col], bad[price_col], marker="x", s=60, linewidth=1, label="✗ Error (en outliers)")

    ax.set_title(title)
    ax.set_ylabel("Precio")
    ax.legend()
    plt.tight_layout(); plt.show()

    # Gráfico de barras de aciertos/errores en outliers
    if not out.empty:
        hits = int(ok.shape[0]); misses = int(bad.shape[0])
        fig, ax = plt.subplots(figsize=(5,3))
        ax.bar(["Aciertos", "Errores"], [hits, misses])
        ax.set_title("Modelo en días OUTLIER")
        for i, v in enumerate([hits, misses]):
            ax.text(i, v, str(v), ha="center", va="bottom")
        plt.tight_layout(); plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

def plot_outlier_hits_stacked(out_tab: pd.DataFrame, title="Modelo en días OUTLIER"):
    assert {"correct"}.issubset(out_tab.columns), "Falta columna 'correct' en out_tab"
    total = len(out_tab)
    hits  = int(out_tab["correct"].sum())
    miss  = int((~out_tab["correct"]).sum())
    fig, ax = plt.subplots(figsize=(6,4))
    ax.bar(["Outliers"], [hits], label="Aciertos")
    ax.bar(["Outliers"], [miss], bottom=[hits], label="Errores")
    ax.set_title(title)
    ax.set_ylabel("Conteo")
    for y, txt in zip([hits, hits+miss], [str(hits), str(miss)]):
        ax.text(0, y, txt, ha="center", va="bottom")
    prec = hits / total if total else 0.0
    ax.set_ylim(0, max(hits+miss, 1))
    ax.legend()
    plt.tight_layout(); plt.show()
    print(f"Aciertos: {hits}  |  Errores: {miss}  |  Precisión: {prec:.1%}")

# uso:
plot_outlier_hits_stacked(out_tab)
```

```python
def plot_outlier_hits_by_type(out_merged: pd.DataFrame, title="Outliers por tipo — aciertos/errores"):
    need = {"out_price","out_vol","correct"}
    assert need.issubset(out_merged.columns), f"Faltan columnas: {need - set(out_merged.columns)}"

    d = out_merged.copy()
    d["tipo"] = np.select(
        [ d["out_price"] & ~d["out_vol"],
          ~d["out_price"] & d["out_vol"],
          d["out_price"] & d["out_vol"] ],
        ["Precio", "Volumen", "Ambos"],
        default="N/A"
    )
    grp = d.groupby("tipo")["correct"].agg(
        aciertos="sum",
        total="count"
    ).reset_index()
    grp["errores"] = grp["total"] - grp["aciertos"]
    grp = grp.sort_values("total", ascending=False)

    # Barras apiladas por categoría
    fig, ax = plt.subplots(figsize=(7,4))
    ax.bar(grp["tipo"], grp["aciertos"], label="Aciertos")
    ax.bar(grp["tipo"], grp["errores"], bottom=grp["aciertos"], label="Errores")
    ax.set_title(title); ax.set_ylabel("Conteo")
    for i, (a,t) in enumerate(zip(grp["aciertos"], grp["total"])):
        ax.text(i, a, str(int(a)), ha="center", va="bottom")
        ax.text(i, t, str(int(t-a)), ha="center", va="bottom")
    ax.legend()
    plt.tight_layout(); plt.show()

    # Barras de precisión (%) por categoría
    grp["precision"] = grp["aciertos"] / grp["total"]
    fig, ax = plt.subplots(figsize=(7,3))
    ax.bar(grp["tipo"], grp["precision"]*100)
    ax.set_title("Precisión por tipo de outlier")
    ax.set_ylabel("% aciertos")
    ax.set_ylim(0, 100)
    for i, p in enumerate(grp["precision"]):
        ax.text(i, p*100, f"{p*100:.1f}%", ha="center", va="bottom")
    plt.tight_layout(); plt.show()

# uso:
plot_outlier_hits_by_type(out_tab)
```

```python
def plot_outlier_precision_single(out_tab: pd.DataFrame, title="% de BULL correctos en OUTLIERS"):
    hits = int(out_tab["correct"].sum())
    total = int(len(out_tab))
    prec = hits / total if total else 0.0
    fig, ax = plt.subplots(figsize=(4,4))
    ax.bar(["Precisión"], [prec*100])
    ax.set_ylim(0, 100)
    ax.set_ylabel("%")
    ax.set_title(title)
    ax.text(0, prec*100, f"{prec*100:.1f}%", ha="center", va="bottom")
    plt.tight_layout(); plt.show()

# uso:
plot_outlier_precision_single(out_tab)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Boxplot del precio (close)
# -----------------------------
def price_boxplot(df: pd.DataFrame,
                  price_col: str = "close",
                  period: tuple[str,str] | None = None,
                  logscale: bool = False,
                  title: str | None = None):
    need = {"Date", price_col}
    if not need.issubset(df.columns):
        raise ValueError(f"Faltan columnas: {need - set(df.columns)}")

    d = df[["Date", price_col]].dropna().copy()
    d["Date"] = pd.to_datetime(d["Date"], errors="coerce")
    if period:
        start, end = pd.to_datetime(period[0]), pd.to_datetime(period[1])
        d = d[(d["Date"] >= start) & (d["Date"] <= end)]

    x = d[price_col].astype(float)
    label = price_col
    if logscale:
        x = np.log(x.replace(0, np.nan)).dropna()
        label = f"log({price_col})"

    fig, ax = plt.subplots(figsize=(6,5))
    ax.boxplot(x, vert=True, showmeans=True)
    ax.set_title(title or f"Boxplot — {label}")
    ax.set_ylabel(label)
    ax.grid(True, linestyle="--", alpha=0.3)
    plt.tight_layout()
    plt.show()

# ----------------------------------------
# Resumen estadístico del precio (close)
#  (arreglado: sin etiquetas duplicadas)
# ----------------------------------------
def price_describe(df: pd.DataFrame,
                   price_col: str = "close",
                   period: tuple[str,str] | None = None) -> pd.DataFrame:
    need = {"Date", price_col}
    if not need.issubset(df.columns):
        raise ValueError(f"Faltan columnas: {need - set(df.columns)}")

    d = df[["Date", price_col]].dropna().copy()
    d["Date"] = pd.to_datetime(d["Date"], errors="coerce")
    if period:
        start, end = pd.to_datetime(period[0]), pd.to_datetime(period[1])
        d = d[(d["Date"] >= start) & (d["Date"] <= end)]

    s = d[price_col].astype(float)

    # describe con percentiles (incluye 25%, 50%, 75% por defecto)
    desc = s.describe(percentiles=[0.01, 0.05, 0.95, 0.99]).to_frame(name="valor")

    # métricas extra (evitamos duplicar 1%,5%,95%,99%)
    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    iqr = q3 - q1
    low_whisker  = q1 - 1.5*iqr
    high_whisker = q3 + 1.5*iqr
    outliers_mask = (s < low_whisker) | (s > high_whisker)

    extra = pd.Series({
        "rango"               : s.max() - s.min(),
        "IQR (Q3-Q1)"         : iqr,
        "coef. variación"     : (s.std() / s.mean()) if s.mean() != 0 else np.nan,
        "asimetría (skew)"    : s.skew(),
        "curtosis (exceso)"   : s.kurt(),
        "Q1 (25%)"            : q1,
        "Q3 (75%)"            : q3,
        "bigote inf (Q1-1.5·IQR)" : low_whisker,
        "bigote sup (Q3+1.5·IQR)" : high_whisker,
        "# outliers (IQR)"    : int(outliers_mask.sum()),
        "% outliers (IQR)"    : 100.0 * outliers_mask.mean(),
        "n"                   : int(s.count())
    }, name="valor").to_frame()

    # combinamos y quitamos posibles duplicados por seguridad
    out = pd.concat([desc, extra])
    out = out[~out.index.duplicated(keep="first")]

    order = [
        "count","n","mean","std","min","1%","5%","25%","Q1 (25%)","50%","75%","Q3 (75%)","95%","99%","max",
        "IQR (Q3-Q1)","rango","coef. variación","asimetría (skew)","curtosis (exceso)",
        "bigote inf (Q1-1.5·IQR)","bigote sup (Q3+1.5·IQR)","# outliers (IQR)","% outliers (IQR)"
    ]
    out = out.reindex([idx for idx in order if idx in out.index])
    return out
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# =========================
# 1) Cálculo de rendimientos
# =========================
def add_returns(df: pd.DataFrame,
                price_col: str = "close",
                ret_col: str = "ret_1d",
                periods: int = 1,
                method: str = "simple"):
    """
    Agrega al DF una columna de rendimientos diarios:
      - method="simple":  r_t = P_t / P_{t-1} - 1
      - method="log":     r_t = ln(P_t / P_{t-1})
    """
    if price_col not in df.columns:
        raise ValueError(f"'{price_col}' no está en df.columns")

    d = df.copy()
    p = d[price_col].astype(float)
    if method == "log":
        r = np.log(p / p.shift(periods))
    else:
        r = p.pct_change(periods)
    d[ret_col] = r
    return d

# ======================================
# 2) Boxplot + describe de rendimientos
# ======================================
def returns_boxplot(df: pd.DataFrame,
                    ret_col: str = "ret_1d",
                    period: tuple[str,str] | None = None,
                    title: str | None = None):
    need = {"Date", ret_col}
    if not need.issubset(df.columns):
        raise ValueError(f"Faltan columnas: {need - set(df.columns)}")

    d = df[["Date", ret_col]].dropna().copy()
    d["Date"] = pd.to_datetime(d["Date"], errors="coerce")
    if period:
        start, end = pd.to_datetime(period[0]), pd.to_datetime(period[1])
        d = d[(d["Date"] >= start) & (d["Date"] <= end)]

    x = d[ret_col].astype(float)

    fig, ax = plt.subplots(figsize=(6,5))
    ax.boxplot(x, vert=True, showmeans=True)
    ax.set_title(title or f"Boxplot — {ret_col}")
    ax.set_ylabel("Rendimiento diario")
    ax.grid(True, linestyle="--", alpha=0.3)
    plt.tight_layout()
    plt.show()

def returns_describe(df: pd.DataFrame,
                     ret_col: str = "ret_1d",
                     period: tuple[str,str] | None = None) -> pd.DataFrame:
    need = {"Date", ret_col}
    if not need.issubset(df.columns):
        raise ValueError(f"Faltan columnas: {need - set(df.columns)}")

    d = df[["Date", ret_col]].dropna().copy()
    d["Date"] = pd.to_datetime(d["Date"], errors="coerce")
    if period:
        start, end = pd.to_datetime(period[0]), pd.to_datetime(period[1])
        d = d[(d["Date"] >= start) & (d["Date"] <= end)]

    s = d[ret_col].astype(float)

    desc = s.describe(percentiles=[0.01, 0.05, 0.95, 0.99]).to_frame(name="valor")

    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    iqr = q3 - q1
    low_whisker  = q1 - 1.5*iqr
    high_whisker = q3 + 1.5*iqr
    outliers = (s < low_whisker) | (s > high_whisker)

    extra = pd.Series({
        "Q1 (25%)"                 : q1,
        "Q3 (75%)"                 : q3,
        "IQR (Q3-Q1)"              : iqr,
        "bigote inf (Q1-1.5·IQR)"  : low_whisker,
        "bigote sup (Q3+1.5·IQR)"  : high_whisker,
        "asimetría (skew)"         : s.skew(),
        "curtosis (exceso)"        : s.kurt(),
        "# outliers (IQR)"         : int(outliers.sum()),
        "% outliers (IQR)"         : 100.0 * outliers.mean(),
        "n"                        : int(s.count())
    }, name="valor").to_frame()

    out = pd.concat([desc, extra])
    out = out[~out.index.duplicated(keep="first")]

    order = [
        "count","n","mean","std","min","1%","5%","25%","Q1 (25%)","50%","75%","Q3 (75%)","95%","99%","max",
        "IQR (Q3-Q1)","asimetría (skew)","curtosis (exceso)",
        "bigote inf (Q1-1.5·IQR)","bigote sup (Q3+1.5·IQR)",
        "# outliers (IQR)","% outliers (IQR)"
    ]
    out = out.reindex([idx for idx in order if idx in out.index])
    return out

# ================================================================
# 3) Outliers en rendimientos + desempeño del modelo en esos días
# ================================================================
def summarize_return_outliers_with_model(df: pd.DataFrame,
                                         ret_col: str = "ret_1d",
                                         proba_col: str = "proba_stacking",
                                         truth_col: str = "y_true",
                                         threshold: float = 0.5,
                                         iqr_k: float = 1.5,
                                         period: tuple[str,str] | None = None):
    """
    Devuelve (summary_df, metrics_dict) y genera 2 gráficos:
      A) Barras: aciertos/errores en outliers vs no-outliers
      B) Barras: precisión Bull en outliers vs no-outliers
    """
    # columnas y fallback para y_true
    need = {"Date", ret_col, proba_col}
    if truth_col not in df.columns:
        if "target" in df.columns:
            df = df.copy()
            df[truth_col] = df["target"].astype(int)
        else:
            raise ValueError(f"No encuentro '{truth_col}' ni 'target' en df.")
    if not need.issubset(df.columns):
        raise ValueError(f"Faltan columnas: {need - set(df.columns)}")

    d = df[["Date", ret_col, proba_col, truth_col]].dropna().copy()
    d["Date"] = pd.to_datetime(d["Date"], errors="coerce")
    if period:
        start, end = pd.to_datetime(period[0]), pd.to_datetime(period[1])
        d = d[(d["Date"] >= start) & (d["Date"] <= end)]

    # IQR en retornos
    s = d[ret_col].astype(float)
    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    iqr = q3 - q1
    low, high = q1 - iqr_k*iqr, q3 + iqr_k*iqr
    d["out_ret"] = (s < low) | (s > high)

    # predicción bull según umbral
    d["pred"] = (d[proba_col] >= float(threshold)).astype(int)
    d["correct"] = (d["pred"] == d[truth_col])

    # resumen fila a fila (útil para exportar)
    summary = d[["Date", ret_col, "out_ret", proba_col, truth_col, "pred", "correct"]].copy()

    # métricas agregadas
    def _prec_bull(sub):
        # precisión de señales bull (positivos predichos)
        mask_pred1 = sub["pred"] == 1
        if not mask_pred1.any():
            return np.nan
        return float((sub.loc[mask_pred1, truth_col] == 1).mean())

    tot_out   = int(d["out_ret"].sum())
    tot_in    = int((~d["out_ret"]).sum())
    bull_out  = int((d["out_ret"] & (d["pred"]==1)).sum())
    bull_in   = int((~d["out_ret"] & (d["pred"]==1)).sum())
    corr_out  = int((d["out_ret"] & d["correct"]).sum())
    corr_in   = int((~d["out_ret"] & d["correct"]).sum())
    prec_out  = _prec_bull(d[d["out_ret"]])
    prec_in   = _prec_bull(d[~d["out_ret"]])

    metrics = {
        "outliers": tot_out, "inliers": tot_in,
        "bull_signals_out": bull_out, "bull_signals_in": bull_in,
        "correct_out": corr_out, "correct_in": corr_in,
        "bull_precision_out": prec_out, "bull_precision_in": prec_in
    }

    # --- Gráfico A: aciertos/errores por grupo ---
    fig, ax = plt.subplots(figsize=(7,4))
    cats = ["Outliers", "No outliers"]
    corrects = [corr_out, corr_in]
    wrongs   = [tot_out - corr_out, tot_in - corr_in]
    x = np.arange(len(cats))
    width = 0.4
    ax.bar(x - width/2, corrects, width, label="Aciertos")
    ax.bar(x + width/2, wrongs,   width, label="Errores")
    ax.set_xticks(x); ax.set_xticklabels(cats)
    ax.set_title("Desempeño del modelo en días con outliers (rendimientos)")
    ax.set_ylabel("Cantidad de días")
    ax.legend()
    ax.grid(True, axis="y", linestyle="--", alpha=0.3)
    for i,(c,w) in enumerate(zip(corrects, wrongs)):
        ax.text(i - width/2, c, f"{c}", ha="center", va="bottom", fontsize=9)
        ax.text(i + width/2, w, f"{w}", ha="center", va="bottom", fontsize=9)
    plt.tight_layout(); plt.show()

    # --- Gráfico B: precisión de señales bull ---
    fig, ax = plt.subplots(figsize=(7,4))
    vals = [prec_out*100 if not np.isnan(prec_out) else 0.0,
            prec_in*100  if not np.isnan(prec_in)  else 0.0]
    bars = ax.bar(["Outliers", "No outliers"], vals)
    ax.set_ylim(0, 100)
    ax.set_ylabel("% de señales BULL correctas")
    ax.set_title("Precisión de señales BULL — Outliers vs No outliers")
    ax.grid(True, axis="y", linestyle="--", alpha=0.3)
    for b,v in zip(bars, vals):
        ax.text(b.get_x()+b.get_width()/2, v, f"{v:.1f}%", ha="center", va="bottom")
    plt.tight_layout(); plt.show()

    return summary, metrics
```

```python
import json, joblib
import numpy as np
import pandas as pd

# ---- Cargar artefactos guardados ----
art_p1 = joblib.load("NVDA_2000_2019_model.joblib")  # {'calibrator','features','threshold',...}
art_p2 = joblib.load("NVDA_2020_2025_model.joblib")
stack_art = joblib.load("NVDA_stack.joblib")         # {'meta_clf','threshold'}
with open("NVDA_ensemble.json","r",encoding="utf-8") as f:
    ens_cfg = json.load(f)

meta_clf = stack_art["meta_clf"]
w1, w2 = ens_cfg["weights"]["p1"], ens_cfg["weights"]["p2"]

def _unique_cols(df): 
    return df.loc[:, ~df.columns.duplicated()].copy()

def _predict_proba_calibrated(df_features: pd.DataFrame, artifact: dict) -> np.ndarray:
    feats = pd.Index(artifact["features"]).unique()
    X = _unique_cols(df_features)
    # Si faltan columnas, se crean con NaN y el pipeline/imputer del calibrador las maneja
    X = X.reindex(columns=feats, fill_value=np.nan)
    return artifact["calibrator"].predict_proba(X)[:, 1]

def _weighted_average(p1, p2, w1, w2):
    s = (w1+w2) if (w1+w2)>0 else 1.0
    return (w1*p1 + w2*p2)/s

def augment_with_probas(df: pd.DataFrame) -> pd.DataFrame:
    """
    Devuelve df + columnas:
      proba_p1, proba_p2, proba_weighted, proba_stacking
    No toca tus etiquetas (y_true/target).
    """
    if "Date" in df.columns:
        d = df.copy()
    else:
        raise ValueError("Se requiere columna 'Date'.")

    # Usamos TODAS las columnas de features disponibles, excepto Date/y/target
    feat_cols = [c for c in d.columns if c not in ["Date","target","y_true"]]
    X_all = d[feat_cols].copy()

    # Probabilidades calibradas por período (usan sus propias listas de features)
    proba1 = _predict_proba_calibrated(X_all, art_p1)
    proba2 = _predict_proba_calibrated(X_all, art_p2)

    # Ensemble ponderado por F1-CV de cada época (lo que guardaste en ens_cfg)
    proba_w = _weighted_average(proba1, proba2, w1, w2)

    # Stacking: meta usa [p1, p2]
    X_meta = np.c_[proba1, proba2]
    proba_s = meta_clf.predict_proba(X_meta)[:,1]

    d["proba_p1"] = proba1
    d["proba_p2"] = proba2
    d["proba_weighted"] = proba_w
    d["proba_stacking"] = proba_s
    return d

# === APLICAR SOBRE TU DF ===
# Si ya tenés base_ret (base + ret_1d), úsalo; si no, usa base y luego calcula ret_1d.
base_aug = augment_with_probas(base_ret)  # <- añade proba_stacking y compañía
```

```python
import re
import joblib
import pandas as pd
import matplotlib.pyplot as plt

# ---------- Clasificador por familias ----------
def classify_features(cols):
    cats = {
        "Técnicas": [],
        "Mercado": [],
        "Económicas (FRED)": [],
        "Noticias (FinBERT)": [],
        "Fundamentales (FMP)": [],
        "Calendario": [],
        "Tasas (rf)": [],
        "Otros": [],
    }
    pat_tecnicas = re.compile(r"^(close|open|high|low|volume|adjClose|ema\d+|ema\d+_above_ema\d+|slope_\d+|roc_\d+|vol_\d+|ewma_vol|ADX_14|ATR_14|TR|PVT|beta_30d|r_nvda|excess_nvda)(_|$)", re.I)
    pat_mercado  = re.compile(r"^(sp500_close|sp_slope_\d+|r_mkt|excess_mkt)(_|$)", re.I)
    pat_fred     = re.compile(r"^(CPIAUCSL|UNRATE|GDPC1|PCE|INDPRO|PAYEMS|GNPCA|M2SL|HOUST)(_|$)", re.I)
    pat_news     = re.compile(r"^(pos(_ewm3)?|neg(_ewm3)?|neu(_ewm3)?|news_count|news_pos_interact)(_|$)", re.I)
    pat_fmp      = re.compile(r"^(keymetrics_|income_|balance_|cashflow_|.*_growth|marketcap_)", re.I)
    pat_calendar = re.compile(r"^(quarter|q_[1-4]|q_sin|q_cos)(_|$)", re.I)
    pat_tasas    = re.compile(r"^(rf)(_|$)", re.I)

    for c in cols:
        base = c.replace("_lag1","").replace("_lag2","")
        if pat_tecnicas.match(base):
            cats["Técnicas"].append(c)
        elif pat_mercado.match(base):
            cats["Mercado"].append(c)
        elif pat_fred.match(base):
            cats["Económicas (FRED)"].append(c)
        elif pat_news.match(base):
            cats["Noticias (FinBERT)"].append(c)
        elif pat_fmp.match(base):
            cats["Fundamentales (FMP)"].append(c)
        elif pat_calendar.match(base):
            cats["Calendario"].append(c)
        elif pat_tasas.match(base):
            cats["Tasas (rf)"].append(c)
        else:
            cats["Otros"].append(c)
    return cats

def counts_table(cats_dict):
    rows = [{"familia": k, "n_features": len(v)} for k,v in cats_dict.items()]
    df = pd.DataFrame(rows).sort_values("n_features", ascending=False).reset_index(drop=True)
    return df

def topN_plus_others(counts: pd.Series, top_n: int = 6) -> pd.Series:
    counts = counts.sort_values(ascending=False)
    if len(counts) <= top_n:
        return counts
    top = counts.iloc[:top_n]
    other_sum = counts.iloc[top_n:].sum()
    if other_sum > 0:
        top.loc["Otros (resto)"] = other_sum
    return top

# ---------- Cargar features desde artefactos ----------
art_p1 = joblib.load("NVDA_2000_2019_model.joblib")   # contiene 'features'
art_p2 = joblib.load("NVDA_2020_2025_model.joblib")
feats_p1 = list(pd.Index(art_p1["features"]).unique())
feats_p2 = list(pd.Index(art_p2["features"]).unique())

# (Opcional) Si ya tienes listas:
# feats_p1 = [...]
# feats_p2 = [...]

# ---------- Clasificar y contar ----------
cats_p1 = classify_features(feats_p1)
cats_p2 = classify_features(feats_p2)
tbl_p1 = counts_table(cats_p1)
tbl_p2 = counts_table(cats_p2)

print("=== Recuento P1 (2000–2019) ===")
print(tbl_p1.to_string(index=False))
print("\n=== Recuento P2 (2020–2025) ===")
print(tbl_p2.to_string(index=False))

# ---------- Gráficos de torta: P1 y P2 ----------
counts_p1 = tbl_p1.set_index("familia")["n_features"]
counts_p2 = tbl_p2.set_index("familia")["n_features"]

# Limitar a top 6 y agrupar el resto en "Otros (resto)" para que se lea mejor
counts_p1_plot = topN_plus_others(counts_p1, top_n=6)
counts_p2_plot = topN_plus_others(counts_p2, top_n=6)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes = axes.ravel()

axes[0].pie(
    counts_p1_plot.values,
    labels=counts_p1_plot.index,
    autopct=lambda pct: f"{pct:.1f}%\n({int(round(pct/100*counts_p1_plot.sum()))})",
    startangle=90
)
axes[0].axis('equal')
axes[0].set_title("Participación por familia — P1 (2000–2019)")

axes[1].pie(
    counts_p2_plot.values,
    labels=counts_p2_plot.index,
    autopct=lambda pct: f"{pct:.1f}%\n({int(round(pct/100*counts_p2_plot.sum()))})",
    startangle=90
)
axes[1].axis('equal')
axes[1].set_title("Participación por familia — P2 (2020–2025)")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def plot_bull_zones_window(df_window: pd.DataFrame, title_suffix=""):
    """Mismo panel que ya tenés, pero aplicado a un sub‐rango."""
    d = df_window.dropna(subset=["close","EMA40","slope10"]).copy()
    if d.empty:
        return
    is_bull = (d["close"] > d["EMA40"]) & (d["slope10"] > 0)

    fig, axes = plt.subplots(2, 1, figsize=(14, 6),
                             sharex=True, gridspec_kw={'height_ratios':[3,1]})
    ax0, ax1 = axes

    ax0.plot(d["Date"], d["close"], label="Cierre (NVDA)")
    ax0.plot(d["Date"], d["EMA40"], label="EMA40", linestyle="--")
    if "EMA20" in d.columns:
        ax0.plot(d["Date"], d["EMA20"], label="EMA20", linestyle=":")
    # Relleno de zonas bull
    for i in range(len(d)-1):
        if is_bull.iloc[i]:
            ax0.axvspan(d["Date"].iloc[i], d["Date"].iloc[i+1], alpha=0.12)

    ax0.set_title(f"NVDA — Precio, EMA40 y Zonas Bull {title_suffix}")
    ax0.set_ylabel("USD")
    ax0.legend(loc="upper left")
    ax0.grid(alpha=0.35)

    # ADX (lag1 si lo tenés)
    adx_col = "ADX_14_lag1" if "ADX_14_lag1" in d.columns else ("ADX_14" if "ADX_14" in d.columns else None)
    if adx_col is not None:
        ax1.plot(d["Date"], d[adx_col], label="ADX_14 (lag1)" if adx_col.endswith("lag1") else "ADX_14")
        ax1.axhline(25, linestyle=":", label="Umbral 25")
        ax1.legend(loc="upper left")
    ax1.set_xlabel("Fecha"); ax1.set_ylabel("ADX")
    ax1.grid(alpha=0.3)

    plt.tight_layout()
    plt.show()

def plot_bull_zones_by_5y_windows(base_df: pd.DataFrame, years_per_win: int = 5, save_prefix: str | None = None):
    """
    Corta el histórico en ventanas de 'years_per_win' años (por defecto 5) y
    dibuja un panel por ventana. Opcionalmente guarda PNGs.
    """
    d = base_df.copy()
    d["Date"] = pd.to_datetime(d["Date"], errors="coerce")
    d = d.dropna(subset=["Date"]).sort_values("Date")
    start_year = d["Date"].dt.year.min()
    end_year   = d["Date"].dt.year.max()

    # Construye bordes [Y, Y+5), [Y+5, Y+10)...
    edges = list(range(start_year, end_year + 1, years_per_win))
    for y0 in edges:
        y1 = min(y0 + years_per_win - 1, end_year)
        mask = (d["Date"] >= pd.Timestamp(f"{y0}-01-01")) & (d["Date"] <= pd.Timestamp(f"{y1}-12-31"))
        df_win = d.loc[mask].reset_index(drop=True)
        if df_win.empty:
            continue
        title_suffix = f"({y0}–{y1})"
        plot_bull_zones_window(df_win, title_suffix=title_suffix)

        if save_prefix:
            fname = f"{save_prefix}_{y0}_{y1}.png"
            plt.savefig(fname, dpi=140, bbox_inches="tight")

# === Usar con tu 'base' ya armado ===
plot_bull_zones_by_5y_windows(base, years_per_win=5)
```
